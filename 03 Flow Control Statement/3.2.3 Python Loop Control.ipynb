## Loop control Statements:

### break statement
- With the 'break' statement we can stop the loop even if the decision statement is satisfied.
- If the execution reaches a break statement, it immediately exits the loop's clause.
- In code, a break statement simply contains the break keyword.
<img src='./Image/3.2 Image f1.jpg' width=30% height=10%/>

```python
# Exit the while loop when i is 4:
i = 1
while i < 6:
    if i == 4:
        break
    print(i, end= ' ')
    i += 1
```

```python
# Exit the for loop when i is 4:
for i in range(1,7):
    if i == 4:
        break
    print(i, end= ' ')
```

### continue statement
- With the 'continue' statement we can skip the current itertion of loop even if the decision statement is satisfied.
- When the program execution reaches a continue statement, the program execution immediately jumps back to the start of the loop and re-evaluates the loop’s condition.
<img src='./Image/3.2 Image g1.jpg' width=30% height=10%/>

```python
# Continue to the next iteration inside while loop if i is 4:
i = 0
while i < 6:
    i += 1
    if i == 4:
        continue
    print(i,end=' ')    
```

```python
# Continue to the next iteration inside for loop if i is 4:
for i in range(1,7):
    if i == 4:
        continue
    print(i, end= ' ')
```

### `pass` vs `continue` vs `...`

These are easy to confuse and do quite different things.

| Statement | Effect |
|---|---|
| `pass` | **Nothing.** Execution carries straight on through the rest of the body. |
| `continue` | **Skip the rest of this iteration** and go to the next one. |
| `...` (`Ellipsis`) | Also nothing — conventionally marks "to be implemented". |

`pass` exists because Python needs *something* where a block is syntactically required.
It is a placeholder, not a loop-control statement — despite living in this notebook.

```python
# pass: a null statement. Execution CONTINUES through the rest of the body.
print("pass:")
for i in range(4):
    if i == 2:
        pass                      # does nothing at all
    print(f"  {i}")               # still runs for i == 2

# continue: skip the REST of this iteration
print("\ncontinue:")
for i in range(4):
    if i == 2:
        continue
    print(f"  {i}")               # never runs for i == 2

# Where `pass` genuinely belongs: a syntactic placeholder
class NotImplementedYet:
    pass

def todo_later():
    pass

# `...` (Ellipsis) is increasingly preferred for "to be filled in",
# because `pass` also legitimately means "deliberately do nothing".
def stub():
    ...

print("\nboth return None:", todo_later(), stub())

# continue as a guard clause - keeps the main body unindented
records = [
    {"name": "Aditya", "score": 88},
    {"name": "", "score": 91},
    {"name": "Rahul", "score": None},
    {"name": "Priya", "score": 95},
]

print("\nvalid records:")
for r in records:
    if not r["name"]:
        continue
    if r["score"] is None:
        continue
    print(f"  {r['name']}: {r['score']}")
```

### ⚠️ `break` only leaves the innermost loop

Python has no labelled `break`. In nested loops, `break` exits **one** level — the loop it
sits directly inside.

There are three standard ways to leave both:

| Approach | When to use |
|---|---|
| **Flag variable** | Quick, but adds state to track |
| **Extract a function and `return`** | Usually the cleanest — `return` leaves everything |
| **`for...else` + `continue`/`break`** | No extra state, but takes a moment to read |

```python
grid = [[1, 5, 3], [7, 2, 8], [4, 6, 9]]
target = 10

# ⚠️ break only exits the INNER loop
print("Naive - break only leaves the inner loop:")
for row in grid:
    for value in row:
        if value == target:
            break            # leaves the inner for, outer keeps going
    print("  finished row", row)


# --- Option 1: a flag variable ---
print("\nOption 1 - flag:")
found = None
for i, row in enumerate(grid):
    for j, value in enumerate(row):
        if value + row[0] == target and j != 0:
            found = (i, j)
            break
    if found:                # re-check at the outer level
        break
print("  found at", found)


# --- Option 2: a function and `return` (usually the cleanest) ---
print("\nOption 2 - function + return:")

def find_value(grid, target):
    for i, row in enumerate(grid):
        for j, value in enumerate(row):
            if value == target:
                return i, j          # returns straight out of BOTH loops
    return None

print("  find 8 ->", find_value(grid, 8))
print("  find 99 ->", find_value(grid, 99))


# --- Option 3: for...else on the outer loop ---
print("\nOption 3 - for/else:")
for row in grid:
    for value in row:
        if value == 8:
            print("  found 8 in", row)
            break
    else:
        continue          # inner loop finished without break -> next row
    break                 # only reached if the inner loop DID break
else:
    print("  never found")
```

### WAP for Signup and login
- During signup store data only if
    - username is unique
    - length of password is greater than equal to 5.
- to restrict user to enter password upto 3 times while login

```python
import sys
acc_detail={'aditya':'123456'}
print('1. Signup','2. Login','3. Exit',sep='\n')
ch= int(input("Enter choice no.: "))
if ch==1:
    print('---------SIGN UP-----------')
    while True:
        name= input("Choose Username: ")
        if name not in acc_detail.keys():
            pwd= input("Choose Password: ")
            if len(pwd)>6:
                acc_detail[name]=pwd
                print("Account Created")
                print(acc_detail)
                break
            else:
                print("Read Guideline")
        else:
            print("Try different Username")
elif ch==2:
    print('----------LOGIN----------')
    name=input("Enter Username: ")
    if name in acc_detail.keys():
        attempt=1
        while attempt<4:
            pwd=input("Enter password: ")
            if pwd==acc_detail[name]:
                print(f'Welcome,{name}!!')
                break
            else:
                print(f"Wrong Password,{3-attempt} attempts left")
            attempt +=1
    else:
        print("Unknown User")
elif ch==3:
    sys.exit()
else:
    print("Invalid Choice")
```

### WAP to guess the correct no. between 1 to 20

```python
from random import randint
from sys import exit
secret_num=randint(1,20)
print("Think of a no. between 1 to 20.")
attempt=0
while True:
    guess= int(input("Guess the no.: "))
    if guess==0: 
        print('You loose!!')
        exit()
    elif guess==secret_num:
        print(f"Congrats you guessed correctly in {attempt} attempts.")
        break
    elif guess<secret_num:
        print("Thinking too low!!")
    elif guess>secret_num:
        print("Thinking too high!!")
    attempt+=1
```

## Common Mistakes & Pitfalls

1. **Expecting `break` to exit all nested loops.** It leaves only the *innermost* one. Use a flag, a function with `return`, or an `else` on the outer loop.
2. **Using `pass` where you meant `continue`.** `pass` does nothing and execution carries on through the rest of the body; `continue` jumps to the next iteration.
3. **Putting `continue` before the counter increment in a `while` loop.** The counter never advances and the loop hangs. This does not happen with `for`.
4. **Forgetting that loop `else` means 'no break'.** It runs when the loop completes normally.
5. **Using `break` inside a `try`/`finally` and expecting `finally` to be skipped.** It still runs.
6. **Over-using `continue` as a filter.** Two or three guards is fine; ten means the loop body wanted to be a function or a comprehension.

## Best Practices

- Use `break` for genuine early exit; use `for...else` to report 'never found'.
- Use `continue` as a guard clause to keep the main body unindented.
- Prefer extracting a nested loop into a function and using `return` over flag variables.
- Use `pass` only as a real placeholder; prefer `...` (Ellipsis) in stubs you intend to fill.
- Keep loop bodies short — if you need `break` and `continue` and a flag, refactor.
- Never write a `while True` without a reachable `break` or an iteration bound.
