## Python Loops
- While programming, you will encounter many situations when you need to execute block of code repeatedly. So instead of writing those statements again and again we use to loops over statement. 
- Python supports two types of loops: **‘for’** and **‘while’**.

## Python ‘while’ Loop
- The code in a 'while clause' will be executed as long as the 'while' statement's condition is True.
    - The 'while' statement condition is checked first and if it returns True, the block of code inside the 'while' loop is executed.
    - You can see that a while statement looks similar to an if statement. The difference is in how they behave. At the end of an 'if' clause, the program execution continues after the 'if' statement but at the end of a while clause, the program execution jumps back to the start of the while statement.
    - If the 'while' statement condition still returns True, then the statements are executed again. If the test expression is False, the 'while' loop is terminated.
- In code, a while statement always consists of the following:
    - The while keyword
    - A condition (that is, an expression that evaluates to True or False)
    - A colon
    - Starting on the next line, an indented block of code (called the while clause)

- **Syntax:**
```python
while(condition):
    block statement
    increment/decrement or any operation for loop execution
else:
    block statement
```

- **Note: remember to increment/decrement, or else the loop will continue forever.**

> ### ⚠️ The infinite loop
>
> A `while` loop repeats until its condition is false. If nothing in the body can ever make
> it false, it never stops.
>
> ```python
> i = 0
> while i < 5:
>     print(i)        # i never changes - this runs forever
> ```
>
> **In Jupyter:** press the ⏹ stop button (Interrupt Kernel) — you'll get a
> `KeyboardInterrupt`. **In a terminal:** `Ctrl-C`.
>
> Two habits prevent almost all of these:
> 1. When you write `while`, immediately write the line that advances the condition.
> 2. Check that the counter advances on **every** path through the body, not just some.

### `while True` and `break`

When the exit condition can only be evaluated **inside** the loop body, testing it in the
`while` header means writing it twice. The idiomatic alternative is an explicitly infinite
loop with a `break`:

```
while True:
    do the work
    if we're done:
        break
```

This is not a code smell — it is the standard shape for input validation, retry loops and
menu systems. What *is* a smell is a loop with no reachable `break` and no bound.

````python
# `while True` with `break` is the standard "loop until valid" shape.
# It avoids duplicating the test above and inside the loop.

attempts = 0
simulated_inputs = iter(["abc", "-5", "200", "42"])   # stands in for input()

while True:
    attempts += 1
    raw = next(simulated_inputs)
    print(f"  attempt {attempts}: {raw!r}")

    try:
        value = int(raw)
    except ValueError:
        print("    not a whole number")
        continue

    if not 0 < value < 150:
        print("    out of range")
        continue

    break

print(f"\nAccepted {value} after {attempts} attempts")

# ⚠️ Always give an externally-driven loop a maximum, so a bad input stream
# cannot hang your program forever.
MAX_ATTEMPTS = 3
for attempt in range(MAX_ATTEMPTS):
    print(f"bounded attempt {attempt + 1}")
else:
    print("gave up after the limit")
````

### The walrus operator in a `while` condition

> **Version note:** `:=` was added in **Python 3.8** (PEP 572). Full treatment in **1.4**.

A very common `while` shape is "read something, and keep going until it's the sentinel".
Without an assignment expression you have to read **twice** — once before the loop and once
at the bottom — and the two calls must be kept in sync forever.

```
while (line := input("> ")) != "quit":
        |     |
        |     +-- assigns to `line`
        +-------- ...and the comparison sees the value
```

````python
# Without the walrus: you must read once before the loop, and again at the end
# line = input("Say something (or 'quit'): ")
# while line != "quit":
#     print(f"  you said: {line}")
#     line = input("Say something (or 'quit'): ")     <- duplicated

# With the walrus (3.8+): read and test in one place
# while (line := input("Say something (or 'quit'): ")) != "quit":
#     print(f"  you said: {line}")

# Same shape, without needing your keyboard - simulate a stream of inputs
incoming = iter(["hello", "world", "quit", "ignored"])

while (line := next(incoming)) != "quit":
    print(f"  received: {line}")
print("stopped at 'quit'")

# The canonical file-reading form (chunked, so memory stays flat):
# with open("big.bin", "rb") as f:
#     while chunk := f.read(8192):
#         process(chunk)

# Draining a queue
queue = [3, 1, 4, 1, 5]
while queue and (item := queue.pop(0)) is not None:
    print(f"  processing {item}, {len(queue)} left")
    if not queue:
        break
````

### Using while, WAP to find multiples/factors of n
- In this task you will be using a while loop to iterate over natural numbers from 1 to 100 and store only the multiples of n in a list and print that list
    - Initialize an empty list factors and a variable num equal to 1

    - Set the condtion in the while loop until the variable num is less than or equal to 100

    - Check using an if statement whether the variable is divisible by n and if is, append that number(i.e. num) to factors

    - Increment the variable num by 1 post checking the above condition. (note that this increment will be outside the if statement, take care of the indentation in order to avoid a scenario of infinite loop)

    - Print out the factors list

````python
# input value
n = int(input("Enter no.: "))

# initialize
factors=[]
num=1

# while loop
while num<100:
    if n%num==0:
        factors.append(num)
    num+=1
    
# display answer
print(factors)
````

### Using while, WAP to find HCF of two no.s

````python
a = int(input("Enter 1st number: "))
b = int(input("Enter 2nd number: "))
if (a > b):
    smaller = b
else:
    smaller = a
hcf = 1
i=1
while i<smaller+1:
    if ((a % i == 0) and (b % i == 0)):
        hcf = i
    i+=1
    
print(f'HCF of {a},{b} is {hcf}')
````

### Using while, WAP to check whether a 3 digit no. is Armstrong number or not.
- An Armstrong number is a number such that the sum of its digits raised to the third power is equal to the number itself.  For example, 371 is an Armstrong number, since `3**3 + 7**3 + 1**3 = 371`.

````python
n = int(input("Enter your number "))
temp = n
s=0
while temp>0:
    r = temp%10
    s += r**3
    temp = temp//10
if s==n:
    print(f'{n} is Armstrong number')
else:
    print(f'{n} is not a Armstrong number')    
````

### The `while ... else` clause

Python loops can have an `else`. Almost nobody expects what it does:

> The `else` block runs when the loop finishes **because its condition became false** —
> that is, when the loop was **not** ended by `break`.

It is genuinely badly named. Read it as **`nobreak`** and it becomes obvious.

```
while condition:
    ...
    break          <-- if this runs, else is SKIPPED
else:
    ...            <-- runs only if the loop ended naturally
```

**Real-world use case:** searching. "Look through these candidates; if none matched, say
so." Without `else` you need a separate `found = False` flag.

````python
# The idiom `while...else` exists for: searching, and reporting failure once.
def find_divisor(n, limit=100):
    d = 2
    while d < limit:
        if n % d == 0:
            print(f"  {n} is divisible by {d}")
            break
        d += 1
    else:
        # runs ONLY if the while condition became False - i.e. no break happened
        print(f"  {n} has no divisor below {limit}")

find_divisor(91)
find_divisor(97)

# Without `else` you need a flag variable to carry the same information
def find_divisor_flag(n, limit=100):
    d, found = 2, False
    while d < limit:
        if n % d == 0:
            found = True
            break
        d += 1
    if not found:
        print(f"  {n} has no divisor below {limit} (flag version)")

find_divisor_flag(97)

# ⚠️ The name is misleading. Read `else` here as "no break".
print("\nMental model: while ... else  ==  while ... nobreak")

# It also runs when the loop body never executes at all
i = 10
while i < 5:
    print("never")
else:
    print("else still runs - the condition was False from the start")
````

## Common Mistakes & Pitfalls

1. **Forgetting to advance the loop variable.** `while i < 5:` with no `i += 1` runs forever. In Jupyter, use the ⏹ Interrupt button.
2. **Off-by-one in the condition.** `while i < 5` runs 5 times from `i = 0`; `while i <= 5` runs 6. Decide which boundary you mean and write it deliberately.
3. **Modifying the condition variable inside a branch only.** If the counter advances in the `if` but not the `else`, some inputs loop forever.
4. **Using `while` where `for` is natural.** If you're counting through a known sequence, `for` is shorter and cannot run away.
5. **Reading input twice** — once to test and once to use — instead of using the walrus operator.
6. **Expecting `else` to run after a `break`.** The loop `else` runs only when the loop finished *without* breaking. This is the opposite of most people's first guess.
7. **Building floats up by repeated addition in a condition.** `while total != 1.0:` may never terminate because of float error. Compare with `<` or use `math.isclose()`.

## Best Practices

- Use `for` when the number of iterations is known; `while` when it depends on a condition.
- Use `while True:` plus `break` for 'loop until valid' — it is clearer than duplicating the test above and inside the loop.
- Use the walrus operator when the loop condition needs the value it just read.
- Keep the loop body short — extract the work into a function once it passes a screenful.
- Give the loop variable a meaningful name (`attempts`, `remaining`) rather than `i`, when it is not just a counter.
- Add a maximum-iteration guard to any loop driven by external input or a network response.
