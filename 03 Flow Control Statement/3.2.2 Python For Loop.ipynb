## Iteration Vs Iterators Vs Iterables:
### Iteration:
- It is the process of taking one element at a time from a sequence of elements i.e, Any time you use a loop, to go over a group of items is iteration.


### Iterators:
- Iterators are objects that can be iterated upon.
    - i.e, An object which will return data, one element at a time.
- We use the next() function to manually iterate through all the items of an iterator. 
- When we reach the end and there is no more data to be returned, it will raise StopIteration


### Iterables:
- An object is called iterable if we can get an iterator from it.
- The iter() function (which in turn calls the __iter__() method) returns an iterator from them.
- So an iterable is an object that you can get an iterator from.
- Most of built-in containers in Python like: list, tuple, string etc. are iterables.

## Python ‘for’ Loop:
- For-loops in python corresponds to what are called **for-each loops** in many other programming languages.
- A **for** loop acts as an iterator in Python, it goes through items that are in a *collection* or any other iterable item.
    - i.e, for loop is used when we want to iterate over a collection.
    - and they request items one-by-one from a collection or more strictly from an iterable series and assign them in turn to a variable we specify.
    - for loop will continue to iterate over it's elements until there is element left in collection.
- 'for' and 'in' keywords are used.
- Collection can be iterator or iterable.
    - Examples of iterable objects are string, list, tuple, dictionary, sets, range object etc.
    - Example of iterator objects are enumerate object, map object, filter object etc.


- **Syntax:** 
```python
for var in sequence:
    block statement(s)
```

- **Actually implementation of for loop:**

```python
# create an iterator object from that iterable
iter_obj = iter(iterable)

# infinite loop
while True:
    try:
        # get the next item
        element = next(iter_obj)
        # do something with element
    except StopIteration:
        # if StopIteration is raised, break from loop
        break
```
- So internally, the for loop creates an iterator object, iter_obj by calling iter() on the iterable.
    - Ironically, this for loop is actually an infinite while loop.
    - Inside the loop, it calls next() to get the next element and executes the body of the for loop with this value. After all the items exhaust, StopIteration is raised which is internally caught and the loop ends. Note that any other kind of exception will pass through.

### WAP to check for anagram of two strings

````python
# NOTE: the original version of this cell used `x` and `y` without ever
# assigning them, so it raised NameError and could never run. The two
# input lines below were missing.

x = input("Enter string 1: ")
y = input("Enter string 2: ")

x = x.lower().replace(' ', '')
y = y.lower().replace(' ', '')

if len(x) != len(y):
    print("Not anagrams (different lengths)")
else:
    # An anagram uses exactly the same letters, so the sorted forms must match
    if sorted(x) == sorted(y):
        print("Anagrams")
    else:
        print("Not anagrams")
````

````python
# The same check, without needing input - and three ways to write it
from collections import Counter


def normalise(text):
    return "".join(ch for ch in text.lower() if ch.isalpha())


def is_anagram_sorted(a, b):
    return sorted(normalise(a)) == sorted(normalise(b))


def is_anagram_counter(a, b):
    return Counter(normalise(a)) == Counter(normalise(b))


def is_anagram_manual(a, b):
    """The explicit loop version - what the other two do for you."""
    a, b = normalise(a), normalise(b)
    if len(a) != len(b):
        return False
    tally = {}
    for ch in a:
        tally[ch] = tally.get(ch, 0) + 1
    for ch in b:
        if ch not in tally:
            return False
        tally[ch] -= 1
        if tally[ch] == 0:
            del tally[ch]
    return not tally


pairs = [
    ("Python is great", "Thorn pig teasy"),
    ("listen", "silent"),
    ("Astronomer", "Moon starer"),
    ("hello", "world"),
]

for a, b in pairs:
    results = {is_anagram_sorted(a, b), is_anagram_counter(a, b), is_anagram_manual(a, b)}
    assert len(results) == 1, f"methods disagree on {a!r}/{b!r}"
    print(f"  {a!r:<22} vs {b!r:<18} -> {results.pop()}")
````

### `enumerate()` — index *and* item, without the bookkeeping

The cell above uses `for i in range(len(marks))` and then indexes back in. It works, but it
is the single most common non-Pythonic loop, and it has three problems: it only works on
things you can index, it says "index" when you mean "item", and it's easy to get the
`len()` wrong.

**`enumerate()`** hands you both at once.

### Syntax breakdown

```
for index, item in enumerate(iterable, start=0):
     |      |                    |         |
     |      |                    |         +-- what the counter starts at
     |      |                    +------------ any iterable
     |      +--------------------------------- the element
     +---------------------------------------- the running count
```

**Real-world use case:** numbering output for humans (`start=1`), and reporting *which*
line of a file failed to parse.

````python
fruits = ["apple", "banana", "cherry"]

# The C-style way - works, but indirect
print("range(len()):")
for i in range(len(fruits)):
    print(f"  {i}: {fruits[i]}")

# enumerate() yields (index, item) pairs - one clear step
print("\nenumerate():")
for i, fruit in enumerate(fruits):
    print(f"  {i}: {fruit}")

# start= shifts the counter, without touching the data
print("\nenumerate(start=1) - human-friendly numbering:")
for n, fruit in enumerate(fruits, start=1):
    print(f"  {n}. {fruit}")

# It works on ANY iterable, not just lists
print("\nover a string:", list(enumerate("abc")))
print("over a dict  :", list(enumerate({"a": 1, "b": 2})))

# Common real use: report the position of a bad row
rows = ["10,20", "30,40", "oops", "50,60"]
for line_no, row in enumerate(rows, start=1):
    if "," not in row:
        print(f"\nLine {line_no}: malformed -> {row!r}")

# enumerate returns an iterator - wrap in list() to see it
print("\nthe object itself:", enumerate(fruits))
````

### `zip()` — walking several sequences in step

`zip()` takes several iterables and yields tuples of their corresponding items. It is how
you loop over parallel data without indexing.

```
zip(names, scores)  ->  ("Aditya", 88), ("Priya", 95), ...
```

> **Version note:** `strict=True` was added in **Python 3.10**. Without it, `zip()`
> **silently stops at the shortest input** — a genuine source of data-loss bugs.

````python
names = ["Aditya", "Priya", "Rahul"]
scores = [88, 95, 72]
grades = ["B", "A", "C"]

# zip() walks several iterables in step, yielding tuples
for name, score in zip(names, scores):
    print(f"  {name:<8} {score}")

# Unpacking works for any number of iterables
print()
for name, score, grade in zip(names, scores, grades):
    print(f"  {name:<8} {score:>3}  {grade}")

# zip + enumerate together
print()
for rank, (name, score) in enumerate(zip(names, scores), start=1):
    print(f"  {rank}. {name:<8} {score}")

# ⚠️ zip stops at the SHORTEST input - silently
short = [1, 2]
print("\nunequal lengths:", list(zip(names, short)), " <- Rahul vanished")

# strict=True (3.10+) turns that silence into an error
try:
    list(zip(names, short, strict=True))
except ValueError as exc:
    print("strict=True     :", exc)

# zip(*rows) transposes
matrix = [(1, 2, 3), (4, 5, 6)]
print("\ntranspose:", list(zip(*matrix)))

# Building a dict from two parallel lists
print("as dict  :", dict(zip(names, scores)))

# ⚠️ zip returns an ITERATOR - it can only be consumed once
z = zip(names, scores)
print("\nfirst pass :", list(z))
print("second pass:", list(z), " <- exhausted")
````

````python
# ---- for ... else ----
# `else` runs only if the loop finished WITHOUT hitting `break`.
# Read it as "nobreak".

def find_first_negative(numbers):
    for n in numbers:
        if n < 0:
            print(f"  found {n}")
            break
    else:
        print("  no negative numbers")

find_first_negative([3, 7, -2, 8])
find_first_negative([3, 7, 2, 8])

# The flag-variable version it replaces
def find_first_negative_flag(numbers):
    found = False
    for n in numbers:
        if n < 0:
            found = True
            break
    if not found:
        print("  no negative numbers (flag version)")

find_first_negative_flag([3, 7, 2, 8])


# ---- Two more loop-header helpers ----
marks = [10, 11, 19, 14]

print("\nreversed:", [m for m in reversed(marks)])
print("sorted  :", [m for m in sorted(marks, reverse=True)])

# ⚠️ The loop variable LEAKS - it survives after the loop
for i in range(3):
    pass
print("\nafter the loop, i =", i)

# ...and if the sequence was empty, it was never created at all
for j in []:
    pass
print("j exists?", "j" in dir())
````

### Using for, WAP to check whether a user input number is prime or not

````python
# Method 1:
num = int(input('Enter your number: '))
count = 0
if (num == 1):
    print('1 is neither prime nor composite')
else:
    for i in range (1, num+1):
        if (num % i == 0):
            count += 1
    if (count == 2):
        print(f'{num} is prime number')
    else:
        print(f'{num} is not a prime number')
````

### WAP to check whether a number is semiprime or not
A semiprime is a natural number that is a product of two prime numbers.

````python
# A semiprime is the product of exactly two primes (not necessarily distinct).
# 4 = 2x2, 6 = 2x3, 9 = 3x3 are semiprimes; 8 = 2x2x2 is not.

def prime_factor_count(n):
    """Count prime factors of n, with multiplicity."""
    count = 0
    factor = 2
    while factor * factor <= n:
        while n % factor == 0:
            n //= factor
            count += 1
        factor += 1
    if n > 1:            # whatever is left is prime
        count += 1
    return count


def is_semiprime(n):
    return n > 1 and prime_factor_count(n) == 2


for n in range(1, 31):
    if is_semiprime(n):
        print(n, end=" ")
print("  <- semiprimes up to 30")

print("\nchecks:")
for n in [4, 6, 8, 9, 15, 16, 21, 25, 30]:
    print(f"  {n:>3}: {prime_factor_count(n)} prime factors -> "
          f"{'semiprime' if is_semiprime(n) else 'not'}")
````

### Signpost: `itertools`

Before writing a clever nested loop, check whether the standard library already has it.
`itertools` is a toolbox of iterator builders, all lazy and memory-efficient.

| Function | Does |
|---|---|
| `product(a, b)` | Every combination — replaces a nested loop |
| `chain(a, b)` | Treat several iterables as one continuous stream |
| `cycle(a)` | Repeat an iterable endlessly |
| `groupby(a)` | Group *consecutive* equal items |
| `accumulate(a)` | Running totals |
| `islice(a, n)` | Slice an iterator you can't index |
| `pairwise(a)` | Every adjacent pair (3.10+) |
| `combinations` / `permutations` | Exactly what they say |

Full treatment in **07 Module and Packages**.

````python
import itertools

colours = ["red", "green"]
sizes = ["S", "M", "L"]

# Every combination of two lists - instead of a nested loop
print("product:", list(itertools.product(colours, sizes))[:4], "...")

# Chain several iterables into one stream, without building a combined list
print("chain  :", list(itertools.chain([1, 2], (3, 4), {5})))

# Run through a short iterable forever (paired with zip, this cycles)
print("cycle  :", list(zip(range(7), itertools.cycle(["a", "b", "c"]))))

# Group consecutive equal items
data = "aaabbbcccd"
print("groupby:", [(k, len(list(g))) for k, g in itertools.groupby(data)])

# Running totals
print("accum  :", list(itertools.accumulate([1, 2, 3, 4, 5])))

# Take the first n of an endless sequence
print("islice :", list(itertools.islice(itertools.count(10, 5), 4)))

# Pairwise (3.10+): every adjacent pair - great for diffs between readings
print("pairwise:", list(itertools.pairwise([1, 4, 9, 16])))
````

## Common Mistakes & Pitfalls

1. **`for i in range(len(seq))` then `seq[i]`.** Iterate the sequence directly, or use `enumerate()` when you genuinely need the index.
2. **Modifying a list while iterating it.** Removing shifts the indices under the loop and silently skips items. Iterate a copy, or build a new list.
3. **Assuming `zip()` complains about unequal lengths.** It stops at the shortest, silently. Pass `strict=True` (3.10+) if that would be a bug.
4. **Reusing an exhausted iterator.** Generators, `zip`, `map` and file objects can only be walked once. A `list` or `range` can be walked repeatedly.
5. **Expecting the loop variable to be scoped to the loop.** It leaks — after `for i in range(3)`, `i` is still `2`. And if the sequence was empty, `i` never exists at all.
6. **Expecting `else` to run after a `break`.** Loop `else` means 'no break happened'.
7. **Iterating a dict expecting pairs.** `for x in d` gives you *keys*. Use `.items()`.
8. **Building a list with `append` in a loop** where a comprehension would say it in one line (see **3.3**).

## Best Practices

- Iterate the object directly: `for item in items:` — not over indices.
- Use `enumerate(items, start=1)` when you need a counter, rather than maintaining one.
- Use `zip()` for parallel iteration; add `strict=True` when the lengths must match.
- Use `.items()` to iterate a dict's pairs, `.values()` when you don't need the keys.
- Use `_` for a loop variable you never read: `for _ in range(3):`.
- Use `for...else` instead of a `found = False` flag when searching.
- Reach for `itertools` before writing a clever nested loop — it's probably already there.
- Prefer a comprehension when the loop's only job is to build a collection.
