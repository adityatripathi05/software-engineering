### WAP to print pattern

## How to think about pattern problems

Every pattern question is the same question in disguise:

> **For row `r`, what goes in that row?**

Answer that on paper and the code writes itself. Three things to work out:

1. **How many rows?** -> the outer loop's `range`.
2. **How many characters in row `r`?** -> the inner loop's `range`, expressed in terms of `r`.
3. **How many leading spaces in row `r`?** -> usually `n - r` for a left-leaning pyramid.

For a right triangle of `n` rows, row `r` has `r` stars. For a pyramid, row `r` has
`2r - 1` stars and `n - r` leading spaces. That's the whole trick.

### Three ways to write the same pattern

Nested loops are the classic teaching approach, but they are rarely the shortest. String
multiplication and f-string alignment often replace the inner loop entirely.

````python
n = 5

# ---- Approach 1: nested loops, character by character ----
print("nested loops:")
for row in range(1, n + 1):
    for col in range(row):
        print("*", end="")      # end="" keeps us on the same line
    print()                     # newline at the end of the row

# ---- Approach 2: build each row as a string ----
print("\nrow strings:")
for row in range(1, n + 1):
    line = "*" * row
    print(line)

# ---- Approach 3: f-string alignment for centring ----
print("\ncentred pyramid:")
for row in range(1, n + 1):
    stars = "*" * (2 * row - 1)
    print(f"{stars:^{2 * n - 1}}")

# The same pyramid with spaces computed by hand - what the alignment does for you
print("\nsame, spaces by hand:")
for row in range(1, n + 1):
    spaces = " " * (n - row)
    stars = "*" * (2 * row - 1)
    print(spaces + stars)
````

### 8. Challenge for you!
- Given an integer n, print a nXn matrix with squares (outermost to innermost) having values n, then n-1 and so on.
For example, given 5, output is:
    
```
5 5 5 5 5
5 4 4 4 5
5 4 3 4 5
5 4 4 4 5 
5 5 5 5 5
```

### Solution to the challenge

Below is one answer to the `n x n` matrix of squares. Try it yourself before reading.

````python
n = 5

# Build the values first, then worry about formatting.
matrix = [[(row * n + col + 1) ** 2 for col in range(n)] for row in range(n)]

# Work out the widest entry so every column lines up
width = max(len(str(v)) for r in matrix for v in r)

for row in matrix:
    print(" ".join(f"{v:>{width}}" for v in row))
````

## Common Mistakes & Pitfalls

1. **Forgetting `end=''`.** `print()` adds a newline every call, so a row printed character by character becomes a column.
2. **Off-by-one in `range()`.** For `n` rows you want `range(1, n + 1)` if the row number is the count, or `range(n)` if you will add 1 yourself. Pick one and be consistent.
3. **Mixing up rows and columns.** The outer loop is rows; the inner loop is what goes *in* a row. Getting this backwards transposes the shape.
4. **Printing spaces with the wrong count** in pyramids — the leading spaces are usually `n - row`, not `n`.
5. **Building the row with `+=` inside the inner loop** rather than `print(..., end='')` or `''.join()`. Correct, but slow (see **2.1**).

## Best Practices

- Work out the **relationship between the row index and the row's contents** on paper first — the code follows immediately.
- Use string multiplication (`'*' * n`) instead of an inner loop when the row is uniform.
- Use f-string alignment (`f'{s:^20}'`) for centring rather than computing spaces by hand.
- Build each row as a string and print it once — easier to debug than character-by-character.
- Name the loop variables `row` and `col`, not `i` and `j`, while learning.
