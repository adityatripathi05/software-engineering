- Program is like a series of instructions. But programming’s real strength isn’t just running one instruction after another. Based on how expressions evaluate, a program can decide to skip instructions, repeat them, or choose one of several instructions to run.

## Flow Control Statement
- These statement controls the flow of execution of code i.e, Flow control statements can decide which Python instructions to execute under which conditions.
- Flow Control statements can be categories as:
    - 1. **Conditional Statements**
        - 1.1 **if** statement
        - 1.2 **if-else** statement
        - 1.3 **if-elif-else** statement
    - 2. **Loop Statements**
        - 2.1 **while** loop statement
        - 2.2 **for** loop statement
    - 3. **Loop Control Statements**
        - 3.1 **break** statement
        - 3.2 **continue** statement
        
 
### Elements of Flow Control
- Flow control statements often start with a part called **the condition** and are always followed by a **block of code** called **the clause**.
<br/><br/>
- **Conditions:**
    - The Boolean expressions we’ve seen so far could all be considered conditions, which are the same thing as expressions; condition is just a more specific name in the context of flow control statements. 
    - Conditions always evaluate down to a Boolean value, True or False. 
    - A flow control statement decides what to do based on whether its condition is True or False, and almost every flow control statement uses a condition.
<br/><br/>
- **Blocks of Code:**
    - Most of the programming languages like C, C++, Java use braces { } to define a block of code. Python uses indentation.
    - Lines of Python code can be grouped together in blocks.
    - A code block (body of a function, loop etc.) starts with indentation and ends with the first unindented line.
    - We can tell when a block begins and ends from the indentation of the lines of code. 
    - There are three rules for blocks:
        - Blocks begin when the indentation increases.
        - Blocks can contain other blocks.
        - Blocks end when the indentation decreases to zero or to a containing block’s indentation.
    - Generally four whitespaces are used for indentation and is preferred over tabs.

## Python Conditional Statements (Decision Making)
- While programming, we will encounter many situations where our program has to execute statements based on certain conditions. These statements are called conditional statements.
- The conditional statements provided by python are:
    - **if** statement
    - **if-else** statement
    - **if-elif-else** statement

## Python 'if' Statement:
- An 'if' statement's clause(block following the if statement) will execute if the statement's condition is True.
- The clause is skipped if the condition is False.
- In plain English, an if statement could be read as, “If this condition is true, execute the code in the clause.”
-  In Python, an if statement consists of the following:
    - The if keyword
    - A condition (that is, an expression that evaluates to True or False)
    - A colon
    - Starting on the next line, an indented block of code (called the if clause)

- **Syntax:**
```python
if (condition):
    block statement(s)
```   
  
- **NOTE:** Space before block statement is called **indentation** and is generally equal to 4 spaces or 1 tab. 
    - In Python, it is required for indicating what block of code a statement belongs to.
    - Indentation works in Python to maintain the structure and order of your code.

## Python if-else Statement:
- An 'if' clause can optionally be followed by an 'else' statement.
- The 'else' clause will execute only when the 'if' statement condition is False.
- In plain English, an else statement could be read as, “If this condition is true, execute this code. Or else, execute that code.” 
- An else statement doesn’t have a condition, and in code, an else statement always consists of the following:
    - The else keyword
    - A colon
    - Starting on the next line, an indented block of code (called the else clause)

- **Syntax:**
```python
if(condition):
    block statement(s)
else:
    block statement(s)
```

## Python if-elif-else Statement:
- While only one of the if or else clauses will execute, we may have cases where we want one of many possibleclauses to execute. For that we use 'elif' statement.
- The 'elif' statement is an 'else if' statement that always follows an 'if' or another elif statement.
- It provides another condition that is checked only if all of the previous conditions were False.
- We can have multiple 'elif' statement.
- In code, an elif statement always consists of the following:
    - The elif keyword
    - A condition (that is, an expression that evaluates to True or False)
    - A colon
    - Starting on the next line, an indented block of code (called the elif clause)

- **Syntax:**
```python
if (condition):
    block statement(s)
elif (condition):
    block statement(s)
elif (condition):    
    block statement(s)    
else:
    block statement(s)
```

### Conditional expressions (Python's ternary)

Many languages have a ternary operator written `condition ? a : b`. Python spells it in
words, and puts the **value first** so the common case reads naturally.

### Syntax breakdown

```
    value_if_true   if   condition   else   value_if_false
          |               |                    |
          +-- used when the condition is truthy
                          +-- any expression
                                               +-- used otherwise
```

Because it is an **expression**, it produces a value — so it can sit inside an f-string, a
function call, a list comprehension or a `return`, where an `if` *statement* cannot.

> **Use it for choosing a value.** Use an `if` statement for choosing an *action*. If your
> conditional expression needs parentheses to be readable, it wanted to be an `if` statement.

````python
age = 20

# As a statement - four lines
if age >= 18:
    status = "adult"
else:
    status = "minor"
print(status)

# As a conditional expression - one line, and `status` is assigned exactly once
status = "adult" if age >= 18 else "minor"
print(status)

# It is an EXPRESSION, so it can go anywhere a value can go
print(f"You are an {'adult' if age >= 18 else 'minor'}")

names = ["alice", "", "bob", ""]
print([n if n else "(anonymous)" for n in names])

# They can be chained, but readability drops fast
score = 85
grade = "A" if score >= 90 else "B" if score >= 80 else "C" if score >= 70 else "F"
print("\ngrade:", grade, " <- works, but an if/elif chain is clearer here")

# ⚠️ Precedence trap: the conditional binds loosely
print("\n" + str(1 + 1 if False else 10))     # (1+1) if False else 10  -> 10
print(str(1 + (1 if False else 10)))           # 1 + 10                  -> 11

# `or` as a shorthand default - common, but beware falsy-but-valid values
username = ""
print("\nor-default :", username or "anonymous")
print("explicit   :", username if username != "" else "anonymous")
````

### Guard clauses: the cure for nested `if`

Every extra level of nesting costs the reader. Once you are three levels deep, the
condition that matters is far away from the code it guards, and the `else` branches are
pages apart from their `if`.

A **guard clause** inverts the test and leaves early — handling the exceptional cases
first, so the main logic ends up unindented at the bottom.

**Analogy:** a bouncer at a door. Rather than one enormous "if you're on the list and old
enough and dressed correctly then come in", you check each rule and turn people away as
they fail. What's left at the end is simply "come in".

````python
# Deeply nested - the "arrow" or "pyramid of doom"
def process_order_nested(order):
    if order is not None:
        if "items" in order:
            if len(order["items"]) > 0:
                if order.get("paid"):
                    return f"Shipping {len(order['items'])} items"
                else:
                    return "Not paid"
            else:
                return "No items"
        else:
            return "Malformed order"
    else:
        return "No order"


# Same logic, guard clauses - handle failures first and leave early
def process_order_guarded(order):
    if order is None:
        return "No order"
    if "items" not in order:
        return "Malformed order"
    if not order["items"]:
        return "No items"
    if not order.get("paid"):
        return "Not paid"

    return f"Shipping {len(order['items'])} items"


cases = [
    None,
    {"paid": True},
    {"items": [], "paid": True},
    {"items": ["book"], "paid": False},
    {"items": ["book", "pen"], "paid": True},
]

for case in cases:
    a, b = process_order_nested(case), process_order_guarded(case)
    assert a == b, (a, b)
    print(f"  {str(case)[:34]:<36} -> {b}")

print("\nIdentical behaviour. The guarded version never indents past one level,")
print("and each rule sits next to the reason it exists.")
````

### What counts as a condition?

`if` does not require a boolean. It calls `bool()` on whatever you give it, so **any**
object works. The full falsy list is in **2.2 Python Numeric Datatype**; the short version:

> Falsy: `None`, `False`, `0`, `0.0`, `""`, `[]`, `()`, `{}`, `set()`.
> **Everything else is truthy** — including `"False"`, `"0"` and `[0]`.

This is why `if items:` is the idiomatic way to ask "is this empty?" — but it's also a trap
when `0` or `""` are legitimate values.

````python
# Every object has a truth value - `if` calls bool() on whatever you give it
items = []
name = ""
count = 0
data = None

for label, value in [("list", items), ("str", name), ("int", count), ("None", data)]:
    print(f"  {label:<5} {value!r:<6} -> {'truthy' if value else 'falsy'}")

# The idiomatic empty check
if not items:
    print("\nlist is empty (Pythonic)")

# ⚠️ but `not x` is NOT the same as `x is None`
def get_discount(user_discount=None):
    # BUG: a legitimate discount of 0 is treated as "not supplied"
    if not user_discount:
        return "default 10%"
    return f"{user_discount}%"

def get_discount_fixed(user_discount=None):
    if user_discount is None:
        return "default 10%"
    return f"{user_discount}%"

print("\ndiscount=0 (buggy):", get_discount(0), " <- wrong, 0 was explicit")
print("discount=0 (fixed):", get_discount_fixed(0))

# Comparing floats in a condition
import math
x = 0.1 + 0.2
print("\nif x == 0.3        ->", x == 0.3)
print("if isclose(x, 0.3) ->", math.isclose(x, 0.3))

# Chained comparisons read like maths
age = 25
print("\n18 <= age <= 65    ->", 18 <= age <= 65)
````

### Beyond `if`/`elif`: two alternatives worth knowing

An `elif` chain is the right tool when each branch tests a genuinely different *condition*.
When you're really just **mapping a value to a result**, there are cleaner options.

| Situation | Better tool |
|---|---|
| Mapping a value to a value | A **dict lookup** |
| Testing membership in a set of options | `in` |
| Dispatching on the **shape** of data (a list of 2, a dict with certain keys, an object of some class) | **`match`/`case`** — see **3.4** |

> `match`/`case` (Python 3.10+) is covered in **3.4 Structural Pattern Matching**. It is
> *not* a C-style switch — it matches structure, not just equality — which is why it earns
> a notebook of its own.

````python
# A long elif chain for simple value mapping...
def describe_role_if(role):
    if role == "admin":
        return "Full access"
    elif role == "editor":
        return "Can edit content"
    elif role == "viewer":
        return "Read only"
    else:
        return "Unknown role"

# ...is often clearer as a dict lookup
ROLE_MESSAGES = {
    "admin": "Full access",
    "editor": "Can edit content",
    "viewer": "Read only",
}

def describe_role_dict(role):
    return ROLE_MESSAGES.get(role, "Unknown role")

for r in ["admin", "editor", "viewer", "ghost"]:
    assert describe_role_if(r) == describe_role_dict(r)
    print(f"  {r:<8} {describe_role_dict(r)}")

print("\nBoth agree. The dict version puts the DATA in one place,")
print("separate from the logic that uses it - easier to extend and to test.")

# Membership beats a long `or` chain
command = "quit"

# Clumsy
if command == "q" or command == "quit" or command == "exit":
    print("\nexiting (or-chain)")

# Better
if command in {"q", "quit", "exit"}:
    print("exiting (membership test)")
````

## Common Mistakes & Pitfalls

1. **Using `=` instead of `==` in a condition.** Python protects you here — `if x = 5:` is a `SyntaxError`, not a silent bug as it is in C. Use `:=` if you genuinely mean to assign.
2. **Writing `if x == True:`.** Just write `if x:`. And `if x is True:` is stricter than you want — `1 == True` but `1 is not True`.
3. **Comparing floats with `==`.** `if 0.1 + 0.2 == 0.3:` never fires. Use `math.isclose()`.
4. **Chaining `if` when you meant `elif`.** Separate `if` statements all get evaluated; an `elif` chain stops at the first match. With overlapping conditions this changes the answer.
5. **Forgetting the colon**, or mixing tabs and spaces in the block. Python 3 raises `TabError`.
6. **Deeply nested `if`s.** Past three levels, readability collapses. Use guard clauses.
7. **`if len(items) > 0:`** instead of `if items:`. The second is idiomatic and works for every container.
8. **Assuming an `else` binds to the outer `if`.** In Python indentation decides, so this is less dangerous than in C — but misindenting an `else` silently changes the logic.

## Best Practices

- Use truthiness: `if items:` not `if len(items) != 0:`.
- Prefer **guard clauses** (early `return`/`continue`) over deep nesting.
- Use `elif` for mutually exclusive branches so only one test runs.
- Use chained comparisons: `if 0 <= x <= 100:`.
- Use `in` against a collection instead of a long `or` chain: `if cmd in {'q', 'quit', 'exit'}:`.
- Keep conditions positive where you can — `if is_valid:` reads better than `if not is_invalid:`.
- Use a conditional expression for simple value selection, an `if` statement for branching logic.
- For multi-way dispatch on the *shape* of data, consider `match`/`case` — see **3.4**.
