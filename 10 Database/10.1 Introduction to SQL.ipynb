## Running the SQL in this notebook

The rest of this notebook is a SQL reference. Rather than leaving it as text you cannot try,
every statement below runs against an **in-memory SQLite database** created by the cell that
follows.

Two things to know:

- **`:memory:`** gives you a throwaway database that exists only while the notebook runs.
  Nothing is written to disk, and re-running the cell starts fresh.
- **SQLite is not MySQL.** The SQL *core* — `CREATE TABLE`, `INSERT`, `SELECT`, `JOIN`,
  `GROUP BY` — is standard and portable. The edges are not. Dialect differences are flagged
  as they come up.

> Connecting to SQLite from Python is the subject of **10.3**; here it is just plumbing, so
> you can execute the SQL.

```python
import sqlite3

# A throwaway database that lives only in RAM
conn = sqlite3.connect(":memory:")
conn.row_factory = sqlite3.Row          # rows behave like dicts - see 10.3
cur = conn.cursor()


def run(sql: str, params=(), show: bool = True):
    """Execute one SQL statement and print the result as a table."""
    cur.execute(sql, params)
    if cur.description is None:                 # not a SELECT
        conn.commit()
        if show:
            print(f"OK ({cur.rowcount} row(s) affected)" if cur.rowcount >= 0 else "OK")
        return None

    rows = cur.fetchall()
    if not show:
        return rows
    if not rows:
        print("(no rows)")
        return rows

    headers = rows[0].keys()
    widths = [max(len(str(h)), *(len(str(r[h])) for r in rows)) for h in headers]
    print("  ".join(str(h).ljust(w) for h, w in zip(headers, widths)))
    print("  ".join("-" * w for w in widths))
    for r in rows:
        print("  ".join(str(r[h]).ljust(w) for h, w in zip(headers, widths)))
    return rows


print("SQLite", sqlite3.sqlite_version, "- in-memory database ready")
```

### SQL Constraints
- SQL Constraints increase the accuracy and reliability of data that goes into the database through a table.
- These constraints make sure that there is no violation in terms of a transaction of data, if found then the action will be terminated. 
- The main use of constraints is to limit the type of data that can go into a table.
- The most used constraints are:
    - **DEFAULT:** When no value is specified then a set of default values for a column is added
    - **NOT NULL:** This makes sure that a NULL value will not be stored in a column
    - **UNIQUE:**  The UNIQUE constraint is used to ensure all the values entered in a column or a table are unique.
    - **INDEX:** This is used to create and also retrieve data from the database
    - **CHECK:** If we want to satisfy a specific condition in a column then we use CHECK constraint
    - **PRIMARY KEY:** It is the candidate key that is selected to uniquely identify a tuple in a relation.
    - **UNIQUE KEY:** The unique key is similar to the primary key, but allows one NULL value in the column.
    - **FOREIGN KEY:** A foreign key is a set of one or more columns in the child table whose values are required to match with corresponding columns in the parent table.
    - **CANDIDATE KEY:** A set of attributes which can uniquely identify a table can be termed as a Candidate Key. A table can have more than one candidate key, and out of the chosen candidate keys, one key can be chosen as a Primary Key. 
    - **SUPER KEY:** The set of attributes which can uniquely identify a tuple is known as Super Key. So, a candidate key, primary key, and a unique key is a superkey, but vice-versa isn’t true.
    - **ALTERNATE KEY:** Alternate Keys are the candidate keys, which are not chosen as a Primary key. 
    - **COMPOSITE KEY:** A composite key is a combination of two or more columns that identify each tuple uniquely.

### How to Read Data?
- **SELECT** query is used for manipulating a database. 
- The select command is used to select the data from the database and display it to the user.
- **Syntax:**
`Select column 1, column 2…..column N
From Table;`
- **Example:**
`SELECT FName,LName FROM student;`
<br/><br/>
- The above example will display all the names from the student table. 
- If we want to display all the fields in the table then we have to use *(Star) operator. This will display the entire table.
- **Example:** 
`SELECT * FROM student;`
<br/><br/>
- If we want to display certain field without any duplicates then we use the DISTINCT keyword along with the select command. 
- **Example:**
`SELECT DISTINCT ADDRESS FROM student;`
<br/><br/>
- **WHERE**
    - If we need only certain records from the table then we use the where clause. 
    - Where clause acts as a Filtering mechanism. 
    - Under the Where section we need to specify certain conditions, only if those conditions are met the records will be extracted.
- **Syntax:**
`SELECT column1, column2, ...column N FROM table_name WHERE condition;`
- **Example:** 
`SELECT FName FROM Students WHERE City='Delhi';`
<br/><br/>
- **Retrieving Data with "?"**
- **Example:**
`SELECT FName,Percentage FROM student WHERE Roll=?, (103,);`
<br/><br/>
- **AND, OR, NOT**
    - If we need to add two or more conditions in the where clause then we can use the above-mentioned operators. 
    - These keywords will add more complexity to the query. 
<br/><br/>
- **AND Operator:** 
    - This operator displays a record if all the conditions separated by AND are TRUE.
- **Syntax:**
`SELECT column1, column2, ... FROM table_name WHERE condition1 AND condition2 AND condition3 ...;`
- **Example:**
`SELECT FName,Percentage FROM student WHERE Address='Delhi' AND Percentage>80;`
<br/><br/>
- **OR Operator:** 
    - This operator displays a record if any of the conditions separated by OR is TRUE.
- **Syntax:**
`SELECT column1, column2, ... FROM table_name WHERE condition1 OR condition2 OR condition3 ...;`
- **Example:**
`SELECT FName,Percentage FROM student WHERE Address='Delhi' OR Percentage>80;`
<br/><br/>
- **NOT Operator:** 
    - This operator displays a record if the condition/conditions are NOT TRUE.
- **Syntax:**
`SELECT column1, column2, ... FROM table_name WHERE NOT condition;`
- **Example:**
`SELECT * FROM Student WHERE NOT Lname='Doe';`
<br/><br/>
- **IN operator:** 
    - This operator is used to specify multiple values inside the WHERE clause. It acts as a short for multiple OR.
- **Syntax:**
`SELECT column_name(s) FROM table_name WHERE column_name IN (value1, value2, ...);`
- **Example:**
`SELECT FName,Percentage FROM student WHERE Address IN ('Delhi','UP');`
<br/><br/>
- **BETWEEN operator:** 
    - This operator will select a particular value within the specified range. It is compulsory to add the beginning and the end value (Range).
- **Syntax:**
`SELECT column_name(s) FROM table_name WHERE column_name BETWEEN value1 AND value2;`
- **Example:**
`SELECT FName,Percentage FROM student WHERE Percentage BETWEEN 85 AND 95;`
<br/><br/>
- **LIKE operator:**
    - This operator is used with a WHERE clause to search for a specified pattern in a column of a table. 
    - There are mainly two wildcards that are used in conjunction with the LIKE operator:
        - **%:** It matches 0 or more character.
        - **_:** It matches exactly one character.

- **Example:**
`SELECT FName,Percentage FROM student WHERE FName LIKE 'A%';`

```python
print("every column - convenient when exploring, fragile in real code:")
run("SELECT * FROM student;")

print("\nnamed columns:")
run("SELECT FName, LName FROM student;")

print("\nDISTINCT:")
run("SELECT DISTINCT Address FROM student;")

print("\nORDER BY - without it, row order is UNDEFINED:")
run("SELECT FName, Percentage FROM student "
    "WHERE Percentage IS NOT NULL ORDER BY Percentage DESC;")

print("\nLIMIT - always use this while exploring:")
run("SELECT FName, Percentage FROM student "
    "WHERE Percentage IS NOT NULL ORDER BY Percentage DESC LIMIT 2;")

print("\nLIMIT with OFFSET, for paging:")
run("SELECT FName FROM student ORDER BY Roll LIMIT 2 OFFSET 2;")

print("""
The logical order SQL evaluates a query in - which is NOT the order you write it:
    FROM -> WHERE -> GROUP BY -> HAVING -> SELECT -> ORDER BY -> LIMIT

That is why you cannot use a SELECT alias in a WHERE clause, but you CAN in
an ORDER BY.
""")
```

### How to give Aliases in SQL?
- Alias is a process of giving a table or a column a temporary name so that it helps when the query is complex. 
- It increases the readability of the query. This renaming is temporary and the table name does not change in the original database. We can alias a column or a table.
<br/><br/>
- **The syntax for Column Aliasing:**
`SELECT column_name AS alias_name FROM table_name;`
- **Example for Column Aliasing:**
`SELECT FName AS First, LName AS Last FROM student;`
<br/><br/>
- **Syntax for Table Aliasing:**
`SELECT column_name(s) FROM table_name AS alias_name;`
- **Example for Table Aliasing:**
`SELECT S.FName, S.LName FROM student as S`

```python
print("column alias:")
run("SELECT FName AS first_name, Percentage AS score FROM student LIMIT 3;")

print("\nan alias CAN be used in ORDER BY (evaluated after SELECT):")
run("SELECT FName, Percentage AS score FROM student "
    "WHERE Percentage IS NOT NULL ORDER BY score DESC LIMIT 3;")

print("\n...but NOT in WHERE (evaluated before SELECT):")
try:
    run("SELECT Percentage AS score FROM student WHERE score > 80;")
except sqlite3.OperationalError as exc:
    print("  ", exc)

print("\ntable alias - the reason joins stay readable:")
run("SELECT s.FName, s.Percentage FROM student AS s WHERE s.Percentage > 80;")
```

### AGGREGATE FUNCTIONS
- An aggregate function is a function where the values of multiple rows are grouped together as input on certain criteria and a single value is returned. 
- We often use aggregate functions with the GROUP BY and HAVING clauses of the SELECT statement. 
- Some of the Aggregate functions are COUNT, SUM, AVG, MIN, MAX.
<br/><br/>
- **COUNT():** 
    - This function returns the number of rows that match specified criteria.
- **Syntax:** 
`SELECT COUNT(column_name) FROM table_name WHERE condition;`
- **Example:**
`SELECT COUNT (Roll) FROM student;`
<br/><br/>
- **AVG():** 
    - This function returns the average value of a numeric column.
- **Syntax:**
`SELECT AVG(column_name) FROM table_name WHERE condition;`
- **Example:**
`SELECT AVG (Percentage) FROM student;`
<br/><br/>
- **SUM():** 
    - This function returns the total sum of a numeric column.
- **Syntax:**
`SELECT SUM(column_name) FROM table_name WHERE condition;`
- **Example:**
`SELECT SUM (Percentage) FROM student;`
<br/><br/>
- **MIN():** 
    - This function returns the smallest value of the selected column.
- **Syntax:**
`SELECT MIN(column_name) FROM table_name WHERE condition;`
- **Example:**
`SELECT MIN (Percentage) AS LeastMarks FROM student;`
<br/><br/>
- **MAX():** 
    - This function returns the largest value of the selected column.
- **Syntax:**
`SELECT MAX(column_name) FROM table_name WHERE condition;`
- **Example:**
`SELECT MAX (Percentage) AS HighestMarks FROM student;`

```python
run("SELECT COUNT(*) AS total_rows FROM student;")

print("\n⚠️ COUNT(*) counts ROWS; COUNT(column) skips NULLs:")
run("SELECT COUNT(*) AS all_rows, COUNT(Percentage) AS with_percentage FROM student;")

print("\nSUM, AVG, MIN, MAX - all ignore NULL:")
run("""
SELECT
    COUNT(Percentage)          AS n,
    ROUND(AVG(Percentage), 2)  AS average,
    MIN(Percentage)            AS lowest,
    MAX(Percentage)            AS highest,
    ROUND(SUM(Percentage), 1)  AS total
FROM student;
""")

print("\naggregates ignore NULL, so AVG is over 6 values, not 7:")
run("SELECT COUNT(*) AS rows_total, COUNT(Percentage) AS non_null FROM student;")

print("\n⚠️ You cannot use an aggregate in WHERE - it is evaluated too early:")
try:
    run("SELECT FName FROM student WHERE Percentage > AVG(Percentage);")
except sqlite3.OperationalError as exc:
    print("  ", exc)

print("\nuse a subquery instead:")
run("SELECT FName, Percentage FROM student "
    "WHERE Percentage > (SELECT AVG(Percentage) FROM student) "
    "ORDER BY Percentage DESC;")
```

### GROUP BY, HAVING, ORDER BY
- These keywords (GROUP BY, HAVING, ORDER BY) are used in a query to increase the functionality.
<br/><br/>
- **GROUP BY:** 
    - This functionality is used to arrange a similar type of data into a group. 
    - For instance, if the column in a table consists of similar data or value in different rows then we can use GROUP BY function to group the data.
- **Syntax:**
`SELECT column_name(s) FROM table_name WHERE condition GROUP BY column_name(s);`
- **ExampleL**
`SELECT COUNT(StudentID), Fname FROM Student GROUP BY Fname;`
<br/><br/>
- **HAVING:** 
    - This clause is used to place conditions where we need to decide which group will be the part of final result-set. 
    - Also, we can not use the aggregate functions like SUM(), COUNT() etc. with WHERE clause. At such situation, we have to use HAVING condition.
- **Syntax:**
`SELECT column_name(s) FROM table_name WHERE condition GROUP BY column_name(s) HAVING condition;`
- **Example:**
`SELECT Fname, SUM(Marks) FROM Student GROUP BY Fname HAVING SUM(Marks)>500;`
<br/><br/>
- **ORDER BY:** 
    - This keyword is used to sort the result-set in ascending or descending order. 
    - The ORDER BY keyword shall sort the records in ascending order by default. 
    - If we want to sort the records in descending order, use the DESC keyword.
- **Syntax:**
`SELECT column1, column2, ... FROM table_name ORDER BY column1, column2, ... ASC|DESC;`
- **Example:**
`SELECT COUNT(StudentID), City FROM Student GROUP BY City ORDER BY COUNT(StudentID) DESC;`

### JOINs

Everything above queries a single table. **Joins** are what make a *relational* database
relational: they combine rows from two tables using a shared value.

Splitting data across tables and joining it back is called **normalisation** — you store each
fact once, and use keys to connect them. Without joins you would have to repeat the student's
name on every marks row.

| Join | Returns |
|---|---|
| `INNER JOIN` | Only rows with a match in **both** tables |
| `LEFT JOIN` | **All** left rows; NULLs where the right side has no match |
| `RIGHT JOIN` | All right rows (SQLite 3.39+; emulate by swapping the tables) |
| `FULL OUTER JOIN` | All rows from both sides (SQLite 3.39+) |
| `CROSS JOIN` | Every combination — usually a mistake |
| Self join | A table joined to itself |

**The most common bug:** using `INNER JOIN` when you meant `LEFT JOIN`, and silently losing
every row that has no match. "Students with no marks" disappear entirely.

```python
# A second table, linked to `student` by Roll
run("""
CREATE TABLE marks (
    id      INTEGER PRIMARY KEY,
    Roll    INTEGER NOT NULL REFERENCES student(Roll),
    Subject TEXT    NOT NULL,
    Score   INTEGER NOT NULL
);
""")

run("""
INSERT INTO marks (Roll, Subject, Score) VALUES
    (102, 'Maths',   72), (102, 'Physics', 65),
    (103, 'Maths',   95), (103, 'Physics', 88),
    (104, 'Maths',   80),
    (105, 'Physics', 90), (105, 'Maths',   87);
    -- note: 106, 107 and 109 have NO marks
""")

print("INNER JOIN - only students who have marks:")
run("""
SELECT s.FName, m.Subject, m.Score
FROM student s
INNER JOIN marks m ON s.Roll = m.Roll
ORDER BY s.FName, m.Subject;
""")

print("\nLEFT JOIN - every student, with NULLs where there are no marks:")
run("""
SELECT s.FName, m.Subject, m.Score
FROM student s
LEFT JOIN marks m ON s.Roll = m.Roll
ORDER BY s.FName;
""")

print("\n^ Vikram, Neha and Meera appear only in the LEFT JOIN.")
print("  That difference is the single most common join bug.")
```

```python
print("finding the rows with NO match - the LEFT JOIN + IS NULL idiom:")
run("""
SELECT s.Roll, s.FName
FROM student s
LEFT JOIN marks m ON s.Roll = m.Roll
WHERE m.Roll IS NULL;
""")

print("\njoin + aggregate: average score per student:")
run("""
SELECT s.FName,
       COUNT(m.id)          AS subjects,
       ROUND(AVG(m.Score),1) AS avg_score
FROM student s
LEFT JOIN marks m ON s.Roll = m.Roll
GROUP BY s.Roll, s.FName
ORDER BY avg_score DESC NULLS LAST;
""")

print("\nCROSS JOIN - every combination. 8 students x 2 subjects:")
run("""
SELECT COUNT(*) AS combinations
FROM student
CROSS JOIN (SELECT DISTINCT Subject FROM marks);
""")

print("\n⚠️ A missing ON clause silently becomes a CROSS JOIN in some dialects -")
print("   which is how a 2-row query turns into a 2-million-row one.")
```

```python
print("GROUP BY - one output row per group:")
run("""
SELECT Department,
       COUNT(*)                  AS students,
       ROUND(AVG(Percentage), 1) AS avg_pct
FROM student
GROUP BY Department;
""")

print("\nGROUP BY on several columns:")
run("""
SELECT Department, City, COUNT(*) AS n
FROM student
GROUP BY Department, City
ORDER BY Department, n DESC;
""")

print("\nHAVING filters GROUPS (WHERE filters ROWS):")
run("""
SELECT City, COUNT(*) AS students
FROM student
WHERE City IS NOT NULL
GROUP BY City
HAVING COUNT(*) > 1;
""")

print("""
WHERE vs HAVING - the distinction that matters:

    WHERE   runs BEFORE grouping, on individual rows,   cannot use aggregates
    HAVING  runs AFTER  grouping, on the groups,        is FOR aggregates

    SELECT city, COUNT(*)
    FROM student
    WHERE percentage IS NOT NULL     <- drops rows before grouping
    GROUP BY city
    HAVING COUNT(*) > 1              <- drops groups after grouping
""")
```

### NULL VALUES
- In SQL, NULL term to represent a missing value. 
- A NULL value in a table is a value appears to be blank.
- A field with a NULL value is a field with no value in SQL.
    - Keep note that a NULL value is different than a zero value or a field that contains spaces.
- To check the null value we are not supposed to use the operators such as <, >, = etc. It’s not supported in SQL. We have special keywords i.e, **IS NULL** and **IS NOT NULL**.
<br/><br/>
- **IS NULL Syntax:**
`SELECT column_names FROM table_name WHERE column_name IS NULL;`
- **Example:**
`Select Fname, Lname From Student Where Marks IS NULL;`
<br/><br/>
- **IS NOT NULL Syntax:**
`SELECT column_names FROM table_name WHERE column_name IS NOT NULL;`
- **Example:**
`Select Fname, Lname From Student Where Marks IS NOT NULL;`

```python
print("rows with a NULL city:")
run("SELECT Roll, FName, City FROM student WHERE City IS NULL;")

print("\n🔴 `= NULL` matches NOTHING - NULL is not a value, it is 'unknown':")
run("SELECT COUNT(*) AS matched FROM student WHERE City = NULL;")
run("SELECT COUNT(*) AS matched FROM student WHERE City IS NULL;")

print("\nNULL propagates through arithmetic and concatenation:")
run("SELECT FName, Percentage, Percentage + 10 AS bumped FROM student WHERE Percentage IS NULL;")

print("\nCOALESCE supplies a fallback:")
run("SELECT FName, COALESCE(City, 'unknown') AS city, "
    "COALESCE(Percentage, 0) AS pct FROM student ORDER BY Roll;")

print("\nIFNULL is the two-argument SQLite/MySQL spelling; NVL in Oracle:")
run("SELECT FName, IFNULL(City, '-') AS city FROM student LIMIT 3;")

print("""
The rule: NULL means UNKNOWN, so any comparison with it is unknown too.
    NULL = NULL     -> NULL (not true!)
    NULL <> 5       -> NULL
    Use IS NULL / IS NOT NULL, and COALESCE to substitute a default.
""")
```

### UPDATE and DELETE
- **UPDATE:** 
    - The Update command is used to modify rows in a table.
    - The update command can be used to update a single field or multiple fields at the same time.
- **Syntax:**
`UPDATE table_name SET column1 = value1, column2 = value2,... WHERE condition;`
- **Example:**
`UPDATE student SET address='Delhi', Age=15 WHERE Roll=107;`
<br/><br/>
- **DELETE:** 
    - The SQL DELETE command is used to delete rows that are no longer required from the database tables.
    - It deletes the whole row from the table.
- **Syntax:**
`DELETE FROM table_name WHERE condition;`
- **Example:**
`DELETE FROM student WHERE Roll=107;`
<br/><br/>
- There is a special case here, if we need to delete the entire table records then we have to specify the table name.
- The data of that particular table will be divided.
- **Example:**
`Delete From Student;`
<br/><br/>
- **What is the difference between DELETE and TRUNCATE command?** 
- DELETE is a DML command whereas TRUNCATE is DDL command.
- Also DELETE deletes records one by one and makes an entry for each and every deletion in the transaction log, whereas TRUNCATE de-allocates pages and makes an entry for deallocation of pages in the transaction log.

```python
print("🔴 ALWAYS run the SELECT first, to see what you are about to change:")
run("SELECT Roll, FName, Percentage FROM student WHERE Roll = 106;")

run("UPDATE student SET Percentage = 60.0 WHERE Roll = 106;")
run("SELECT Roll, FName, Percentage FROM student WHERE Roll = 106;")

print("\nupdating several columns at once:")
run("UPDATE student SET City = 'Bengaluru', Department = 'Science' WHERE Roll = 107;")
run("SELECT Roll, FName, City, Department FROM student WHERE Roll = 107;")

print("\n🔴 An UPDATE with no WHERE hits EVERY row. On a copy, to prove it:")
run("CREATE TABLE danger AS SELECT * FROM student;")
run("UPDATE danger SET Department = 'OOPS';")
run("SELECT COUNT(*) AS rows_changed FROM danger WHERE Department = 'OOPS';")
run("DROP TABLE danger;")

print("\nDELETE with a WHERE:")
run("SELECT COUNT(*) AS before FROM student;")
run("DELETE FROM student WHERE Roll = 109;")
run("SELECT COUNT(*) AS after FROM student;")

print("""
Two habits that prevent most data-loss incidents:

  1. Write it as SELECT first. Only when the row set is right, change the
     verb to UPDATE or DELETE.
  2. Wrap it in a transaction you can roll back (see 10.3).
""")

conn.close()
print("\nin-memory database closed - everything in it is gone")
```

## Common Mistakes & Pitfalls

1. **`SELECT *` in production code.** It breaks the moment a column is added or reordered. Name the columns you need.
2. **`UPDATE` or `DELETE` without a `WHERE`.** It silently affects every row. Run the equivalent `SELECT` first, always.
3. **Comparing to NULL with `=`.** `WHERE city = NULL` matches nothing. Use `IS NULL`.
4. **Expecting aggregates to ignore NULL the way you expect.** `COUNT(col)` skips NULLs; `COUNT(*)` does not. `AVG` ignores them, which may or may not be what you want.
5. **Confusing `WHERE` and `HAVING`.** `WHERE` filters rows *before* grouping; `HAVING` filters groups *after*.
6. **Assuming row order without `ORDER BY`.** SQL result order is undefined unless you ask.
7. **`TRUNCATE` when you meant `DELETE`.** `TRUNCATE` removes every row and cannot be filtered or, usually, rolled back.
8. **Writing dialect-specific SQL and assuming it is portable.** `AUTO_INCREMENT` (MySQL), `AUTOINCREMENT` (SQLite) and `SERIAL` (PostgreSQL) all do the same job differently.
9. **Building queries by string concatenation.** That is SQL injection — see **10.3**.

## Best Practices

- Name your columns explicitly rather than using `SELECT *`.
- Write the `SELECT` before the `UPDATE`/`DELETE` to see exactly which rows you will hit.
- Use `IS NULL` / `IS NOT NULL`, never `= NULL`.
- Use table aliases in joins (`FROM student s JOIN marks m`) to keep queries readable.
- Put SQL keywords in UPPERCASE and identifiers in lowercase — the near-universal convention.
- Add `LIMIT` while exploring, so a mistake returns 10 rows and not 10 million.
- Prefer explicit `JOIN ... ON` over comma-joins with a `WHERE` — it separates the join condition from the filter.
- Learn the standard first, then the dialect. Most of SQL is portable; the edges are not.
