## Static Vs Dynamic Programming Language:
- **Static programming language** are those languages in which we can't write a code for performing logical operations using variables etc. Eg HTML etc.
- **Dynamic programming language** are those languages in which we can write a code for performing logical operations using variables etc. Eg. C, C++, Python, Java etc.

## Statically typed Vs Dynamically typed Programming Language:

This is about **when types are checked**, not about how strict they are.

- **Statically typed programming languages** are those in which
    - 1. variables **must be declared before they're used**, and the declaration fixes the type.
        - Variables may be initialised anywhere, but the compiler has to know the type before the variable is used.
    - 2. variable types are explicitly declared (or inferred at compile time) and are therefore determined **at compile time**. Ex. C, C++, Java, Rust, Go.
    - 3. type checking (i.e. the process of verifying and enforcing the constraints of types) is done at **compile time** (a static check).
        - <img src='./Images/1.1 Image f.png' width=20% height=20%/>
        - This lets the compiler decide whether a given variable can perform the actions requested from it or not.
    - 4. once a variable has been declared with a type, it cannot be assigned a value of a different type; doing so raises a type error **at compile time**, before the program ever runs.

- **Features of Static typed programming languages:**
    - A large class of errors are caught in the early stage of the development process.
    - Static typing usually results in compiled code that executes more quickly, because when the compiler knows the exact data types in use it can produce optimised machine code (i.e. faster and/or using less memory).
    - Editors can offer better autocomplete and refactoring, because they know what every name is.

```c
/* C code */
#include<stdio.h>
void sum()
{
int num, sum; // explicit declaration 
num = 5; // now use the variables 
sum = 10; 
sum = sum + num;
}
```

```java
/* Java code */
int data;
data = 50;
data = “Hello World!”; // causes an compilation error
```

- **Dynamically typed programming languages** are those in which
    - 1. variables **do not need to be declared before use** — assigning to a name creates it.
    - 2. they do not require the explicit declaration of the data type of variables. Ex. Python, JavaScript, PHP, Ruby.
    - 3. type checking is done at **run time** (a dynamic check).
        - <img src='./Images/1.1 Image g.png' width=10% height=10%/>
    - 4. variables are bound to objects at run time by means of assignment statements, and it is possible to bind the same variable to objects of different types during the execution of the program.

- **Features of Dynamic typed programming languages:**
    - Dynamic type checking typically results in less optimised code than static type checking.
    - It also includes the possibility of run-time type errors, and forces run-time checks to occur on every execution of the program (instead of just once at compile time).
    - Implementations generally associate each run-time object with a type tag (a reference to a type) holding its type information. This run-time type information (RTTI) also enables dynamic dispatch, late binding, down-casting and reflection.
    - The absence of a separate compilation step means you don't have to wait for a compiler to finish before testing a change. This makes the debug cycle much shorter and less cumbersome.

> **The trade-off in one line:** static typing catches more mistakes *before* you run;
> dynamic typing lets you run sooner. Python takes the second deal — and then, since 3.5,
> hands you **optional type hints** so you can opt back into the first one where it pays off.
> You'll meet those in **17 Type Hints and Static Typing**.

## Strong Typing Vs Weak Typing:
- **Strong Typing:** Programming languages in which variables have specific data types are **strong typed**. 
    - This implies that in strong typed languages, variables are necessarily bound to a particular data type.
    - Ex. Python, Java.
<br/><br/>    
- **Weak Typing:** Weak typed languages are those in which variables are not of a specific data type. 
    - It should be noted that this does not imply that variables do not have types; it does mean that variables are not "bound" to a specific data type. 
    - Ex. PHP and C.

```python
/* Python code */ 
foo = "x" 
foo = foo + 2 
>>>
Traceback (most recent call last): 
  File "<pyshell#3>", line 1, in ? 
    foo = foo + 2 
TypeError: cannot concatenate 'str' and 'int' objects 
>>>
```

```php
/* PHP code */ 
<?php 
$foo = "x"; 
$foo = $foo + 2; // not an error 
echo $foo; 
?>
```

### Summary: 
- Python is dynamic typed i.e, there is no type checking of the code prior to running it.
    - and strong typed i.e, there is no way to circumvent that type
- Java is static typed and strong typed. 
- PHP is dynamic typed and weak typed. 
- C is static typed and weak typed.

### Python is Interpreted
Many languages are compiled, meaning the source code you create needs to be translated into machine code, the language of your computer’s processor, before it can be run. Programs written in an interpreted language are passed straight to an interpreter that runs them directly.

This makes for a quicker development cycle because you just type in your code and run it, without the intermediate compilation step.

One potential downside to interpreted languages is execution speed. Programs that are compiled into the native language of the computer processor tend to run more quickly than interpreted programs. For some applications that are particularly computationally intensive, like graphics processing or intense number crunching, this can be limiting.

## Architecture of Python:
<img src='./Images/1.1 Image d.png' width=60% height=50%/>

- **Source file** contains high level instructions written by programmer. Extension of this file is `.py`.
<br/><br/>
- **Python Compiler** is the integral part of python interpreter and checks the syntax by compiling the program. 
    - If there is syntax error, the output of the program will be "SyntaxError".
    - If there is no syntax error it will generate a intermediate code i.e, `byte code` from source file. Extension of this file is `.pyc`.
    - Compiled code is run by python interpreter.
<br/><br/>        
- **Bytecode** is program code that has been compiled from source code into low-level code designed for a software interpreter. Byte code converted to machine code everytime when the program runs.
    - Bytecode is an implementation detail of the CPython interpreter.
    - .pyc files Python leaves lying around aren't just some "faster" or "optimized" version of our source code; they're the bytecode instructions that will be executed by Python's virtual machine as our program runs.
    - Prevents Python from having to re-parse the source code every time the program runs.
    - The byte-code is loaded into the Python run-time and interpreted by a virtual machine, which is a piece of code that reads each instruction in the byte-code and executes whatever operation is indicated. 
<br/><br/>
- **PVM (Python Virtual Machine)** is the integral part of python interpreter and it generate low level code form python byte code in "on the fly mode" i.e, saved no where. Machine code is directly handover to CPU.
    - Steps in PVM:
        - **Step 1) Laxing:** It is responsible for dividing program into tokens. (Tokens are smallest unit of a program)
        - **Step 2) Parsing:** Tree(Bunch of nodes) formation on which token is placed at each node.
        Parent node and child node
        - **Step 3) Compiler:** Convert high level language to low level language. In python convert unicode code to binary code.
        - **Step 4) Execution:** Operation is performed and output is generated in binary form which is again converted to unicode code and final output on output screen.
<br/><br/>
- **NOTE: After completing the execution of program interpreter delete the .pyc file. Machine code is not saved in interpreter but we can save it by using it as a module**

## Python versions: what changed, and what to target

Python 2 reached **end of life on 1 January 2020**. It is gone; you will not write it.
The useful question today is not "2 or 3?" but **"which 3.x, and what does it give me?"**

### Release cadence
- A new minor version (3.12 -> 3.13 -> 3.14) ships **every October**.
- Each release gets **2 years** of full support, then **3 more** of security fixes — about 5 years in total.
- Version numbering is `3.MINOR.PATCH`; the *minor* number is the one that matters for features.

### What arrived when

| Version | Year | Headline features you'll actually use |

## The REPL: Python's interactive console

**REPL** stands for **Read-Evaluate-Print-Loop**:

1. **Read** whatever you type,
2. **Evaluate** it,
3. **Print** the result,
4. **Loop** back to the beginning.

Type `python` (or `py`) in a terminal with no filename and you land in it — the `>>>` prompt.
It is the fastest way to answer "what does this do?" without creating a file.

```pycon
>>> 2 + 3
5
>>> name = "Shiva"
>>> name.upper()
'SHIVA'
>>> _ + "!"          # `_` holds the last printed value (REPL only)
'SHIVA!'
>>> exit()           # or Ctrl-Z then Enter on Windows, Ctrl-D on macOS/Linux
```

Two things worth knowing:
- The REPL **echoes the value of an expression automatically** — you do not need `print()`.
  In a `.py` file you do. This trips up nearly everyone once.
- The special `_` variable holds the most recently printed value. **This works only in the
  REPL**, not in scripts.

> **Version note (3.13+):** Python 3.13 shipped a completely rewritten interactive REPL with
> colour, multi-line editing, block-aware history, and `help` / `exit` that work without
> parentheses. If your REPL looks plain, you are on 3.12 or earlier — everything above still works.

## Common Mistakes & Pitfalls

1. **Confusing *dynamic* with *weak* typing.** Python is dynamically typed (types checked at run time) but *strongly* typed (`"x" + 2` is an error, not a silent coercion). These are independent axes — PHP is dynamic *and* weak; Java is static *and* strong.
2. **Forgetting to tick "Add python.exe to PATH"**, then finding `python` is not a recognised command. Re-run the installer and choose Modify.
3. **Building on the system Python** on macOS/Linux. It belongs to the OS, and installing packages into it can break system tools. Use your own install plus a virtual environment.
4. **Expecting the REPL's auto-echo inside a `.py` file.** In the REPL, `2 + 3` prints `5`. In a script it computes `5` and throws it away — you need `print(2 + 3)`.
5. **Following a Python 2 tutorial.** If you see `print "hello"` or `raw_input()`, close the tab.

## Best Practices

- Target the **current stable release** for new work; target the **oldest version you must support** in libraries.
- Pin your Python version per project (`.python-version`, or `requires-python` in `pyproject.toml`).
- Use a **virtual environment per project** — never install project dependencies globally.
- Read the *last* line of a traceback first: it names the actual error.
- Keep <https://docs.python.org/3/whatsnew/> bookmarked — it is the definitive record of what changed.
