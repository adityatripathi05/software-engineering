## Method to get output
- **syntax:** ```print(value1,value2, ..., sep=' ', end='\n')```
    - Parameter **'sep'** for seperating values. By default the value of this parameter is ```' '``` i.e, space
    - Parameter **'end'**. By default the value of this parameter is '\n' i.e, the new line character.

````python
print('Ducat', "India")
````

````python
print("India", 'Ducat', sep='\n')
````

````python
print("India", 'Ducat', sep='\n')
print("Python")
print("India", 'Ducat', end='_')
print('Python')
````

### The four parameters of `print()`

```
print(*values, sep=' ', end='\n', file=sys.stdout, flush=False)
```

| Parameter | Default | What it does |
|---|---|---|
| `*values` | — | Any number of values, of any type |
| `sep` | `' '` | Placed **between** values |
| `end` | `'\n'` | Placed **after** the final value |
| `file` | `sys.stdout` | Where to write — use `sys.stderr` for errors |
| `flush` | `False` | Force immediate output instead of buffering |

`file` and `flush` matter the moment your output is piped somewhere: error messages
belong on `stderr` so they can be separated from real output, and progress indicators
need `flush=True` or they appear all at once at the end.

````python
import sys

# print() has four keyword parameters
print("a", "b", "c")                        # default sep=' ', end='\n'
print("a", "b", "c", sep="-")               # sep: what goes BETWEEN values
print("a", "b", "c", sep="")                # glue them together
print("no newline", end=" -> ")             # end: what goes AFTER the last value
print("continued on the same line")

# sep can be multi-character, including a newline
print("first", "second", "third", sep="\n  ")

# file: where the output goes (default sys.stdout)
print("this is an error message", file=sys.stderr)

# flush: force the output out immediately instead of buffering it
print("processing...", end="", flush=True)
print(" done")
````

## Formatting output

`print()` puts values on the screen; **formatting** decides what they look like when
they get there. Python has three styles, and only one of them is worth writing today.

### Syntax breakdown

```
f"{ expression : format_spec }"
  |      |            |
  |      |            +-- how to render it:  .2f  ,  >10  05d  %
  |      +--------------- any Python expression
  +---------------------- the f prefix makes it an f-string
```

````python
name = "Shiva"
age = 23
pi = 3.14159265
total = 1234567.8915

# 1. f-strings (3.6+) - the modern default. Expressions go inside the braces.
print(f"{name} is {age} years old, and next year will be {age + 1}.")

# 2. Format specifications: {value:spec}
print(f"Pi to 2dp      : {pi:.2f}")
print(f"Pi to 4dp      : {pi:.4f}")
print(f"With separators: {total:,.2f}")
print(f"Percentage     : {0.8734:.1%}")
print(f"Padded number  : {42:05d}")

# 3. Alignment: < left, > right, ^ centre
print(f"|{name:<12}|{name:^12}|{name:>12}|")

# 4. Self-documenting expressions (3.8+) - excellent for debugging
print(f"{age=}")
print(f"{age * 2 =}")

# 5. !r shows the repr() - quotes visible, invaluable for spotting stray whitespace
value = " Shiva "
print(f"str : {value}")
print(f"repr: {value!r}")
````

#### A practical formatting example

Format specs let you build aligned output without any manual padding arithmetic:

````python
products = [
    ("Keyboard", 2, 1299.5),
    ("Monitor", 1, 12499.0),
    ("USB-C cable", 3, 349.99),
]

print(f"{'Item':<15}{'Qty':>5}{'Price':>12}{'Total':>12}")
print("-" * 44)

grand_total = 0
for item, qty, price in products:
    line_total = qty * price
    grand_total += line_total
    print(f"{item:<15}{qty:>5}{price:>12,.2f}{line_total:>12,.2f}")

print("-" * 44)
print(f"{'Grand total':<15}{'':>5}{'':>12}{grand_total:>12,.2f}")
````

#### The two older styles — recognise them, don't write them

You will meet these in existing code and older tutorials:

````python
name, age = "Shiva", 23

# str.format() - pre-3.6 standard. Still useful when the template is stored
# separately from the values (e.g. loaded from a config file or a translation table).
print("{} is {} years old".format(name, age))
print("{0} is {1}; {0} again".format(name, age))
print("{n} is {a}".format(n=name, a=age))

TEMPLATE = "Dear {title} {surname}, your order {order_id} has shipped."
print(TEMPLATE.format(title="Dr", surname="Rao", order_id="A-1187"))

# %-formatting - the oldest style, inherited from C. LEGACY: do not write new code with it.
print("%s is %d years old" % (name, age))
print("Pi is %.2f" % 3.14159)
````

**Which to use:** f-strings, essentially always. They are the fastest, the shortest, and
the only one where the value sits visually where it is printed.

Use `.format()` only when the template string must exist separately from its values.
Never write `%` formatting in new code.

**Real-world use case:** the `:,.2f` and alignment specs above are what turn a raw data
dump into a readable report — invoices, CLI tables, log lines. Learning them once saves
writing padding logic forever.

> **NOTE on conversions:** `float(input(...))` happily accepts `7` and stores `7.0`,
> because widening `int` -> `float` loses nothing.
>
> The reverse direction *is* allowed but **truncates**: `int(7.9)` is `7`, not `8`.
> And converting a number to text always works — `str(7)` gives `"7"`. What fails is
> converting text that is not numeric: `int("abc")` raises `ValueError`.

### Validating input properly

`int(input(...))` is fine in a tutorial and unacceptable in a real program: the moment a
user types `twenty` instead of `20`, the program dies with an unhandled `ValueError`.

Real code asks again.

````python
# The fragile version: one stray letter and the program crashes
# age = int(input("Enter your age: "))     # ValueError if the user types "abc"

# The robust version: keep asking until the input is actually valid
while True:
    raw = input("Enter your age: ").strip()
    try:
        age = int(raw)
    except ValueError:
        print(f"  {raw!r} is not a whole number. Try again.")
        continue

    if not 0 < age < 130:
        print("  Age must be between 1 and 129. Try again.")
        continue

    break

print(f"Thank you. Next year you will be {age + 1}.")
````

The shape of that loop is worth memorising, because it recurs constantly:

```
while True:
    read the input
    try to convert it        -> on failure: explain, continue
    check it makes sense     -> on failure: explain, continue
    break
```

Note `.strip()` on the raw text. Users type trailing spaces, and `int(" 23 ")` happens to
tolerate them but `" yes " == "yes"` does not — so strip early and stop thinking about it.

**Real-world use case:** this is the entire basis of CLI tools, setup wizards and any
interactive script. The same try/validate/retry pattern reappears when you validate web
form data, config files and API payloads — the source changes, the shape does not.

### `eval()` — powerful, and a genuine security hole

`eval()` takes a **string and executes it as Python code**, so `eval(input())` will
evaluate the datatype of whatever the user typed — `5` becomes an `int`, `[1,2]` becomes
a `list`, `'hi'` (in quotes) becomes a `str`.

> ### ⚠️ Never use `eval(input())` on input you do not control.
>
> It does not just *parse* the text — it **runs** it, with your program's full privileges.
> A user who types
> ```python
> __import__('os').system('rm -rf ~')
> ```
> has just deleted the home directory. Typing `__import__('os').listdir('.')` lists your
> files; anything Python can do, that string can do.
>
> This is one of the most common security holes in beginner code, and it appears in a
> depressing number of tutorials. It is shown here so you **recognise and avoid** it.

**What to use instead:**

| You want | Use |
|---|---|
| A number | `int(input())` / `float(input())` |
| A literal list, dict, tuple, string, number | `ast.literal_eval()` — parses literals only, executes nothing |
| Structured data from a file or API | `json.loads()` |
| A choice from a fixed set | Compare against a list/dict of allowed values |

````python
import ast

# The old, dangerous way — shown for recognition only. DO NOT SHIP THIS.
# data = eval(input("Enter anything: "))

# Safe equivalent: literal_eval parses literals and refuses to execute code.
text = input("Enter a number, list or quoted string: ")

try:
    data = ast.literal_eval(text)
    print(f"{data!r} belongs to {type(data)}")
except (ValueError, SyntaxError):
    print(f"Not a valid Python literal - treating it as text: {text!r}")

# Proof of the difference:
print("\nliteral_eval on a literal :", ast.literal_eval("[1, 2, 3]"))
try:
    ast.literal_eval("__import__('os').getcwd()")   # eval() would RUN this
except ValueError as exc:
    print("literal_eval on code     : refused ->", exc)
````

- **Multiple runtime input:**
    - **split():** This method returns a list of strings after breaking the given string by the specified separator. 
        - **Syntax:** ```str.split(separator, maxsplit)```

````python
name, branch, year = input('Enter your name, branch, year seperated ny comma: ').split(',')
print(f'My name is {name} and my branch is {branch} and presently in {year} year.')
````

⚠️ The cell above breaks if the user types the wrong number of fields — `ValueError:
not enough values to unpack`. It also keeps the spaces around each value, so `branch`
becomes `" CSE"` rather than `"CSE"`.

A more defensive version:

````python
# Safer: check the field count before unpacking
raw = input("Enter name, branch, year separated by commas: ")
parts = [p.strip() for p in raw.split(",")]

if len(parts) != 3:
    print(f"Expected 3 values, got {len(parts)}. Please try again.")
else:
    name, branch, year = parts
    print(f"My name is {name}, my branch is {branch}, and I am in year {year}.")
````

## Common Mistakes & Pitfalls

1. **Forgetting `input()` always returns a `str`.** `input("Age: ")` then `age + 1` raises `TypeError`. Convert explicitly: `int(input(...))`.
2. **Wrapping `input()` in `int()` with no error handling.** One stray letter and the program dies with `ValueError`. Real programs loop until the input is valid.
3. **Using `eval(input())`.** It executes whatever the user types, as code. See the warning above.
4. **Unpacking a `.split()` into the wrong number of names.** `a, b = input().split(',')` raises `ValueError` if the user types one value or three.
5. **Forgetting `.strip()`.** Users type trailing spaces. `" yes " == "yes"` is `False`.
6. **Assuming `print()` flushes immediately.** When output is piped to a file it is buffered; pass `flush=True` for live progress output.

## Best Practices

- Use **f-strings** for all new formatting. Reach for `.format()` only when the template is stored separately from the values; never write `%` formatting in new code.
- Always `.strip()` text you got from a human.
- Validate in a loop with `try` / `except ValueError` rather than trusting the first attempt.
- Use format specs (`:.2f`, `:,`, `:>10`) instead of manually rounding and padding.
- Use `f"{value=}"` while debugging — it prints the expression *and* its value.
- Write user-facing prompts that state the expected format: `"Enter date (YYYY-MM-DD): "`.
