## Comment Statement in python
- Comments in the program are written to make the program more readable. These lines are ignored by the interpreter.
    - Single-line comment in Python start with the hash character `#` and are automatically terminated by the end of line.
    - Multiline comment in python starts and ends with either `'''`(triple single quotes) or `"""`(triple double quotes).

````python
# comment in single line
'''
for multiline
comment use
triple quotes
'''
````

> **Correction worth internalising:** a triple-quoted string is **not a comment.**
>
> Python has exactly one comment syntax: `#`. A `'''...'''` block on its own line is a
> *string expression* that gets created and immediately discarded. In practice it is
> harmless — but two things follow:
>
> 1. It costs a tiny amount of work at run time; `#` costs literally nothing.
> 2. As the **first statement** in a module, function or class, that string is not discarded
>    at all — it becomes the **docstring**, retrievable via `help()` and `__doc__`.
>
> Use `#` to comment. Use `""".."""` to *document*.

```python
def greet(name):
    """Return a greeting for `name`."""     # <- docstring, not a comment
    # Build the string separately for readability   <- an actual comment
    return f"Hello, {name}!"
```

# Language Primitives
- **Lexical analysis:** A Python program is read by a parser. Input to the parser is a stream of tokens, generated by the lexical analyzer i.e, lexical analyzer breaks a file into tokens.

## Tokens:
- Token is the smallest unit inside the given program.

## Python Keywords

- Keywords are **case sensitive**. 
    - Mostly Keywords contain lowercase letters only except True, False, None.
    - Some keywords begin simple statements or clauses of compound statements, while other keywords are operators.
- They represent the syntax and structure of a Python program. We cannot use keywords as regular identifiers.

- **‘keywords’** is a Python builtin module.
    - For now, a module is simply a python script having a collection of related attributes. We’ll learn about modules in detail later in this tutorial.
- **‘kwlist’** is one of the attributes of the keyword module which returns a list of all Python Keywords.
- **iskeyword()** is an function of keyword module. It test whether a keyword is valid or not.
<br/><br/>
- **print()** is a builtins module function. Prints the values to a output by default.
- **len()** is a builtins module function. It returns the length of a sequence.

### Hard keywords vs soft keywords

> **Version note (3.10+):** Python gained **soft keywords**. These are words that carry
> special meaning *only in one specific position* — everywhere else they behave as ordinary
> names. `match`, `case`, `type` and `_` are soft keywords.

This is why adding `match`/`case` in Python 3.10 did **not** break the thousands of programs
that already used `match` as a variable (a `re` match object, most commonly).

- **Hard keyword** (`if`, `for`, `class`, `None`...) — reserved. `for = 5` is a `SyntaxError`.
- **Soft keyword** (`match`, `case`, `type`, `_`) — contextual. `match = 5` is perfectly fine.

Generate both lists live, so this cell is never out of date:

````python
import keyword

# Hard keywords: reserved, can never be used as names
print(f"Hard keywords ({len(keyword.kwlist)}):")
print(keyword.kwlist)

# Soft keywords (3.10+): special ONLY in specific syntax, otherwise ordinary names
print(f"\nSoft keywords ({len(keyword.softkwlist)}):")
print(keyword.softkwlist)

print("\nIs 'match' a hard keyword?", keyword.iskeyword("match"))
print("Is 'match' a soft keyword?  ", keyword.issoftkeyword("match"))

# Proof: `match` is still perfectly usable as a variable name
match = "this is legal"
print(match)
````

## Python Identifiers
- Python Identifiers are user-defined names to represent a variable, function, class, module or any other object.. 
- It helps in differentiating one entity from another.
<br/><br/>
- **Python Identifier writing rules:**
    - An identifier starts with a letter (A to Z or a to z) or an underscore (_) followed by zero or more letters, underscores, and digits (0 to 9). 
        - For example, myvar, student_Class, employee0, _var1, var_100 are valid identifiers.
    - An identifier cannot be started with a digit character. 
        - For example, myvar9 is valid identifier while 9myvar is not.
    - Case is significant in Python: lowercase and uppercase letters are distinct.
    - Python keywords are not allowed to be used as identifiers.
    - Python does not allow punctuation characters such as '@', '$', and '%' within identifiers.
    - Normal Python style is to start class names with an uppercase letter and all other identifiers with a lowercase letter.
    - Starting an identifier with a single leading underscore indicates by convention that the identifier is meant to be private. 
    - Starting an identifier with two leading underscores indicates a strongly private identifier; if the identifier also ends with two trailing underscores, the identifier is a language-defined special name.
<br/><br/>  
- **isidentifier():** Test whether a Python identifier is valid or not.

````python
print('_var1'.isidentifier())
print('9func'.isidentifier())
````

### Naming conventions (PEP 8)

Python does not *enforce* naming style — but every Python codebase you will ever join
follows **PEP 8**, the official style guide. Matching it is how your code stops looking
like a beginner's.

| Thing | Convention | Example |
|---|---|---|
| Variable, function, method | `snake_case` | `student_name`, `calculate_total()` |
| Constant | `UPPER_SNAKE_CASE` | `MAX_RETRIES`, `PI` |
| Class | `PascalCase` | `StudentRecord`, `BankAccount` |
| Module / package | short `lowercase` | `utils`, `models` |
| "Internal, don't touch" | leading underscore | `_cache`, `_helper()` |
| Deliberately unused | single underscore | `for _ in range(3):` |

A few hard rules that go with them:
- **4 spaces** per indentation level — never tabs (Python 3 raises `TabError` if you mix them).
- Names are **case-sensitive**: `age`, `Age` and `AGE` are three different variables.
- Never shadow a builtin (`list`, `str`, `id`, `type`, `sum`, `input`, `max`). It works, and
  then quietly breaks the builtin for the rest of your scope.

> **Tip:** don't memorise PEP 8 — install a formatter (`black` or `ruff format`) and a linter
> (`ruff`) and let the tools enforce it. Covered in **18 Tooling, Packaging and Environments**.

````python
# PEP 8 in practice
MAX_RETRIES = 3                 # constant: UPPER_SNAKE_CASE (a convention, not enforced)
student_name = "Shiva"          # variable: snake_case
total_marks_scored = 452        # descriptive beats short

def calculate_percentage(scored, total):   # function: snake_case, a verb phrase
    return scored / total * 100

class StudentRecord:            # class: PascalCase, a noun
    pass

print(calculate_percentage(total_marks_scored, 500))

# `_` marks a value you are deliberately ignoring
for _ in range(3):
    print("tick")

# Shadowing a builtin: legal, and a trap
list = [1, 2, 3]                # `list` the type is now hidden
print(list)
# print(list("abc"))            # TypeError: 'list' object is not callable
del list                        # undo the damage
print(list("abc"))              # the builtin is back
````

## Python Literals
- Literals can be defined as a data that is given in a variable or constant.
- Python support the following literals:
    - **1. String literals:** A string literal is a sequence of characters surrounded by quotes. Eg. 'xyz' or "xyz"
        - We can use both single, double or triple quotes for a string. 
        - And, a character literal is a single character surrounded by single or double quotes.
        <br/><br/>
    - **2. Numeric literals:** Numeric Literals are immutable(unchangeable). Eg. 2, 2.5, 2+5j
        - Numeric literals can belong to 3 different numerical types Integer, Float, and Complex.
        <br/><br/>
    - **3. Boolean literals:** A Boolean literal can have any of the two values: True or False.
        <br/><br/>
    - **4. Special literals:** Python contains one special literal i.e., None. 
        - None is used to specify to that field that is not created. 
        - It is also used for end of lists in Python.
        <br/><br/>
    - **5. Literal Collections:** There are four different literal collections List literals, Tuple literals, Dict literals, and Set literals.
        - Eg. (2, 32, 'xyz', 2.5),[2, 32, 'xyz', 2.5],{'Roll no':2, 'Age':32, 'name':'xyz', 2.5:'avg'}, {2, 32, 'xyz', 2.5}

### Modern literal forms worth knowing

Two conveniences that make numeric literals far more readable:

- **Underscores as digit separators** (3.6+) — `1_000_000` is exactly the same value as
  `1000000`, but you can read it at a glance. Python ignores the underscores entirely.
- **Base prefixes** — `0b` binary, `0o` octal, `0x` hex.

**Real-world use case:** underscore separators are near-universal in configuration and
finance code, where a misread `10000000` vs `100000000` is a real and expensive bug.

````python
# Underscores in numeric literals (3.6+): visual grouping, ignored by Python
population = 1_400_000_000
salary = 75_000
print(population, salary)
print(population == 1400000000)   # the underscores are purely cosmetic

# Integers can be written in four bases
decimal = 255
binary = 0b1111_1111     # 0b prefix
octal = 0o377            # 0o prefix
hexadecimal = 0xFF       # 0x prefix
print(decimal, binary, octal, hexadecimal)
print(decimal == binary == octal == hexadecimal)

# ...and converted back for display
print(f"{255:b}  {255:o}  {255:x}  {255:X}")

# Scientific notation and complex literals
print(1.5e3, 2.5e-4, 3 + 4j, type(3 + 4j))
````

## Variable or Reference Variable
- A variable is used to store value or data which can later be used in the program.
- A Python variable is an entity which stores the memory address of referenced value(object) stored inside memory.
    - Variable is created inside in stack memory
    - Referenced value is stored inside heap memory.
    - Similar to pointers in C, variables in Python refer to values (or objects) stored somewhere in memory.
- In python, we don't need to declare a variable before using it. Simply assign a value to the variable and it will exist.
- Python is a dynamically typed language hence we need not define the type of the variable while declaring it. The interpreter implicitly binds the value with its type.
<br/><br/>
- In fact, all variable names in Python are said to be references to the values, some of which are front loaded by Python and therefore exist before the name references occur. 
- Python keeps an internal counter on how many references an object has. Once the counter goes to zero - meaning that no reference is made to the object - the garbage collector in Python removes the object , thus freeing up the memory.

Hence, python allocate same memory location to variable with same value(Automatic memory allocation)

````python
var3= 10
print(var3)
print(id(var3))
````

- Both var2 and var3 refer to same memory location(i.e, same object) because python optimises the memory utilization by allocating the same object reference to a new variable, if object already exist with the same value.

> **Important caveat — this is a CPython optimisation, not a language rule.**
>
> Python does **not** guarantee that equal values share a memory location. What you are
> seeing is **small-integer caching**: CPython pre-creates the integers **-5 to 256** at
> startup and hands out the same object every time. Step outside that range and the
> behaviour changes.
>
> The same applies to short strings, which CPython may *intern*.
>
> **Never write logic that depends on this.** Use `==` to compare values; reserve `is` for
> `None`, `True` and `False`. Run the cell below to see the seam.

````python
# Small integers (-5 to 256) are pre-created and cached by CPython
a = 10
b = 10
print("a is b (small int):", a is b, "| same id:", id(a) == id(b))

# Outside that range, equal values are usually SEPARATE objects
x = 1000
y = 1000
print("x is y (large int):", x is y, "| same id:", id(x) == id(y))

# ...but equality still works perfectly, because `==` compares VALUES
print("x == y:", x == y)

# The same trap with strings: short/identifier-like literals get interned
s1 = "hello"
s2 = "hello"
print("s1 is s2:", s1 is s2)

s3 = "hello world!"
s4 = "hello world!"
print("s3 is s4:", s3 is s4)   # may be True or False - do not depend on it

# THE RULE: use `==` to compare values. Reserve `is` for None, True, False.
value = None
print("Correct None check:", value is None)
````

## Common Mistakes & Pitfalls

1. **Treating a triple-quoted string as a comment.** It is a *string expression* that gets built and thrown away. Harmless, but not free — and inside a function or class it silently becomes the docstring. Use `#` for real comments.
2. **Assuming `id()` values are stable or meaningful.** They are memory addresses for the current run only. Never compare them, never store them, never rely on small-int caching.
3. **Shadowing a builtin.** Naming a variable `list`, `str`, `id`, `type`, `sum` or `input` works — and then breaks the builtin for the rest of the cell. Add a trailing underscore (`list_`) or pick a better noun.
4. **Mixing tabs and spaces for indentation.** Python 3 rejects this outright with `TabError`. Configure your editor to insert 4 spaces.
5. **Believing a variable holds a value.** In Python a name is a *label bound to an object*. `b = a` copies the label, not the object — which matters enormously for lists and dicts (see **02 Datatypes**).

## Best Practices

- `snake_case` for variables and functions, `PascalCase` for classes, `UPPER_SNAKE_CASE` for constants (PEP 8).
- Names should say *what the thing is*: `student_count`, not `sc` or `n`.
- 4 spaces per indent level. Never tabs.
- Write comments that explain **why**, not **what** — the code already says what.
- Keep lines under ~79 characters (PEP 8) or ~88 (the Black formatter's default). Pick one and be consistent.
- Use `_` for a value you are deliberately ignoring: `for _ in range(3):`
