### What is Programming?
* Programming is the implementation of logic to facilitate specific computing operations and functionality.
* In simple terms, programming is a set of logically built Instructions.
* Programming can be done in one or more languages, which differ by application, domain and programming model.

## From idea to running program

A useful analogy: a **recipe**.

- The **recipe** written on paper is the *source code* — instructions in a language humans read.
- The **cook** is the *processor* — but this cook only understands one very terse language.
- Something has to **translate** the recipe into that terse language.

That translation step is the whole difference between the two big families of languages:

| | Compiled | Interpreted |
|---|---|---|
| **Analogy** | Translate the entire cookbook up front, hand over the translation | A translator stands beside the cook, converting line by line as they go |
| **When translation happens** | Once, before running (build step) | While running |
| **Errors surface** | Many at compile time, before you run | At run time, when that line is reached |
| **Speed to run** | Usually faster | Usually slower |
| **Speed to iterate** | Slower — rebuild after every change | Faster — just run it again |
| **Examples** | C, C++, Rust, Go | Python, JavaScript, Ruby |

Python is (mostly) **interpreted**, which is why you can type a line and see the result
immediately — no build step. That fast feedback loop is a large part of why Python is
such a good first language.

> The real picture is slightly muddier: Python first *compiles* your source to an
> intermediate form called **bytecode** (the `.pyc` files you'll see in `__pycache__/`),
> then interprets that. You'll see exactly how in **1.1 Introduction to Python**.

### What are Program Instruction and How to give Instructions?
* For a computer to know how to do anything, it must be provided instructions.
* Instructions are a segment of code containing steps that need to be executed by the processor. 
* At the lowest level, each instruction is a sequence of 0s and 1s that describes a physical operation the computer is to perform.
* These instructions are stored on the hard disk drive or another storage device and are not executed until they are run by the user.

### What an instruction looks like, at each level

The same idea, expressed at three levels of abstraction:

| Level | Looks like | Who writes it |
|---|---|---|
| **Machine code** | `10110000 01100001` | Nobody, by hand |
| **Assembly** | `MOV AL, 61h` | Systems programmers, rarely |
| **High-level (Python)** | `total = price + tax` | You |

Each level up trades a little control for a lot of readability. Python sits near the top:
one line of Python can stand in for dozens of machine instructions.

### Why do we have so many languages?
- Languages, which differ by application, domain and programming mode.
<br/><br/>
    - **FORTRAN:**   FORmula TRANslation is a general-purpose, compiled imperative programming language that is especially suited to **numeric computation and scientific computing**. Developed in 1957.
<br/><br/>
    - **COBOL:** COmmon Business Oriented Language is a compiled English-like computer programming language designed for **business use**. Designed in 1959.
<br/><br/>
    - **BASIC:** Beginners' All-purpose Symbolic Instruction Code is a family of general-purpose, high-level programming languages whose design philosophy emphasizes ease of use. Developed in the mid 1960s.
<br/><br/>
    - **C:** C is a procedural programming language, initially developed by Dennis Ritchie in 1972. It was mainly developed as a system programming language to write an **operating system**.
<br/><br/>
    - **C++:** C++ is a cross-platform language that can be used to create **high-performance applications**. 
        - C++ was developed by Bjarne Stroustrup, as an extension to the C language. C++ gives programmers a high level of control over system resources and memory. Developed in 1980. 
<br/><br/>
    - **ABC:** ABC is an interactive programming language and environment for personal computing, originally intended as a good replacement for BASIC.
<br/><br/>
    - **Python:** Python is an interpreted, high-level, **general-purpose** programming language.
        - Created by Guido van Rossum and first released in 1991, Python's design philosophy emphasizes code readability with its notable use of significant whitespace.
<br/><br/>
    - **JAVA:** Java is a set of technologies (programming language and computing platform) for creating and running software. Java released in 1995.

Enough theory. Run the cell below (`Shift + Enter`).

```python
print("Hello, world!")
```

That single line already involves everything discussed above:

1. You wrote an **instruction** in a high-level language.
2. Python **compiled** it to bytecode.
3. The interpreter **executed** that bytecode.
4. The processor pushed characters to your screen.

Now something with a little more going on — a variable, arithmetic, and formatted output.
Don't worry about the syntax yet; the point is that it reads almost like English.

```python
# A program is just instructions, executed top to bottom.
price = 250
tax_rate = 0.18

tax = price * tax_rate
total = price + tax

print(f"Price:  {price:>8.2f}")
print(f"Tax:    {tax:>8.2f}")
print(f"Total:  {total:>8.2f}")
```

Read that again and notice: no type declarations, no semicolons, no `main()` function,
no compile command. Five lines of setup in some languages is zero lines here.

That is the design philosophy you'll see throughout Python — **readability counts**.

## Common Mistakes & Pitfalls

1. **Expecting the computer to infer intent.** It executes exactly what you wrote, in order, every time. Most bugs are a mismatch between what you meant and what you said.
2. **Reading code without running it.** You cannot learn programming by reading alone — run every cell, then change a number and run it again.
3. **Skipping error messages.** A traceback is not a failure, it's the interpreter telling you precisely what confused it. Read the last line first.
