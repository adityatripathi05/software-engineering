#### ⚠️ The comma makes the tuple, not the parentheses

`(5)` is just the integer `5` wrapped in grouping parentheses. `(5,)` is a one-element
tuple. This catches everyone at least once, and the resulting bug is quiet — a stray
trailing comma turns a value into a tuple with no error at all.

````python
# THE most common tuple mistake
not_a_tuple = (5)          # parentheses here are just grouping
a_tuple = (5,)             # the COMMA is what makes a tuple

print("(5)  ->", not_a_tuple, type(not_a_tuple))
print("(5,) ->", a_tuple, type(a_tuple))

# Parentheses are optional; the comma is not
also_a_tuple = 5,
print("5,   ->", also_a_tuple, type(also_a_tuple))

# This is why a stray comma can silently change a value
value = 42,
print("\nvalue = 42,  ->", value, "  <- probably a typo, and a hard bug to spot")

# It bites hardest in function calls
print("\nlen((5))  would fail - int has no len()")
print("len((5,)) =", len((5,)))
````

### Unpacking

Unpacking is one of Python's most-used conveniences: assign several names from one sequence
in a single statement. It works on **any** iterable, not just tuples.

### Syntax breakdown

```
first, *middle, last = (1, 2, 3, 4, 5)
  |       |       |
  |       |       +-- takes the final item
  |       +---------- the star absorbs everything left over, as a LIST
  +------------------ takes the first item
```

Rules: **at most one** starred name, and the non-starred names must all be satisfiable.

````python
point = (3, 7)
x, y = point                     # basic unpacking
print(f"x={x}, y={y}")

# Swapping needs no temporary variable - the right side builds a tuple first
a, b = 1, 2
a, b = b, a
print(f"\nswapped: a={a}, b={b}")

# Starred unpacking (3.0+): one name absorbs "the rest", always as a LIST
first, *rest = (1, 2, 3, 4, 5)
print(f"\nfirst={first}, rest={rest}")

*most, last = (1, 2, 3, 4, 5)
print(f"most={most}, last={last}")

first, *middle, last = (1, 2, 3, 4, 5)
print(f"first={first}, middle={middle}, last={last}")

# Nested unpacking mirrors the structure
record = ("Aditya", (95.6, "CBSE"), 101)
name, (percentage, board), roll = record
print(f"\n{name}: {percentage}% ({board}), roll {roll}")

# Ignore what you do not need with _
_, _, year = ("15", "08", "1947")
print(f"\nyear: {year}")

# Unpacking in a for loop - this is why zip() and enumerate() feel natural
pairs = (("Phy", 88), ("Chem", 91), ("Maths", 79))
for subject, mark in pairs:
    print(f"  {subject:<6} {mark}")

# Too many / too few raises immediately
try:
    p, q = (1, 2, 3)
except ValueError as exc:
    print("\nMismatch:", exc)
````

### ⚠️ Immutable means the *bindings* are fixed, not the contents

A tuple guarantees that slot 0 will always point at the same object. It does **not**
guarantee that the object itself cannot change.

**Analogy:** a tuple is a row of numbered lockers welded shut. You cannot swap what's in
locker 3 for something else — but if locker 3 contains a bag, you can still put things in
the bag.

This has a direct consequence: **a tuple is hashable only if everything inside it is.**
That is what determines whether it can be a dictionary key or a set member.

````python
# A tuple freezes its BINDINGS, not the objects they point to
t = ([1, 2], "fixed", 3)

print("before:", t)
t[0].append(99)              # the LIST inside is still mutable
print("after :", t)

# What you cannot do is rebind a slot
try:
    t[0] = [9, 9]
except TypeError as exc:
    print("\nRebinding a slot:", exc)

# This is exactly why hashability follows the same rule
safe = (1, 2, "three")
unsafe = ([1, 2], 3)

print("\nhash((1, 2, 'three')):", hash(safe))
try:
    hash(unsafe)
except TypeError as exc:
    print("hash(([1,2], 3))     :", exc)

# ...and therefore what can be a dict key
locations = {(19.07, 72.87): "Mumbai", (28.61, 77.20): "Delhi"}
print("\nTuple as dict key:", locations[(28.61, 77.20)])
````

## `namedtuple` and `NamedTuple` — tuples that explain themselves

Look back at `stud_1 = 101, 'Aditya', 'CBSE', 95.6`. To read the percentage you write
`stud_1[3]`. Six months later, nobody knows what `3` means — and if someone inserts a field,
every index in your codebase silently shifts.

A **named tuple** fixes this. It is a real tuple — same immutability, same indexing, same
unpacking, same memory efficiency — with **names attached to the positions**.

### Two spellings, same idea

| | Import | Style |
|---|---|---|
| `collections.namedtuple` | `from collections import namedtuple` | Older, no type hints |
| `typing.NamedTuple` | `from typing import NamedTuple` | Modern, class syntax, typed |

Prefer **`typing.NamedTuple`** in new code — it reads like a class, supports type hints, and
your editor can autocomplete the fields.

**Real-world use case:** database rows, coordinates, RGB colours, parsed CSV records, and
anything returned from a function with more than two values. It is the cheapest possible
upgrade from "tuple nobody can read" to "self-documenting record" — and unlike a
`dataclass` (see **17 Modern Python Features**), it stays hashable and immutable.

````python
from collections import namedtuple

# The original, untyped form - still very common in existing code
Student = namedtuple("Student", ["roll", "name", "board", "percentage"])

s1 = Student(101, "Aditya", "CBSE", 95.6)
print(s1)
print("name:", s1.name, "| percentage:", s1.percentage)

# Compare with the plain tuple from earlier in this notebook:
plain = (101, "Aditya", "CBSE", 95.6)
print("\nplain[3]  ->", plain[3], "   (what is 3 again?)")
print("s1.percentage ->", s1.percentage, "   (obvious)")
````

````python
from typing import NamedTuple

# The modern, typed form - a class statement, but still a real tuple
class Student(NamedTuple):
    roll: int
    name: str
    board: str
    percentage: float

s = Student(101, "Aditya", "CBSE", 95.6)

print("By name  :", s.name, s.percentage)
print("By index :", s[1], s[3], "   <- still works")
print("Unpacking:", *s)
print("repr     :", s)

# It IS a tuple - every tuple operation still applies
print("\nisinstance(s, tuple):", isinstance(s, tuple))
print("len                 :", len(s))
print("hashable            :", hash(s) is not None)

# Immutable, like any tuple
try:
    s.percentage = 99.0
except AttributeError as exc:
    print("\nAssignment:", exc)

# _replace() returns a NEW instance with a field changed
better = s._replace(percentage=99.0)
print("\noriginal:", better is s, "|", s.percentage)
print("updated :", better.percentage)

# _asdict() for serialisation
print("\nas dict:", s._asdict())
````

## Common Mistakes & Pitfalls

1. **Forgetting the trailing comma in a single-element tuple.** `(5)` is the integer 5; `(5,)` is a tuple. This is the most common tuple bug there is.
2. **Thinking immutable means 'contents cannot change'.** A tuple's *bindings* are fixed, but a mutable object inside it can still be mutated: `t = ([1], 2); t[0].append(3)` works.
3. **Assuming every tuple is hashable.** A tuple is hashable only if *everything inside it* is. `([1,2], 3)` cannot be a dict key.
4. **Using a tuple where a class or `NamedTuple` belongs.** `stud[3]` tells the reader nothing; `stud.percentage` tells them everything.
5. **Expecting `+` to modify in place.** `t1 + t2` builds a new tuple; tuples have no `append()`.
6. **Over-unpacking.** `a, b = (1, 2, 3)` raises `ValueError: too many values to unpack`. Use `a, b, *rest` when the length varies.

## Best Practices

- Use a **tuple** for a fixed-length heterogeneous record (a row, a coordinate, a return value); use a **list** for a variable-length homogeneous collection.
- Reach for **`NamedTuple`** as soon as a tuple has more than two or three fields.
- Use unpacking rather than indexing — `x, y = point` beats `point[0], point[1]`.
- Use `_` for values you intentionally discard: `_, _, year = date_parts`.
- Return a tuple from a function that has several results; the caller unpacks it.
- Prefer tuples for dict keys and set members — they are hashable and cannot be mutated out from under you.
