## String in Python: 
- Strings in Python are ordered collections of Unicode characters delimited by single, double or triple quotes.
- Strings are used in Python to record text information, such as name.
- Python strings are immutable i.e, once we've constructed a string we can't modify its contents.
- Python strings are iterable objects.
<br/><br/>
- Strings in python can be indexed often synonymously called subscripted as well. 
    - In forward indexing first character of the string has index 0.
    - In backward indexing last character of string has index -1.

### Mutable vs Immutable:
- Objects which can be modified after creation **in the same memory location** are called
  **mutable**; those which cannot are **immutable**.

| | Types |
|---|---|
| **Immutable** | `int`, `float`, `complex`, `bool`, `str`, `tuple`, `frozenset`, `bytes` |
| **Mutable** | `list`, `dict`, `set`, `bytearray`, most user-defined classes |

> **Corrected from the original note:** `long` was a *Python 2* type. In Python 3 there is
> only `int`, and it has unlimited precision — `long` no longer exists.

**Why this matters for strings:** every "modification" of a string actually builds a
**new** string. `s += "x"` does not extend `s`; it creates a new object and rebinds the
name. Run the cell below and watch the `id()` change.

````python
s = "hello"
print("original :", s, "| id:", id(s))

s += " world"          # looks like modification, is actually re-creation
print("after += :", s, "| id:", id(s), " <- a different object")

# Strings genuinely cannot be modified in place
try:
    s[0] = "H"
except TypeError as exc:
    print("\nAssignment:", exc)

# Compare with a list, which IS mutable - same object throughout
lst = [1, 2, 3]
print("\nlist before:", lst, "| id:", id(lst))
lst.append(4)
print("list after :", lst, "| id:", id(lst), " <- same object")

# The correct way to "change" a string: build a new one
s = "hello"
print("\nreplaced:", s.replace("h", "H"), "| original unchanged:", s)
````

### Characters and code points: `ord()` and `chr()`
- Every character in a string is a **Unicode code point** — under the hood, a number. `ord()` shows the number behind a character; `chr()` turns a number back into its character.
- This is why `'a' < 'b'` works, why sorting puts `'Z'` before `'a'`, and how every script — Latin, Devanagari, emoji — fits in the one `str` type.

````python
print(ord('0'),'to', ord('9'))
print(ord('a'),'to', ord('z'))
print(ord('A'),'to', ord('Z'))
print(ord('अ'),'to', ord('ह'))
````

````python
print(chr(48), chr(57))
print(chr(97), chr(122))
print(chr(65), chr(90))
````

## Access string elements

- **Forward indexing** starts with 0.    
- **Backward indexing** starts with -1.


### String Indexing: 
- Used to select the letter at a particular index/position
    - Syntax: `object[index]`

````python
filename = "backup_2026-08-21.tar.gz"

print(filename[0])       # first character
print(filename[7])       # character at index 7
print(filename[-1])      # last character
print(filename[-2])      # second from the end
````

### Substring or Index slicing in python: 
- Process to select a subset of an entire string.
- **syntax:** ```object[start : stop : stride(step)]```
   - **start (optional):** starting number, the default is 0(in fwd indexing).
   - **stop (optional):** ending number, is not included in the generated sequence, default is len(str)(in fwd indexing).
   - **step (optional):** the difference between each item in the sequence, the default is 1(in fwd indexing).

````python
print(filename)
print(filename[7:17])    # the date buried in the filename
print(filename[:6])      # the first 6 characters
print(filename[-6:])     # the extension, counted from the end
print(filename[::2])     # every second character (step = 2)
print(filename[::-1])    # step -1 walks backwards: the reverse
````

## Case & cleaning
Nothing a user types arrives clean. A signup form hands you `"  Aditya.Tripathi@Example.COM \n"` — stray spaces, a trailing newline, random capitalisation. Before you can store that address or compare it against your records, you **normalise** it.

That job belongs to two method families:

- **Case mapping** — `upper()`, `lower()`, `casefold()`, `title()`, `capitalize()`, `swapcase()`
- **Trimming** — `strip()`, `lstrip()`, `rstrip()`

Remember: strings are immutable, so every one of these **returns a new string** and leaves the original alone. `s.lower()` on its own line does nothing useful — assign the result.

````python
raw = "  Aditya.Tripathi@Example.COM \n"
email = raw.strip().lower()        # trim the ends, then normalise the case
print(repr(raw))                   # repr() makes the hidden whitespace visible
print(repr(email))

# Now two differently-typed versions of the same address compare equal
typed_later = "ADITYA.TRIPATHI@example.com  "
print(email == typed_later.strip().lower())
````

#### Conversion Functions:
- **upper():** Convert all lower alphabet character elements of the string to upper case.
<br/><br/>
- **lower():** Convert all upper alphabet character elements of the string to lower case.
<br/><br/>
- **swapcase():** Interchange between upper and lower case alphabetic character.
<br/><br/>    
- **capitalize():** It will make the first alphabet of string in upper-case and all other in lower-case.
<br/><br/>        
- **title():** It will make the first alphabet of every word of string in upper-case.

````python
heading = "quarterly usage report"
print(heading.upper())         # SHOUTING — constants, log levels
print(heading.title())         # Every Word Capitalised — display headings
print(heading.capitalize())    # Sentence style — first letter only
````

````python
level = "WARNING"
print(level.lower())           # normalise a log level before comparing
print("PascalCase".swapcase()) # upper<->lower; rarely needed, but it exists
````

#### `casefold()` and `partition()`

Two methods the original notes skip that you will reach for constantly:

- **`casefold()`** — like `lower()`, but more aggressive, and designed for *caseless
  matching* across languages. Use it whenever you compare user input case-insensitively.
- **`partition(sep)`** — splits into exactly three parts and **always returns three**, so
  unpacking it never raises. This makes it safer than `split()` when the separator might
  be missing.

````python
# .lower() is not enough for correct caseless comparison
german = "STRASSE"
also_german = "straße"          # ß uppercases to SS

print("lower()   :", german.lower() == also_german.lower())      # False
print("casefold():", german.casefold() == also_german.casefold())  # True

# partition() splits into exactly three parts - always, even when absent
email = "aditya@example.com"
user, at, domain = email.partition("@")
print(f"\nuser={user!r}  sep={at!r}  domain={domain!r}")

missing = "no-at-sign"
print("no separator:", missing.partition("@"))    # ('no-at-sign', '', '')

# Because it always returns 3 items, unpacking never raises - unlike split()
print("\nsplit  :", "a=b=c".split("=", 1))
print("partition:", "a=b=c".partition("="))
````

#### Padding Functions
- **strip():** It removes all leading and trailing whitespace of a string and can also be used to remove particular character from leading and trailing.
    - **Syntax:** `str.strip(characters)`
    - characters (optional) : A set of characters to remove as leading characters.
<br/><br/>
- **lstrip():** It removes all leading whitespace of a string and can also be used to remove particular character from leading.
    - **Syntax:** `str.lstrip('characters')`
    - characters (optional) : A set of characters to remove as leading characters.
<br/><br/>
- **rstrip():** It removes all trailing whitespace of a string and can also be used to remove particular character from trailing.
    - **Syntax:** `str.rstrip(characters)`
    - characters (optional) : A set of characters to remove as trailing characters.

````python
username = "   aditya_07   "
print(repr(username.strip()))      # both ends
print(repr(username.lstrip()))     # left only
print(repr(username.rstrip()))     # right only

# The optional argument is a SET of characters to remove from the ends
route = "///api/v2/users///"
print(route.strip('/'))
print(route.lstrip('/'))
````

#### ⚠️ `strip()` does not remove a substring — `removeprefix`/`removesuffix` do

> **Version note:** `str.removeprefix()` and `str.removesuffix()` were added in **Python 3.9**.

`strip(chars)` treats its argument as a **set of characters** and keeps removing any of
them from each end until it hits something else. `"report_2024.csv.csv".strip(".csv")` therefore
eats **both** trailing `.csv`s at once — and a name like `"stats.csv"` even loses its
leading `s`, giving `'tat'`. This surprises everyone exactly once.

If you want to remove an exact prefix or suffix, say so:

````python
filename = "report_2024.csv.csv"

# strip() removes CHARACTERS, not a substring - it strips any char in the set, repeatedly
print("strip('.csv')      :", filename.strip(".csv"))     # eats far too much!

# removesuffix() removes the exact suffix, once, or nothing at all  (3.9+)
print("removesuffix('.csv'):", filename.removesuffix(".csv"))

url = "https://example.com/page"
print("\nremoveprefix('https://'):", url.removeprefix("https://"))
print("removeprefix('http://') :", url.removeprefix("http://"), " <- unchanged, no match")

# The pre-3.9 idiom you will still see in older code
def old_removeprefix(s, prefix):
    return s[len(prefix):] if s.startswith(prefix) else s

print("\nOld idiom gives    :", old_removeprefix(url, "https://"))
````

#### String Substitution Functions

## Searching & testing
Scenario: your monitoring script reads one line from a server log:

```
2026-08-21 14:03:22 ERROR disk /dev/sda1 92% full
```

The questions you ask of a string are always the same four:

- **Is something in it?** → the `in` operator
- **Where exactly?** → `find()` / `index()`
- **Does it start or end a certain way?** → `startswith()` / `endswith()`
- **Is the whole string one kind of thing** (digits, letters, spaces)? → the `is...()` family

- **Membership Operator(in/not in):**
    - Test substring membership

````python
line = "2026-08-21 14:03:22 ERROR disk /dev/sda1 92% full"

print("ERROR" in line)        # the everyday substring test
print("WARNING" in line)
print("WARNING" not in line)
````

- **find():** Returns the lowest index of the substring if it is found in given string. If its is not found then it returns -1.
    - **Syntax:** `str.find(sub, start=0, stop=len(str))`
        - **sub:** It’s the substring which needs to be searched in the given string.
        - **start:** Starting position where sub is needs to be checked within the string and by default it takes 0.
        - **stop:** Ending position up to which sub is searched within the string; by default it takes `len(str)`, so the last character is included in the search.
<br/><br/>
- **rfind():** Return the highest index in the string where substring sub is found.
    - **Syntax:** `str.rfind(sub, start=0, stop=len(str))`

````python
print(line.find("ERROR"))     # the index where it starts
print(line.find("TRACE"))     # -1 means "not found" — no exception

path = "logs/app/2026/app.log"
print(path.find("/"))         # first slash
print(path.rfind("/"))        # last slash — where the filename begins
print(path[path.rfind("/") + 1:])   # slice from there: the bare filename
````

#### Search Functions:
- **index():** It returns the lowest index of occurrence of substring in a string. And raises ValueError if argument string is not found.
    - **Syntax:** `str.index(sub, start=0, stop=len(str))`
<br/><br/>
- **rindex():** It returns the highest index of occurrence of substring in a string. And raises ValueError if argument string is not found.
    - **Syntax:** `str.rindex(sub, start=0, stop=len(str))`

````python
print(line.index("ERROR"))    # same answer as find() when present...
# print(line.index("TRACE"))  # ...but ValueError: substring not found, when absent
print(path.rindex("/"))

# Rule of thumb: find() when absence is normal; index() when absence is a bug
# you want to fail loudly.
````

- **count():** It counts number of times substring occurs.
    - **Syntax:** `object.count(substring, start, stop)`

````python
record = "101,Aditya,dev,Pune"
print(record.count(","))          # 3 commas -> 4 fields
print(line.count("/"))            # slashes in the device path
print(line.count("ERROR"))
````

- **startswith():** It is used to check whether a given string starts with some particular string.
    - **Syntax:** `str.startswith(sub, start=0, stop=len(str))`

````python
print(line.startswith("2026-"))       # does the line begin with this year?
print(line.startswith("ERROR"))
print(line.startswith("ERROR", 20))   # start the check at index 20 instead
````

- **endswith():** It is used to check whether a given Sentence ends with some particular string.
    - **Syntax:** `str.endswith(sub, start, stop=len(str))`
        - end index is excluded

````python
print(line.endswith("full"))
print("report.csv".endswith(".csv"))

# startswith/endswith accept a TUPLE of options — one call, several suffixes
image = "diagram.jpeg"
print(image.endswith((".png", ".jpg", ".jpeg")))
````

#### Comparison Functions:
- **isupper():** Check whether all characters of string are in upper-case only.
<br/><br/>
- **islower():** Check whether all characters of string are in lower-case only.
<br/><br/>  
- **isdigit():** Check whether all characters of string are numeric only.
<br/><br/>    
- **isalpha():** Check whether all characters of string are alphabet only.
<br/><br/>        
- **isalnum():** Check whether all characters of string are either alphabetic or numeric only.
<br/><br/>
- **isspace():** Check whether all characters of string are only whitespace characters.
<br/><br/>
- **istitle():** Check whether the string is a titlecased string.

````python
# Validating form fields before converting or storing them
print("8080".isdigit())        # port number: digits only? safe for int()
print("80.5".isdigit())        # the dot is not a digit
print("Pune".isalpha())        # letters only
print("web01".isalnum())       # letters and digits
print("web_01".isalnum())      # the underscore fails the test
print("   ".isspace())
print("OK".isupper(), "ok".islower())
print("Deploy Report".istitle())
````

## Splitting & joining
Text usually arrives as **one string carrying several values** — a CSV row, a timestamp, a file path. `split()` and friends take it apart; `join()` puts pieces back together. Between them they are probably the two most-used string methods in working code.

- **split():** Splits a string on a separator and returns a list of substrings. If maxsplit is given, at most maxsplit splits are done, the leftmost ones.
    - Syntax: `str.split(sep=None, maxsplit=-1)`
    - ⚠️ The default is `sep=None`, **not** `' '`: with no argument, `split()` splits on *runs* of whitespace and drops empty strings; `split(' ')` cuts at every single space and keeps empties (see Common Mistakes item 6).
<br/><br/>
- **rsplit():** Return a list of the words in the string, using sep as the separator. If maxsplit is given, at most maxsplit splits are done, the rightmost ones.
    - Syntax: `str.rsplit(sep=None, maxsplit=-1)`

````python
record = "101,Aditya,dev,Pune"
print(record.split(","))              # one string -> a list of fields

stamp = "2026-08-21T14:03:22"
date, clock = stamp.split("T")        # exactly one separator? unpack directly
print(date, "|", clock)

version = "2.4.1-beta-hotfix"
print(version.rsplit("-", 1))         # maxsplit=1 from the RIGHT: keep the head intact
print(version.split("-", 1))         # ...versus the same split from the left

# The sep=None trap from the syntax note above
spaced = "a  b"
print(spaced.split())      # runs of whitespace collapse, empties dropped
print(spaced.split(" "))   # every single space cuts -> an empty string appears
````

- **splitlines([keepends]):** Return a list of the lines in the string, breaking at line boundaries.

````python
log = "INFO service started\nWARN slow response: 2.4s\nERROR disk full"
print(log.splitlines())
print(log.splitlines(True))   # keepends=True keeps the newline on each line
````

- **join():** This method returns a string in which the elements of sequence have been joined by str separator. 
    - **Syntax:** ```separator.join(iterable)```

````python
fields = ['101', 'Aditya', 'dev', 'Pune']
print(",".join(fields))                               # back into a CSV row

print("/".join(["logs", "app", "2026", "app.log"]))   # build a path

# A classic real-world combo: page title -> URL slug
title = "My First Blog Post!"
slug = "-".join(title.lower().replace("!", "").split())
print(slug)
````

#### ⚠️ `join()` vs `+=` in a loop — the O(n²) trap

Because strings are immutable, `result += word` inside a loop **rebuilds the whole string
every iteration**. For 20,000 words that is 20,000 allocations, each one copying everything
accumulated so far.

`str.join()` walks the sequence once, works out the total size, allocates once, and fills it.

This is the most common accidental performance bug in beginner Python. Measure it yourself:

````python
import time

words = ["Python"] * 20_000

# SLOW: each += builds a brand-new string and copies everything so far -> O(n^2)
start = time.perf_counter()
slow = ""
for w in words:
    slow += w + " "
slow_time = time.perf_counter() - start

# FAST: join allocates once -> O(n)
start = time.perf_counter()
fast = " ".join(words) + " "
fast_time = time.perf_counter() - start

print(f"+= in a loop : {slow_time * 1000:7.2f} ms")
print(f"str.join()   : {fast_time * 1000:7.2f} ms")
print(f"join is roughly {slow_time / fast_time:.0f}x faster here")
print("Same result:", slow == fast)

# join() only accepts strings - convert numbers first
numbers = [1, 2, 3]
# print(", ".join(numbers))          # TypeError: sequence item 0: expected str
print("\nJoined numbers:", ", ".join(str(n) for n in numbers))
````

## Formatting values into strings
You have data in variables and need a sentence — for a log line, a report, a UI label. Python has three formatting styles; you will *meet* all of them in real code and *write* mostly one. The same example rendered all three ways:

- **String Formatting:**
    - To add formatted objects to string
        - We can use %d %f %s method
            - `%[flags][width][.precision]type` 
        - We can use the .format() method
        - We can use the f-string method

````python
service = "auth-api"
requests = 1284
uptime = 99.97

report1 = "Service %s handled %d requests at %.2f%% uptime." % (service, requests, uptime)
print(report1)
# Order and type of the values must match the placeholders
````

- Python's **str.format()** method of the str class allows you to carry out variable substitutions and value formatting. 
    - Formatters work by putting in one or more replacement fields or placeholders defined by a pair of curly braces {} into a string and calling the str.format() method.
    - When we pass the value we want to concatenate with the string into the method, this value will appear in the same place that your placeholder is positioned when we run the program.

````python
report2 = "Service {} handled {} requests at {:.2f}% uptime.".format(service, requests, uptime)
print(report2)
# placeholders are filled left to right (position matching)
````

````python
report3 = "Service {2} handled {0} requests at {1:.2f}% uptime.".format(requests, uptime, service)
print(report3)
# {0}, {1}, {2} pick values by index — handy when one value repeats
````

````python
report4 = "Service {s} handled {n} requests at {u:.2f}% uptime.".format(n=requests, u=uptime, s=service)
print(report4)
# named placeholders — the order in the call no longer matters
````

````python
report5 = f"Service {service} handled {requests} requests at {uptime:.2f}% uptime."
print(report5)
# the f-string: each value sits exactly where it prints
````

> **Which of those four should you write?** The last one — the **f-string**.
>
> | Style | Verdict |
> |---|---|
> | `%s` / `%d` | **Legacy.** Recognise it in old code; never write it. |
> | `.format()` | Fine when the template is stored apart from the values (config, i18n tables). |
> | `.format()` with `{0}` / `{name}` | Same — useful when a value is repeated or reordered. |
> | **f-string** | **The default.** Shortest, fastest, and the value sits where it prints. |
>
> The full format-spec mini-language (`:.2f`, `:,`, `:>10`, `{x=}`) is covered in
> **1.3 Python Input Output** — it applies identically to all three styles.

## Bytes and str: text vs raw data

This is one of the genuinely important distinctions in Python 3, and the original notes
covered it in three lines. It is worth more.

### The concept

- A **`str`** is a sequence of **Unicode code points** — abstract characters. `"नमस्कार"`,
  `"café"`, `"A"`. It has no inherent representation in memory that you should reason about.
- A **`bytes`** object is a sequence of **integers 0–255** — raw octets. This is what
  actually travels over a network, sits in a file, or arrives from a socket.

An **encoding** (almost always **UTF-8**) is the rule for converting between them.

```
   str  --.encode('utf-8')-->  bytes
   str  <--.decode('utf-8')--  bytes
```

**Analogy:** `str` is the *meaning* of a sentence; `bytes` is the ink pattern on the page.
You need an agreed alphabet (the encoding) to get from one to the other.

### Why you care
Every file you open in binary mode, every HTTP response body, every socket `recv()`, and
every database blob hands you `bytes`. If you try to use it as text without decoding, you
get `TypeError: a bytes-like object is required, not 'str'` — one of the most common
Python 3 errors there is.

> **Version note:** in Python 2 `str` *was* bytes, and mixing the two silently sort-of-worked.
> Python 3 made them separate types on purpose. This is the single largest source of
> "Python 2 code that won't port".

````python
string = "Python is interesting."
# string with encoding 'utf-8'
arr = bytes(string, 'utf-8')
print(arr)
````

### Encoding and decoding in practice

````python
text = "Grüße"           # 5 characters, but NOT 5 bytes in UTF-8

encoded = text.encode("utf-8")
print("str    :", text, "| len:", len(text))
print("bytes  :", encoded, "| len:", len(encoded))
print("decoded:", encoded.decode("utf-8"))

# Different encodings produce different bytes for the same text
print("\nutf-8    :", text.encode("utf-8"))
print("utf-16   :", text.encode("utf-16"))
print("latin-1  :", text.encode("latin-1"))

# ...and decoding with the WRONG one gives mojibake or an error
print("\nWrong decoding:", text.encode("utf-8").decode("latin-1"))
try:
    "नमस्कार".encode("latin-1")
except UnicodeEncodeError as exc:
    print("Cannot encode Devanagari as latin-1:", exc.reason)
````

````python
# bytes are IMMUTABLE; bytearray is the mutable version
data = bytearray(b"hello")
data[0] = ord("H")
print(data, "->", bytes(data))

# Indexing bytes gives an INTEGER, not a 1-character bytes object
raw = b"ABC"
print("raw[0]      :", raw[0], type(raw[0]))     # 65, an int
print("raw[0:1]    :", raw[0:1])                 # b'A', a bytes slice

# The classic Python 3 error, and its fix
try:
    b"prefix-" + "suffix"
except TypeError as exc:
    print("\nMixing bytes and str:", exc)

print("Fixed:", b"prefix-" + "suffix".encode("utf-8"))

# Real-world shape: read raw, decode once, work in str, encode on the way out
raw_from_network = "Grüße, नमस्कार".encode("utf-8")
print("\nOn the wire :", raw_from_network)
text = raw_from_network.decode("utf-8")
print("After decode:", text)
````

## Common Mistakes & Pitfalls

1. **Expecting a string method to modify in place.** `s.upper()` *returns* a new string; it does not change `s`. `s = s.upper()` is what you meant.
2. **Using `strip(".csv")` to remove a suffix.** It strips *characters*, not a substring. Use `removesuffix()`.
3. **Building strings with `+=` in a loop.** O(n^2). Use `str.join()`.
4. **`.lower()` for caseless comparison across languages.** Use `.casefold()`.
5. **Confusing `find()` with `index()`.** `find()` returns `-1` when absent; `index()` raises `ValueError`. Picking the wrong one either hides a bug or crashes.
6. **Forgetting `str.split()` with no argument is special.** It splits on *runs* of whitespace and drops empties; `split(' ')` does not. `'a  b'.split()` gives 2 items, `.split(' ')` gives 3.
7. **Treating `bytes` as text.** Decode once at the boundary, work in `str`, encode on the way out.
8. **Indexing `bytes` and expecting a byte.** `b"ABC"[0]` is the integer `65`.

## Best Practices

- Use **f-strings** for formatting. Reserve `.format()` for stored templates; never write `%`.
- Use `str.join()` to assemble strings from sequences.
- Prefer `in` for substring tests (`if "x" in s`) over `find(...) != -1`.
- Use raw strings (`r"..."`) for regex patterns and Windows paths.
- Decode bytes at the edge of your program — the 'Unicode sandwich'.
- Use `.casefold()` for case-insensitive comparison, `.strip()` on anything a human typed.
- Triple-quoted strings for multi-line text; `textwrap.dedent()` when indentation gets in the way.
