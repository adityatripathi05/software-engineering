## String in Python: 
- Strings in Python are ordered collections of Unicode characters delimited by single, double or triple quotes.
- Strings are used in Python to record text information, such as name.
- Python strings are immutable i.e, once we've constructed a string we can't modify its contents.
- Python strings are iterable objects.
<br/><br/>
- Strings in python can be indexed often synonymously called subscripted as well. 
    - In forward indexing first character of the string has index 0.
    - In backward indexing last character of string has index -1.

### Mutable vs Immutable:
- Objects which can be modified after creation **in the same memory location** are called
  **mutable**; those which cannot are **immutable**.

| | Types |
|---|---|
| **Immutable** | `int`, `float`, `complex`, `bool`, `str`, `tuple`, `frozenset`, `bytes` |
| **Mutable** | `list`, `dict`, `set`, `bytearray`, most user-defined classes |

> **Corrected from the original note:** `long` was a *Python 2* type. In Python 3 there is
> only `int`, and it has unlimited precision — `long` no longer exists.

**Why this matters for strings:** every "modification" of a string actually builds a
**new** string. `s += "x"` does not extend `s`; it creates a new object and rebinds the
name. Run the cell below and watch the `id()` change.

````python
s = "hello"
print("original :", s, "| id:", id(s))

s += " world"          # looks like modification, is actually re-creation
print("after += :", s, "| id:", id(s), " <- a different object")

# Strings genuinely cannot be modified in place
try:
    s[0] = "H"
except TypeError as exc:
    print("\nAssignment:", exc)

# Compare with a list, which IS mutable - same object throughout
lst = [1, 2, 3]
print("\nlist before:", lst, "| id:", id(lst))
lst.append(4)
print("list after :", lst, "| id:", id(lst), " <- same object")

# The correct way to "change" a string: build a new one
s = "hello"
print("\nreplaced:", s.replace("h", "H"), "| original unchanged:", s)
````

> | Style | Verdict |

#### `casefold()` and `partition()`

Two methods the original notes skip that you will reach for constantly:

- **`casefold()`** — like `lower()`, but more aggressive, and designed for *caseless
  matching* across languages. Use it whenever you compare user input case-insensitively.
- **`partition(sep)`** — splits into exactly three parts and **always returns three**, so
  unpacking it never raises. This makes it safer than `split()` when the separator might
  be missing.

````python
# .lower() is not enough for correct caseless comparison
german = "STRASSE"
also_german = "straße"          # ß uppercases to SS

print("lower()   :", german.lower() == also_german.lower())      # False
print("casefold():", german.casefold() == also_german.casefold())  # True

# partition() splits into exactly three parts - always, even when absent
email = "aditya@example.com"
user, at, domain = email.partition("@")
print(f"\nuser={user!r}  sep={at!r}  domain={domain!r}")

missing = "no-at-sign"
print("no separator:", missing.partition("@"))    # ('no-at-sign', '', '')

# Because it always returns 3 items, unpacking never raises - unlike split()
print("\nsplit  :", "a=b=c".split("=", 1))
print("partition:", "a=b=c".partition("="))
````

#### ⚠️ `strip()` does not remove a substring — `removeprefix`/`removesuffix` do

> **Version note:** `str.removeprefix()` and `str.removesuffix()` were added in **Python 3.9**.

`strip(chars)` treats its argument as a **set of characters** and keeps removing any of
them from each end until it hits something else. `"report.csv".strip(".csv")` therefore
also eats the `r`, the `p`, the `o`... This surprises everyone exactly once.

If you want to remove an exact prefix or suffix, say so:

````python
filename = "report_2024.csv.csv"

# strip() removes CHARACTERS, not a substring - it strips any char in the set, repeatedly
print("strip('.csv')      :", filename.strip(".csv"))     # eats far too much!

# removesuffix() removes the exact suffix, once, or nothing at all  (3.9+)
print("removesuffix('.csv'):", filename.removesuffix(".csv"))

url = "https://example.com/page"
print("\nremoveprefix('https://'):", url.removeprefix("https://"))
print("removeprefix('http://') :", url.removeprefix("http://"), " <- unchanged, no match")

# The pre-3.9 idiom you will still see in older code
def old_removeprefix(s, prefix):
    return s[len(prefix):] if s.startswith(prefix) else s

print("\nOld idiom gives    :", old_removeprefix(url, "https://"))
````

#### ⚠️ `join()` vs `+=` in a loop — the O(n²) trap

Because strings are immutable, `result += word` inside a loop **rebuilds the whole string
every iteration**. For 20,000 words that is 20,000 allocations, each one copying everything
accumulated so far.

`str.join()` walks the sequence once, works out the total size, allocates once, and fills it.

This is the most common accidental performance bug in beginner Python. Measure it yourself:

````python
import time

words = ["Python"] * 20_000

# SLOW: each += builds a brand-new string and copies everything so far -> O(n^2)
start = time.perf_counter()
slow = ""
for w in words:
    slow += w + " "
slow_time = time.perf_counter() - start

# FAST: join allocates once -> O(n)
start = time.perf_counter()
fast = " ".join(words) + " "
fast_time = time.perf_counter() - start

print(f"+= in a loop : {slow_time * 1000:7.2f} ms")
print(f"str.join()   : {fast_time * 1000:7.2f} ms")
print(f"join is roughly {slow_time / fast_time:.0f}x faster here")
print("Same result:", slow == fast)

# join() only accepts strings - convert numbers first
numbers = [1, 2, 3]
# print(", ".join(numbers))          # TypeError: sequence item 0: expected str
print("\nJoined numbers:", ", ".join(str(n) for n in numbers))
````

## Bytes and str: text vs raw data

This is one of the genuinely important distinctions in Python 3, and the original notes
covered it in three lines. It is worth more.

### The concept

- A **`str`** is a sequence of **Unicode code points** — abstract characters. `"नमस्कार"`,
  `"café"`, `"A"`. It has no inherent representation in memory that you should reason about.
- A **`bytes`** object is a sequence of **integers 0–255** — raw octets. This is what
  actually travels over a network, sits in a file, or arrives from a socket.

An **encoding** (almost always **UTF-8**) is the rule for converting between them.

```
   str  --.encode('utf-8')-->  bytes
   str  <--.decode('utf-8')--  bytes
```

**Analogy:** `str` is the *meaning* of a sentence; `bytes` is the ink pattern on the page.
You need an agreed alphabet (the encoding) to get from one to the other.

### Why you care
Every file you open in binary mode, every HTTP response body, every socket `recv()`, and
every database blob hands you `bytes`. If you try to use it as text without decoding, you
get `TypeError: a bytes-like object is required, not 'str'` — one of the most common
Python 3 errors there is.

> **Version note:** in Python 2 `str` *was* bytes, and mixing the two silently sort-of-worked.
> Python 3 made them separate types on purpose. This is the single largest source of
> "Python 2 code that won't port".

````python
string = "Python is interesting."
# string with encoding 'utf-8'
arr = bytes(string, 'utf-8')
print(arr)
````

- bytes is a sequence of bytes, str is a sequence of Unicode codepoints

````python
# Create a byte of given integer size
size = 5
arr = bytes(size)
print(arr)
````

````python
# Convert iterable list to bytes
rList = [1, 2, 3, 4, 5]
arr = bytes(rList)
print(arr)
````

### Encoding and decoding in practice

````python
text = "Grüße"           # 5 characters, but NOT 5 bytes in UTF-8

encoded = text.encode("utf-8")
print("str    :", text, "| len:", len(text))
print("bytes  :", encoded, "| len:", len(encoded))
print("decoded:", encoded.decode("utf-8"))

# Different encodings produce different bytes for the same text
print("\nutf-8    :", text.encode("utf-8"))
print("utf-16   :", text.encode("utf-16"))
print("latin-1  :", text.encode("latin-1"))

# ...and decoding with the WRONG one gives mojibake or an error
print("\nWrong decoding:", text.encode("utf-8").decode("latin-1"))
try:
    "नमस्कार".encode("latin-1")
except UnicodeEncodeError as exc:
    print("Cannot encode Devanagari as latin-1:", exc.reason)
````

````python
# bytes are IMMUTABLE; bytearray is the mutable version
data = bytearray(b"hello")
data[0] = ord("H")
print(data, "->", bytes(data))

# Indexing bytes gives an INTEGER, not a 1-character bytes object
raw = b"ABC"
print("raw[0]      :", raw[0], type(raw[0]))     # 65, an int
print("raw[0:1]    :", raw[0:1])                 # b'A', a bytes slice

# The classic Python 3 error, and its fix
try:
    b"prefix-" + "suffix"
except TypeError as exc:
    print("\nMixing bytes and str:", exc)

print("Fixed:", b"prefix-" + "suffix".encode("utf-8"))

# Real-world shape: read raw, decode once, work in str, encode on the way out
raw_from_network = "Grüße, नमस्कार".encode("utf-8")
print("\nOn the wire :", raw_from_network)
text = raw_from_network.decode("utf-8")
print("After decode:", text)
````

## Common Mistakes & Pitfalls

1. **Expecting a string method to modify in place.** `s.upper()` *returns* a new string; it does not change `s`. `s = s.upper()` is what you meant.
2. **Using `strip(".csv")` to remove a suffix.** It strips *characters*, not a substring. Use `removesuffix()`.
3. **Building strings with `+=` in a loop.** O(n^2). Use `str.join()`.
4. **`.lower()` for caseless comparison across languages.** Use `.casefold()`.
5. **Confusing `find()` with `index()`.** `find()` returns `-1` when absent; `index()` raises `ValueError`. Picking the wrong one either hides a bug or crashes.
6. **Forgetting `str.split()` with no argument is special.** It splits on *runs* of whitespace and drops empties; `split(' ')` does not. `'a  b'.split()` gives 2 items, `.split(' ')` gives 3.
7. **Treating `bytes` as text.** Decode once at the boundary, work in `str`, encode on the way out.
8. **Indexing `bytes` and expecting a byte.** `b"ABC"[0]` is the integer `65`.

## Best Practices

- Use **f-strings** for formatting. Reserve `.format()` for stored templates; never write `%`.
- Use `str.join()` to assemble strings from sequences.
- Prefer `in` for substring tests (`if "x" in s`) over `find(...) != -1`.
- Use raw strings (`r"..."`) for regex patterns and Windows paths.
- Decode bytes at the edge of your program — the 'Unicode sandwich'.
- Use `.casefold()` for case-insensitive comparison, `.strip()` on anything a human typed.
- Triple-quoted strings for multi-line text; `textwrap.dedent()` when indentation gets in the way.
