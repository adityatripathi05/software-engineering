### Complex numbers

Python has complex numbers as a **built-in type** — unusual, and a legacy of its scientific
roots. The imaginary unit is written **`j`** (electrical-engineering convention), not `i`.

**Real-world use case:** signal processing, control systems, electrical engineering, and
the FFT. If you're not doing those, you will never touch `complex` — but it's worth knowing
it's there, because it explains why `int`/`float`/`complex` form a *numeric tower* with
automatic widening.

```python
z = 3 + 4j                 # note: j, not i - Python follows the engineering convention

print("z          :", z, type(z))
print("real part  :", z.real)
print("imag part  :", z.imag)
print("conjugate  :", z.conjugate())
print("magnitude  :", abs(z))          # sqrt(3^2 + 4^2) = 5.0

# Arithmetic works as you would expect
print("\n(3+4j) * (1-2j) =", (3 + 4j) * (1 - 2j))

# complex() constructor
print("complex(3, 4)   =", complex(3, 4))
print("complex('3+4j') =", complex("3+4j"))

# The numeric tower: int -> float -> complex, widening is automatic
print("\n1 + 2.5      =", 1 + 2.5, type(1 + 2.5))
print("2.5 + (1+0j) =", 2.5 + (1 + 0j), type(2.5 + (1 + 0j)))

# ...but narrowing is not, and comparison with complex is undefined
try:
    print(3 + 4j > 1 + 2j)
except TypeError as exc:
    print("\nOrdering complex numbers:", exc)
```

## Precision Handling in Python

Floating-point numbers are **approximations**. A `float` is a 64-bit IEEE-754 double, and
most decimal fractions simply cannot be represented exactly in binary — in the same way
that `1/3` cannot be written exactly in decimal.

This is not a Python quirk; it is how essentially every language represents floats.

### Three different jobs, three different tools

| You want | Use |
|---|---|
| To **display** a number with N decimals | f-string format spec: `f"{x:.2f}"` |
| To **round** a value for further calculation | `round(x, n)` |
| **Exact** decimal arithmetic (money!) | `decimal.Decimal` |
| **Exact** rational arithmetic | `fractions.Fraction` |
| To **compare** two floats | `math.isclose()` — never `==` |

> The original note listed `%` formatting first. Use an f-string; see **1.3** for the full
> format-spec mini-language.

### ⚠️ `round()` does not round the way you were taught at school

Python uses **banker's rounding** (round-half-to-even) for exact halves, as mandated by
IEEE-754. `round(2.5)` is `2`, not `3`. This is deliberate — it removes the upward bias you
get from always rounding halves away from zero, which matters when you sum thousands of
rounded values.

```python
import math
pi = math.pi
print('pi:', pi)

var = round(pi, 4)#roundoff after 4 palce of decimal
print('pi:',var)

print('pi: %.3f' %pi)#roundoff after 3 palce of decimal

print('pi: {0} roundoff after 2 of place decimal= {0:.2f}'.format(pi))#roundoff after 2 of place
```

### Seeing the problem for yourself

```python
import math

# The canonical demonstration
print("0.1 + 0.2      =", 0.1 + 0.2)
print("== 0.3         ?", 0.1 + 0.2 == 0.3)
print("What 0.1 really is:", f"{0.1:.20f}")

# The correct way to compare floats
print("\nmath.isclose(0.1+0.2, 0.3):", math.isclose(0.1 + 0.2, 0.3))
print("with tolerance            :", math.isclose(1000.0, 1000.1, rel_tol=1e-3))

# Banker's rounding: exact halves go to the nearest EVEN number
print("\nround(0.5) =", round(0.5), "   round(1.5) =", round(1.5))
print("round(2.5) =", round(2.5), "   round(3.5) =", round(3.5))
print("round(2.675, 2) =", round(2.675, 2), " <- 2.675 is really 2.67499...")

# Display rounding vs value rounding are different operations
value = 3.14159
print(f"\ndisplay only : {value:.2f}   (value unchanged: {value})")
print(f"actual round : {round(value, 2)}")
```

```python
from decimal import Decimal, getcontext, ROUND_HALF_UP

# Decimal stores decimal digits, so decimal fractions are EXACT
print("float  :", 0.1 + 0.2 == 0.3)
print("Decimal:", Decimal("0.1") + Decimal("0.2") == Decimal("0.3"))

# ALWAYS construct from a string - Decimal(0.1) inherits the float's error
print("\nDecimal('0.1'):", Decimal("0.1"))
print("Decimal(0.1)  :", Decimal(0.1), " <- the float was already wrong")

# Money: the classic case
price = Decimal("19.99")
qty = 3
tax_rate = Decimal("0.18")

subtotal = price * qty
tax = (subtotal * tax_rate).quantize(Decimal("0.01"), rounding=ROUND_HALF_UP)
print(f"\nSubtotal: {subtotal}")
print(f"Tax     : {tax}")
print(f"Total   : {subtotal + tax}")

# Compare with floats accumulating error over many operations
float_total = sum(0.1 for _ in range(10))
dec_total = sum(Decimal("0.1") for _ in range(10))
print(f"\nfloat  10 x 0.1 = {float_total!r}")
print(f"Decimal 10 x 0.1 = {dec_total}")
```

```python
from fractions import Fraction

# Fraction stores a numerator and denominator - exact rational arithmetic
third = Fraction(1, 3)
print("1/3 as Fraction:", third)
print("three thirds sum:", third + third + third, " <- exactly 1")
print("as float        :", 1/3 + 1/3 + 1/3)

# Fractions can be built from strings and floats
print("\nFraction('0.25') :", Fraction("0.25"))
print("Fraction(0.1)    :", Fraction(0.1))        # reveals what 0.1 REALLY is
print("limit_denominator:", Fraction(0.1).limit_denominator(100))

# Choosing between them
print("""
float    -> fast, approximate. Science, graphics, statistics.
Decimal  -> exact decimal. Money, invoices, tax, anything a human audits.
Fraction -> exact rational. Symbolic maths, probabilities, exact ratios.
""")
```

### `int` has unlimited precision; `float` does not

> **Version note:** Python 2 had separate `int` and `long` types. Python 3 merged them —
> there is only `int`, and it grows to whatever size it needs.

This is a genuine and unusual strength. In C, Java or Go, integer arithmetic silently wraps
around at 64 bits. In Python it simply keeps going. Floats, being fixed-width IEEE-754
doubles, do still overflow — to `inf`.

```python
# Python integers have UNLIMITED precision - no overflow, ever
big = 2 ** 1000
print("2**1000 has", len(str(big)), "digits")
print("First 60:", str(big)[:60], "...")

print("\n100! =", len(str(__import__("math").factorial(100))), "digits")

# In C or Java this would overflow silently or raise. In Python it just grows.
print("\n(2**63 - 1) + 1 =", (2 ** 63 - 1) + 1, " <- would overflow a C int64")

# The cost is speed and memory, not correctness
import sys
print("\nsys.getsizeof(1)      :", sys.getsizeof(1), "bytes")
print("sys.getsizeof(2**1000):", sys.getsizeof(2 ** 1000), "bytes")

# Floats DO have limits - they overflow to infinity
import math
print("\nfloat max      :", sys.float_info.max)
print("1e308 * 10     :", 1e308 * 10)
print("is inf         :", math.isinf(1e308 * 10))

# NaN: not equal to anything, including itself
nan = float("nan")
print("\nnan == nan     :", nan == nan, " <- always False!")
print("math.isnan(nan):", math.isnan(nan), " <- the correct test")
```

### Calculate the compound interest (C.I.)
- Formula for simple interest for amount P after n years is
    - `I= (P*R*T)/100`
- Formula for compound interest for amount P after n years is
    - `A = P(1 + r/100)^n`
    - Here, A is the Amount after n years, r is the rate of interest, and P is the principal.
- Task is to calculate the amount after two years on a principal amount of USD 1000 at the rate of 10 % p.a.

```python
# store variable names
P,r,n = 1000, 10, 2

# compound interest formula
A = P*pow((1+(r/100)),n)

# display compound interest
print(A)

# display interest
interest= A-P
print(interest)
```

## None

- Python has a special literal for the null value: **`None`**.
- `None` represents *the absence of a value* — not zero, not empty, not `False`.
- It is a singleton: there is exactly **one** `None` object in a running program, which is
  precisely why `is None` is the correct way to test for it.
- `None` is the implicit return value of any function that doesn't return anything.

### Why `is None` and not `== None`

Because `None` is a singleton, `is` is both faster and safer. `==` calls `__eq__`, which a
custom class can override — so `x == None` can be made to lie. `x is None` cannot.

**Real-world use case:** `None` is the standard *sentinel* for "no value supplied yet".
It's how you write an optional parameter whose default must not be shared between calls —
the mutable-default-argument fix you'll meet in **04 Functions**.

```python
a = None
```

```python
print(a is None)
```

```python
def find_user(user_id):
    """Return a user dict, or None when there is no such user."""
    users = {1: {"name": "Aditya"}, 2: {"name": "Priya"}}
    return users.get(user_id)          # .get() returns None when the key is absent

result = find_user(1)
print("found  :", result)

result = find_user(99)
print("missing:", result)

# The correct check
if result is None:
    print("No such user")

# A function with no return statement returns None implicitly
def greet(name):
    print(f"Hello, {name}")

returned = greet("Aditya")
print("\ngreet() returned:", returned)

# WATCH OUT: `if not result` is NOT the same as `if result is None`
for value in [None, 0, "", [], False]:
    print(f"  {value!r:>6}: is None -> {value is None:<5}  not value -> {not value}")

# ...so a function returning 0 or "" legitimately would be mishandled by `if not result`
```

## Bool Datatype:
- The Boolean type represents a logical state. There are exactly two values: **`True`** and
  **`False`** (both capitalised).
- Booleans drive every decision statement and `while` loop in Python.
- `bool()` converts other types, following the **truthiness** rules below.

### `bool` is a subclass of `int`

This surprises people: `True` really *is* `1` and `False` really *is* `0`. You can do
arithmetic with them — and `sum()` over a list of booleans is the idiomatic way to count
how many are true.

### Truthiness: what counts as False

Every object has a truth value. Only these are **falsy**:

| Falsy | Example |
|---|---|
| `None` | `None` |
| `False` | `False` |
| Zero of any numeric type | `0`, `0.0`, `0j`, `Decimal(0)` |
| Empty sequence | `''`, `()`, `[]` |
| Empty mapping / set | `{}`, `set()` |
| Objects defining `__bool__()` returning `False` | custom classes |
| Objects defining `__len__()` returning `0` | custom containers |

**Everything else is truthy** — including `"False"`, `"0"`, `[0]` and `-1`.

```python
print(bool(0), bool(0.0))
```

```python
print(bool(100), bool(-1), bool(53.4))
```

```python
print(bool([])) # empty list
```

```python
# bool is literally a subclass of int
print("issubclass(bool, int):", issubclass(bool, int))
print("True + True  =", True + True)
print("sum of flags =", sum([True, False, True, True]))     # idiomatic counting

scores = [45, 82, 91, 30, 67]
print("How many passed:", sum(score >= 50 for score in scores))

# Falsy values
for value in [None, False, 0, 0.0, "", (), [], {}, set()]:
    print(f"  bool({value!r:>8}) = {bool(value)}")

# ...and some truthy ones that surprise people
print()
for value in ["False", "0", " ", [0], (0,), {"a": None}, -1]:
    print(f"  bool({value!r:>12}) = {bool(value)}")

# THE IDIOM: test the object itself, not len() or == True
items = []
if not items:                       # good
    print("\nempty (Pythonic check)")
# if len(items) == 0:               # works, but noisy
# if items == []:                   # works, but type-specific
# if bool(items) == True:           # never write this
```

## Common Mistakes & Pitfalls

1. **Comparing floats with `==`.** `0.1 + 0.2 == 0.3` is `False`. Use `math.isclose()`.
2. **Using `float` for money.** Errors accumulate and auditors notice. Use `Decimal`, constructed **from strings** — `Decimal(0.1)` inherits the float's error.
3. **Expecting `round(2.5)` to be `3`.** Python uses banker's rounding; it is `2`.
4. **Testing `nan` with `==`.** `float('nan') == float('nan')` is `False` by definition. Use `math.isnan()`.
5. **Confusing display rounding with value rounding.** `f"{x:.2f}"` changes what you see; `round(x, 2)` changes the value.
6. **Assuming integer overflow.** Python `int` never overflows — but it does get slow and memory-hungry at extreme sizes.
7. **Writing `if x == True:`.** Just write `if x:`. And `if x is True:` is stricter than you probably want, since `1 == True` but `1 is not True`.
8. **Using `not x` when you mean `x is None`.** `0`, `""` and `[]` are falsy but are not `None`.

## Best Practices

- `Decimal` for currency, `Fraction` for exact ratios, `float` for measurement and science.
- Always build `Decimal` from a string: `Decimal("0.1")`.
- Use `math.isclose()` for float comparison; pass `rel_tol`/`abs_tol` deliberately.
- Use underscores in long numeric literals: `1_000_000`.
- Use `is None` / `is not None` for null checks — never `==`.
- Prefer truthiness (`if items:`) over `len(items) == 0` or `items == []`.
- Use `sum(condition for x in items)` to count matches — `bool` is an `int`.
