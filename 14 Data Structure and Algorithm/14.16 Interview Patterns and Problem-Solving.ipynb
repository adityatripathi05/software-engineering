## A framework for the unfamiliar problem

Under pressure, people jump straight to code and get stuck. A sequence prevents that.

```
   1. UNDERSTAND   restate it. Ask about edge cases and constraints.
   2. EXAMPLES     work one small case BY HAND. Then an edge case.
   3. APPROACH     brute force first. Say its complexity out loud.
   4. OPTIMISE     name the bottleneck, then the pattern that removes it.
   5. CODE         only now.
   6. TEST         the example, an edge case, and the case you fear.
```

### Step 1 - the questions worth asking

| Ask | Why it changes the answer |
|---|---|
| How large is n? | decides the acceptable complexity (see below) |
| Is the input **sorted**? | unlocks two pointers and binary search (**14.3**, **14.11**) |
| May I **sort** it? | O(n log n) + O(n) often beats O(n²) |
| May I **modify** the input? | decides whether O(1) space is achievable |
| Duplicates? Negatives? Empty? | sliding windows break on negatives (**14.3**) |
| One query or many? | many queries justify preprocessing (**14.3** prefix sums) |

> 🔴 **Always give the brute force first.** "The naive approach is O(n²) — let me improve it" is a strong opening. Silence while you search for the clever answer is not, and a brute force you can state is worth more than an optimal one you cannot reach.

### The three questions that pick a pattern

1. **Is it sorted?** → binary search or two pointers.
2. **Am I searching inside a loop?** → a hash map turns O(n²) into O(n).
3. **Do the choices interact?** → DP if yes, greedy if provably not.

## Reading the constraints

The size of n tells you which complexity is intended. This is not a trick — interviewers and problem setters choose the bound deliberately.

| n up to | Acceptable | Suggests |
|---|---|---|
| 10-12 | O(n!) | permutations, brute force (**14.12**) |
| ~20 | O(2ⁿ) | subsets, bitmask DP (**14.15**) |
| ~500 | O(n³) | triple loop, Floyd-Warshall |
| ~5,000 | O(n²) | a nested loop is fine, 2-D DP (**14.13**) |
| ~1,000,000 | O(n log n) | sorting, heaps, divide and conquer |
| ~10,000,000 | O(n) | one pass, hash map, two pointers |
| enormous | O(log n) or O(1) | binary search, direct formula |

> **Work backwards.** "n ≤ 10⁵ and I have an O(n²) idea — that is 10¹⁰ operations, far too slow. I need O(n log n)." Saying that out loud shows you can size a solution before writing it.

🔴 Equally: **n ≤ 20 is a strong hint that exponential is expected**, and reaching for something clever is wasted effort.

````python
# What the table means in seconds, at roughly 10^7 Python operations/sec.
import math

OPS_PER_SECOND = 10_000_000


def readable(seconds):
    if seconds < 1e-3:
        return "instant"
    if seconds < 60:
        return f"{seconds:.2f} s"
    if seconds < 3600:
        return f"{seconds / 60:.1f} min"
    if seconds < 86400 * 365:
        return f"{seconds / 3600:.1f} hours"
    return f"{seconds / (86400 * 365):.3g} years"


SIZES = (10, 20, 1_000, 100_000, 10_000_000)
COMPLEXITIES = [
    ("O(log n)", lambda n: math.log2(n)),
    ("O(n)", lambda n: n),
    ("O(n log n)", lambda n: n * math.log2(n)),
    ("O(n^2)", lambda n: n ** 2),
    ("O(2^n)", lambda n: 2 ** n if n <= 40 else float("inf")),
    ("O(n!)", lambda n: math.factorial(n) if n <= 20 else float("inf")),
]

print(f"{'complexity':<12}" + "".join(f"{n:>17,}" for n in SIZES))
print("-" * 97)
for label, formula in COMPLEXITIES:
    row = f"{label:<12}"
    for n in SIZES:
        operations = formula(n)
        row += f"{readable(operations / OPS_PER_SECOND):>17}" if operations != float("inf") \
            else f"{'never':>17}"
    print(row)

print("\n  Read down a column to see what is affordable at that size.")
print("  At n = 100,000: O(n log n) is instant and O(n^2) takes 17 minutes.")
print("  At n = 20: even O(n!) finishes - which is why n<=20 hints at")
print("  an exponential solution being expected.")
````

## What is actually being assessed

The correct answer is roughly a third of it. The rest:

| Signal | How to show it |
|---|---|
| **Communication** | narrate your thinking; state assumptions |
| **Complexity awareness** | give time *and* space, unprompted |
| **Trade-off judgement** | "this is O(n) time but O(n) space — worth it here because..." |
| **Edge cases** | empty, one element, duplicates, negatives, overflow |
| **Testing instinct** | walk your code through an example by hand |
| **Coachability** | take a hint gracefully and use it |

### Things that read badly

🔴 Long silence. Say what you are considering, even if you reject it.

🔴 Jumping to code before agreeing the approach.

🔴 Claiming a complexity you have not checked — especially forgetting that `x in list` is O(n) or that slicing copies (**14.2**).

🔴 Ignoring a hint. A hint means the interviewer wants you to succeed.

> **The most valuable habit in this whole folder:** when you propose a greedy solution, spend a moment trying to break it (**14.14**). Either you find the counter-example — and now you know you need DP — or you cannot, and you can sketch why. Both outcomes look like engineering judgement.

# Two warm-up problems

> These are the two problems from the original `Advance Coding.ipynb`, rewritten. In the original, one used `input()` — so it could never be run unattended — and neither had tests. They are good warm-ups: easy enough to finish quickly, with real edge cases.

### 1. Diagonal difference

*Given a square matrix, find the absolute difference between the sums of its two diagonals.*

```
    11   2   4          primary   (top-left to bottom-right): 11 + 5 + (-12) = 4
     4   5   6          secondary (top-right to bottom-left):  4 + 5 +  10   = 19
    10   8 -12          |4 - 19| = 15
```

**The insight:** both diagonals are reachable in **one pass** — for row `i`, the primary element is at column `i` and the secondary at column `n-1-i`. O(n) time, O(1) space, where the naive version builds two lists.

### 2. Fractions of positive, negative and zero

*Given an array, print the fractions that are positive, negative and zero, to six decimal places.*

🔴 The interesting parts are the edge cases: an **empty array** (division by zero) and **float formatting**.

````python
def diagonal_difference(matrix):
    """Absolute difference between the two diagonal sums. O(n) time, O(1) space.

    Row i contributes matrix[i][i] to the primary diagonal and
    matrix[i][n-1-i] to the secondary - so one loop does both.
    """
    n = len(matrix)
    primary = secondary = 0
    for i in range(n):
        primary += matrix[i][i]
        secondary += matrix[i][n - 1 - i]
    return abs(primary - secondary)


def diagonal_difference_naive(matrix):
    """The obvious version, for comparison - two passes and two lists."""
    n = len(matrix)
    primary = [matrix[i][i] for i in range(n)]
    secondary = [matrix[i][n - 1 - i] for i in range(n)]
    return abs(sum(primary) - sum(secondary))


cases = [
    ([[11, 2, 4], [4, 5, 6], [10, 8, -12]], 15),
    ([[1, 2], [3, 4]], 0),
    ([[5]], 0),
    ([], 0),
    ([[1, 0, 0], [0, 1, 0], [0, 0, 1]], 2),
]

print(f"{'matrix':<40}{'result':>8}{'expected':>10}{'ok':>5}")
print("-" * 64)
for matrix, expected in cases:
    got = diagonal_difference(matrix)
    label = str(matrix)[:38]
    print(f"{label:<40}{got:>8}{expected:>10}{str(got == expected):>5}")

print("\n  agrees with the two-pass version:",
      all(diagonal_difference(m) == diagonal_difference_naive(m) for m, _ in cases))
print("\n  🔴 The 3x3 identity matrix is the case worth noticing: the two")
print("     diagonals SHARE the centre element when n is odd. That is fine")
print("     here - both sums include it - but it is the kind of overlap")
print("     worth stating out loud.")
````

````python
def fraction_report(values):
    """Fractions that are positive, negative and zero. O(n) time, O(1) space."""
    if not values:
        raise ValueError("cannot compute fractions of an empty array")
    positive = negative = zero = 0
    for value in values:
        if value > 0:
            positive += 1
        elif value < 0:
            negative += 1
        else:
            zero += 1
    total = len(values)
    return positive / total, negative / total, zero / total


samples = [
    [-4, 3, -9, 0, 4, 1],
    [1, 1, 0, -1, -1],
    [0, 0, 0],
    [7],
]
for values in samples:
    positive, negative, zero = fraction_report(values)
    print(f"  {str(values):<24} +{positive:.6f}  -{negative:.6f}  0:{zero:.6f}")
    assert abs(positive + negative + zero - 1.0) < 1e-9

print("\n  the three fractions always sum to 1.0 (asserted above)")

try:
    fraction_report([])
except ValueError as exc:
    print(f"\n  🔴 empty input: ValueError: {exc}")
    print("     The original version would have raised ZeroDivisionError.")
    print("     Deciding what an empty input MEANS - error, or all zeros? -")
    print("     is exactly the kind of question to ask before coding.")

print("\n  and a float-formatting trap worth knowing:")
print(f"    1/3 printed with .6f  -> {1 / 3:.6f}")
print(f"    0.1 + 0.2 == 0.3      -> {0.1 + 0.2 == 0.3}")
print(f"    0.1 + 0.2             -> {0.1 + 0.2!r}")
print("    ^ never compare floats with == ; compare with a tolerance (02)")
````

````python
# A harness for practising: write a brute force, then verify the fast version.
import random


def verify(fast, slow, generate, trials=500, seed=15):
    """Compare two implementations on random inputs.

    🔴 This is the single most useful habit in the folder. A brute force
    you trust plus a random generator will find bugs that hand-picked
    examples never do - and it is how 14.9, 14.13 and 14.14 caught mine.
    """
    rng = random.Random(seed)
    for trial in range(trials):
        case = generate(rng)
        expected = slow(case)
        actual = fast(case)
        if actual != expected:
            return False, case, expected, actual
    return True, None, None, None


# Example: 'maximum subarray' - Kadane against exhaustive search
def max_subarray_fast(data):
    if not data:
        return 0
    best = current = data[0]
    for value in data[1:]:
        current = max(value, current + value)
        best = max(best, current)
    return best


def max_subarray_slow(data):
    if not data:
        return 0
    return max(sum(data[i:j])
               for i in range(len(data))
               for j in range(i + 1, len(data) + 1))


def make_array(rng):
    return [rng.randint(-20, 20) for _ in range(rng.randint(1, 12))]


ok, case, expected, actual = verify(max_subarray_fast, max_subarray_slow, make_array)
print(f"max_subarray: 500 random trials -> {'all correct' if ok else 'MISMATCH'}")
if not ok:
    print(f"  failing case: {case}, expected {expected}, got {actual}")


# And a deliberately BROKEN version, to show the harness earning its keep
def max_subarray_broken(data):
    """Returns 0 for all-negative input - the classic bug (14.3)."""
    best = current = 0
    for value in data:
        current = max(0, current + value)
        best = max(best, current)
    return best


ok, case, expected, actual = verify(max_subarray_broken, max_subarray_slow, make_array)
print(f"\nthe broken version: {'all correct' if ok else 'MISMATCH found'}")
if not ok:
    print(f"  input    : {case}")
    print(f"  expected : {expected}")
    print(f"  got      : {actual}")
    print("\n  🔴 Found in well under a second, on an input nobody would have")
    print("     thought to try. Hand-picked test cases would very likely")
    print("     have missed this.")
````

## Common Mistakes & Pitfalls

1. 🔴 **Coding before agreeing the approach.** State the brute force and its complexity first.
2. 🔴 **Silence.** Narrate what you are considering, including what you reject.
3. 🔴 **Claiming a complexity you have not checked.** `x in list` is O(n); slicing copies (**14.2**).
4. 🔴 **Not asking about constraints.** n's size tells you the intended complexity.
5. **Ignoring a hint.** It means the interviewer wants you to get there.
6. **Skipping edge cases.** Empty, one element, duplicates, negatives, all-negative.
7. **Optimising before it is correct.** A working O(n²) beats a broken O(n).
8. **Memorising solutions rather than patterns.** The costume changes; the pattern does not.
9. **Forgetting to state space complexity.** Volunteering it reads well.

## Best Practices

- Restate the problem in your own words before starting.
- Work one small example by hand, then one edge case.
- Give the brute force and its complexity, then improve it.
- Say time *and* space complexity unprompted.
- Use the pattern table's three questions: sorted? searching in a loop? choices interact?
- Size the solution from the constraints before writing it.
- Test by walking your code through an example line by line.
- Verify against a brute force on random inputs whenever you can.
