### Division: `/` vs `//` vs `%`

This is one of the few places where Python 3 differs sharply from both Python 2 and from
C-family languages, so it is worth being precise.

| Operator | Name | `10 ? 3` | Returns |
|---|---|---|---|
| `/` | True division | `3.333...` | **Always a float**, even for `10 / 2` |
| `//` | Floor division | `3` | Floors the result toward negative infinity |
| `%` | Modulo | `1` | The remainder; sign follows the **divisor** |
| `**` | Exponent | `1000` | Right-associative |

⚠️ **The negative-number trap.** Python floors, C truncates:

| Expression | Python | C / Java |
|---|---|---|
| `-7 // 2` | `-4` | `-3` |
| `-7 % 2` | `1` | `-1` |

Python's choice keeps the identity `(a // b) * b + (a % b) == a` true for all signs, and
makes `%` always return a result with the divisor's sign — which is why `-1 % 12` gives
`11`, exactly what you want for clock arithmetic and circular buffers.

````python
print("True division  10 / 3   =", 10 / 3)      # always a float
print("True division  10 / 2   =", 10 / 2)      # 5.0, NOT 5
print("Floor division 10 // 3  =", 10 // 3)     # int in, int out
print("Floor division 10.0 // 3 =", 10.0 // 3)  # float in, float out

# Negative numbers: Python FLOORS (rounds toward negative infinity)
print("\n -7 // 2  =", -7 // 2, "  (C and Java give -3)")
print(" -7 %  2  =", -7 % 2,  "  (sign follows the DIVISOR)")
print("  7 % -2  =", 7 % -2)

# The identity that always holds:
a, b = -7, 2
print(f"\n(a // b) * b + (a % b) == a  ->  {(a // b) * b + (a % b) == a}")

# divmod() gives you both at once - one operation instead of two
print("\ndivmod(17, 5) =", divmod(17, 5))

# Classic use: converting a total into units
total_seconds = 3725
minutes, seconds = divmod(total_seconds, 60)
hours, minutes = divmod(minutes, 60)
print(f"{total_seconds}s = {hours}h {minutes}m {seconds}s")

# Exponentiation is RIGHT-associative
print("\n2 ** 3 ** 2 =", 2 ** 3 ** 2, " (that is 2**(3**2) = 2**9, not (2**3)**2)")
````

**Real-world use case:** `divmod()` is the standard way to break a total into units —
seconds into h/m/s, paise into rupees, items into pages. Doing it with `//` and `%`
separately performs the same division twice.

### Python Assignment Operators:
- In python, the operators that are used to assign values to variables are called assignment operators. The major assignment operator is an equal sign (=). 
<br/><br/>
    - 1. **=:** Assigns value at the right side to the variable at left. **Eg:** a = 100, assigns 100 to variable a.
<br/><br/>
    - 2. **+=:** Adds the operand at the right side to the variable at left and assigns the result to the variable at left. **Eg:** a += 1, equivalent to a = a + 1.
<br/><br/>
    - 3. **-=:** Subtracts the operand at the right side from the variable at left and assigns the result to the variable at left. **Eg:** a -= 1, equivalent to a = a – 1.
<br/><br/>
    - 4. ***=:** Multiplies the operand at the right side to the variable at left and assigns the result to the variable at left. **Eg:** a *= 2, equivalent to a = a * 2.
<br/><br/>
    - 5. **/=:** Divides the variable at the left side by the operand at right and assigns the result to the variable at left. **Eg:** a /= 2, equivalent to a = a / 2.
<br/><br/>
    - 6. **%=:** Performs modulo operation on two operands and assigns the result to the variable at left. **Eg:** a %= 2, equivalent to a = a % 2.
<br/><br/>
    - 7. ****=:** Performs an exponential operation on the operands and assigns the result to the variable at left. **Eg:** a ***= 2, equivalent to a = a ** 2.
<br/><br/>
    - 8. **//=:** Performs floor division on the operands and assigns the resulting value to the left variable. **Eg:** a //= 2, equivalent to a = a // 2.

### The walrus operator `:=` (assignment expression)

> **Version note:** added in **Python 3.8** (PEP 572).

Normal assignment (`x = 5`) is a **statement** — it does something but has no value, which
is why `if x = 5:` is a `SyntaxError` (a deliberate design choice, to prevent the classic
C bug of typing `=` when you meant `==`).

The **walrus operator** `:=` is an assignment *expression*: it assigns **and** evaluates to
the assigned value, so it can be used where a value is expected.

```
if (n := len(data)) > 5:
     |   |    |
     |   |    +-- the expression whose value we want
     |   +------- the name it gets bound to
     +----------- := assigns AND returns, so the comparison sees the value
```

Use it when it removes a genuine duplicated call. Do not use it to compress unrelated
statements onto one line — readability first.

````python
# Problem: you need the value twice - once to test it, once to use it.
# Without the walrus, you either call it twice or add an extra line.

data = [1, 2, 3, 4, 5, 6, 7, 8]

# Before:
n = len(data)
if n > 5:
    print(f"List is long ({n} items)")

# After - assign and test in one expression:
if (n := len(data)) > 5:
    print(f"List is long ({n} items)")

# Where it genuinely shines: loops that must read before they can test.
# Without walrus you need input() twice, or a while True + break.
# while (line := input("Say something (or 'quit'): ")) != "quit":
#     print(f"  you said: {line}")

# In comprehensions, to avoid computing an expensive value twice:
numbers = [1, 2, 3, 4, 5, 6]
results = [y for x in numbers if (y := x * x) > 10]
print("\nSquares over 10:", results)

# Reading a file in chunks - the canonical real-world example:
# while (chunk := file.read(8192)):
#     process(chunk)
````

### Python Comparison (Relational) Operators:
- Also called Relational operators, they compare the values on either side of the operand and decide the relation among them.
    - Return True or False based on the relation.
    - i.e, Comparison operators compare two values and evaluate down to a single Boolean value.
<br/><br/>
- The comparison operators allowed in python are:
    - 1. **== (Equal to):** Returns True if the operands at both sides are equal, returns False otherwise. **Eg:** a == b.
<br/><br/>
    - 2. **!= (Not equal to):** Returns True if the operands are not equal, returns False otherwise. **Eg:** a != b.
<br/><br/>
    - 3. **> (Greater than):** Returns True if the left operand is greater than the operand at right, returns False otherwise. **Eg:** a > b.
<br/><br/>
    - 4. **>= (Greater than or equal to):** Returns True if the left operand is either greater than or equal to the operand at right, returns False otherwise. **Eg:** a >= b.
<br/><br/>
    - 5. **< (Less than):** Returns True if the left operand is less than the operand at right, returns False otherwise. **Eg:** a < b.
<br/><br/>
    - 6. **<= (Less than or equal to):** Returns True if the left operand less than or equal to the operand at right, returns False otherwise. **Eg:** a <= b.

````python
num_1, num_2= 20, 3

print(f'Is {num_1} equal to {num_2}: {num_1 == num_2}')

print(f'Is {num_1} not equal to {num_2}: {num_1 != num_2}')

print(f'Is {num_1} greater than {num_2}: {num_1 > num_2}')

print(f'Is {num_1} greater than or equal to {num_2}: {num_1 >= num_2}')

print(f'Is {num_1} less than {num_2}: {num_1 < num_2}')

print(f'Is {-num_1} less than or equal to {num_2}: {-num_1 <= num_2}')
````

### Chained comparisons

Python lets you chain comparison operators the way mathematics does. `18 <= age <= 65`
means exactly `age >= 18 and age <= 65` — but it reads better *and* evaluates `age` only
once, which matters when the middle term is a function call.

Most languages do not have this. In C or JavaScript, `1 < x < 10` parses as `(1 < x) < 10`,
which compares a boolean against 10 and is almost never what you meant.

````python
age = 25

# The verbose way
print(age >= 18 and age <= 65)

# The Pythonic way - chained comparison, reads like mathematics
print(18 <= age <= 65)

# Chains can be longer, and mix operators
x = 5
print(1 < x < 10 < 100)
print(0 <= x != 7)

# Chaining also short-circuits: the middle value is evaluated only ONCE
def get_value():
    print("  get_value() called")
    return 5

print("\n1 < get_value() < 10 :")
print(" ", 1 < get_value() < 10)     # get_value() runs a single time
````

### Python Logical Operators:
- The logical operators in python allow us to combine two or more conditional operators and return True or False based on the results of those conditional operators combined with logical operators. 
<br/><br/>
- The logical operators in python are:
    - 1. **and:** Returns True if both the conditions are True otherwise returns False. **Eg:** (a > 10) and (a < 20).
<br/><br/>
    - 2. **or:** Returns True if one of the conditions is True, otherwise returns False. **Eg:** (a > 10) or (a < 20).
<br/><br/>
    - 3. **not:** Also called negation, this operator is used to reverse the logical state of its operand. **Eg:** not (a > 10).

````python
num1, num2 = 20, 3
print(f'{(num1>15)} logical_and {(num2<5)} is {(num1>15) and (num2<5)}')
print(f'{(num1<15)} logical_and {(num2<5)} is {(num1<15) and (num2<5)}')

print(f'{(num1>15)} logical_or {(num2<5)} is {(num1>15) or (num2<5)}')
print(f'{(num1<15)} logical_or {(num2<5)} is {(num1<15) or (num2<5)}')

print(f'logical_not of {num1>15} is {not(num1>15)}')
print(f'logical_not of {num1<15} is {not(num1<15)}')
````

### Short-circuit evaluation

Python's `and` / `or` are **lazy**: they evaluate the left side, and only touch the right
side if they still need to.

- `A and B` — if `A` is falsy, the answer is already `A`. `B` is never evaluated.
- `A or B` — if `A` is truthy, the answer is already `A`. `B` is never evaluated.

Two consequences worth knowing:

1. **They enable guard clauses.** `if user is not None and user.is_active:` is safe —
   if `user` is `None`, the second half never runs, so there is no `AttributeError`.
2. **They return an operand, not a boolean.** `5 and 3` is `3`, not `True`. This is what
   makes `name or "anonymous"` work as a default-value idiom.

````python
def expensive_check():
    print("  ...expensive_check() actually ran")
    return True

# `and` stops at the first falsy value - the right side never runs
print("False and expensive_check():")
result = False and expensive_check()
print("  result:", result)

# `or` stops at the first truthy value
print("\nTrue or expensive_check():")
result = True or expensive_check()
print("  result:", result)

# This is what makes the "guard" idiom safe:
data = None
if data is not None and len(data) > 0:      # len(None) would raise TypeError
    print("has data")
else:
    print("\nGuard worked - len() was never called on None")

# They return an OPERAND, not a boolean
print("\n5 and 3   ->", 5 and 3)        # 3   (last truthy value)
print("0 or 'hi'  ->", 0 or "hi")        # hi  (first truthy value)
print("None or [] ->", None or [])       # []  (last value, all falsy)

# The common idiom this enables: a default for a possibly-empty value
username = ""
display = username or "anonymous"
print("display   ->", display)
````

4. **^:** Performs bitwise XOR(Exclusive OR) operation. **Eg:** a ^ b.

### `is` vs `==` — the distinction that actually matters

| | `==` | `is` |
|---|---|---|
| **Asks** | Do these have the same *value*? | Are these the *same object* in memory? |
| **Calls** | `__eq__()` | Compares `id()` |
| **Use for** | Almost everything | `None`, `True`, `False` |

An analogy: `==` asks *"are these two banknotes the same denomination?"*;
`is` asks *"are these the same physical banknote?"*

> ⚠️ `x is 10` and `x is "hello"` raise a **`SyntaxWarning`** on Python 3.8+. They may
> appear to work because of small-integer caching and string interning (see **1.2**), but
> that is an implementation detail you must never rely on.

**Real-world use case:** `if result is None:` is the standard way to check "did this
function return nothing?" — and it is correct precisely *because* there is only ever one
`None` object in a running program.

````python
import math

a = [1, 2, 3]
b = [1, 2, 3]
c_ref = a                      # a NAME for the same object, not a copy

print(f"a == b     : {a == b}      (equal values)")
print(f"a is b     : {a is b}      (different objects)")
print(f"a is c_ref : {a is c_ref}  (same object)")

# Mutating through one name is visible through the other - this is the whole point
c_ref.append(4)
print(f"a after c_ref.append(4): {a}")

# The classic float trap
print(f"\n0.1 + 0.2          = {0.1 + 0.2!r}")
print(f"0.1 + 0.2 == 0.3   = {0.1 + 0.2 == 0.3}")
print(f"math.isclose(...)  = {math.isclose(0.1 + 0.2, 0.3)}")

# For money, floats are the wrong tool entirely
from decimal import Decimal
print(f"\nDecimal: {Decimal('0.1') + Decimal('0.2') == Decimal('0.3')}")
````

## Common Mistakes & Pitfalls

1. **Using `is` to compare values.** `x is 10` raises a `SyntaxWarning` on 3.8+ and gives unreliable answers. `is` asks *are these the same object*; `==` asks *are these equal*. Reserve `is` for `None`, `True`, `False`.
2. **Expecting `/` to give an integer.** In Python 3, `10 / 2` is `5.0`, a float. Use `//` for integer division.
3. **Assuming `-7 // 2` is `-3`.** Python floors toward negative infinity, so it is `-4`. C and Java truncate toward zero and give `-3`. This bites when porting algorithms.
4. **Confusing `%` the modulo operator with `%` the (legacy) string-formatting operator.** Same symbol, unrelated jobs.
5. **Floating-point equality.** `0.1 + 0.2 == 0.3` is `False`. Compare with `math.isclose()`, or use `decimal.Decimal` for money.
6. **Confusing bitwise `&`/`|` with logical `and`/`or`.** `True & False` happens to work, but `5 and 3` gives `3` while `5 & 3` gives `1`. They are different operators.
7. **Assigning inside a condition by accident.** Python makes this hard on purpose: `if x = 5:` is a `SyntaxError`. If you genuinely mean it, use the walrus `:=`.

## Best Practices

- Use parentheses to make precedence explicit even when they are technically unnecessary — `(a * b) + c` costs nothing and removes doubt.
- Use `==` for values, `is` only for `None` / `True` / `False`.
- Use chained comparisons (`0 <= x <= 100`) instead of `x >= 0 and x <= 100`.
- Use `math.isclose()` for float comparison; `decimal.Decimal` for currency.
- Reach for the walrus `:=` only when it removes a genuine duplicate call — never to cram two ideas onto one line.
- Put spaces around binary operators (`a + b`, not `a+b`) per PEP 8.
