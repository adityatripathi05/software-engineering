**Given a square matrix, calculate the absolute difference between the sums of its diagonals.**

Sample Input:

3

11 2 4

4 5 6

10 8 -12

Sample Output:

15

```python
import math

n = int(input().strip())
arr = []
for z in range(n):
    arr.append(list(map(int, input().rstrip().split())))
pd, sd = 0, 0
for i in range(len(arr)): 
    for j in range(len(arr)):
        if (i == j):
            pd = pd + arr[i][j]
for i in range(len(arr)): 
    for j in range(len(arr)):
        if (i+j == len(arr)-1):
            sd = sd + arr[i][j]
print(abs(pd-sd))
```

**Given an array of integers, calculate the fractions of its elements that are positive, negative, and are zeros. Print the decimal value of each fraction on a new line. (answer must be upto 6 places of decimal only)**

Sample Input:

6

-4 3 -9 0 4 1 

Sample Output:

0.500000

0.333333

0.166667

```python
n = int(input())
arr = list(map(int, input().rstrip().split()))
pc, nc, zc = 0, 0 ,0
for i in arr:
    if (i<0):
        nc = nc+1
    elif(i>0):
        pc = pc+1
    else:
        zc = zc +1
    
print(format(pc/len(arr),'.6f'))
print(format(nc/len(arr),'.6f'))
print(format(zc/len(arr),'.6f'))
```
