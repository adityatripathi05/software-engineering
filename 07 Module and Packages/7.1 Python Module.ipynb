## Module in Python:
- All Python programs can call a basic set of functions called built-in functions, including the print(), input(), and len() functions you’ve seen before. These functions are defined inside a **builtins module**. 
- Python also comes with a set of modules called the **standard library**.
- Each module is a Python program that contains a related group of functions that can be embedded in our programs.
- A module is a python file that has collection of identifiers that have been bundled together in a single python file. 
    - i.e, grouping of similar code for easy access.

***
### Advantages of using Module:
- Used for code organization, packaging and reusability
    - **Reusability:** Module can be used in some other python code. Hence it provides the facility of code reusability.
- **Categorization:** Similar type of attributes can be placed in one module.
- **Security**
- **Execution time:** Is fast

- **NOTE: It saves the .pyc file, so that module doesn't run everytime for compilation. It just run .pyc file.**

***
### Types of Module:
- There are two types of module in Python:
    - 1) **Standard Library Modules:** Some of them are as follows: math, random , threading , collections , os , mailbox , string , time , tkinter etc.
        - Ref. <https://docs.python.org/3/py-modindex.html>
    - 2) **User Defined Modules**

### Create a module:
- To create a module just save the code you want in a file with the file extension `.py`.


### Use a module:
- Before we can use the functions in a module, we must load that module in our python code.
- Python provides two types of statements to load a module as defined below:
    - 1. **import statement:** The import statement is used to import all the functionality of one module.
        - **Syntax:** `import module1,module2,........ module n`
    <br/><br/>
    - **NOTE: We can import multiple modules with a single import statement, but a module is loaded once regardless of the number of times, it has been imported into our file.**
    <br/><br/>
    - 2) **from-import statement:** Instead of importing the whole module into the namespace, python provides the flexibility to import only the specific attributes of a module.
        - **Syntax:** `from <module-name> import <name 1>, <name 2>..,<name n>`
<br/><br/>
- The from...import statement is always better to use if we know the attributes to be imported from the module in advance. It doesn't let our code to be heavier.
- **NOTE:** Using from-import statement, we can also import all the attributes from a module by using *
    - `from module_name import *`
- **NOTE: When importing using the from keyword, do not use the module name when referring to elements in the module.**


### Renaming a module:
- We can create an alias when we import a module, by using the `as` keyword.
- **Syntax:** `import <module-name> as <specific-name>` 

### Python Module Attributes:
- Ref. <https://docs.python.org/3/reference/import.html#import-related-module-attributes>

#### `__name__` Attribute: 
- This attribute returns the name of the module. This name is used to uniquely identify the module in the import system.
- By default, the name of the file (excluding the extension .py) is the value of `__name__` attribute.
    - However, value of the `__name__` attribute is set to `__main__` when checked on the interpreter.
    - when the module is imported, its `__name__` is set to its filename.

#### `__doc__` Attribute:
- The `__doc__` attribute denotes the documentation string (docstring) line written in a module code.
     - i.e, `__doc__` attribute will return a string defined at the beginning of the module code.
     
     
#### `__file__` Attribute:
- `__file__` is an optional attribute which holds the name and path of the module file from which it is loaded.


#### `__dict__` Attribute:
- The `__dict__` attribute will return a dictionary object of module attributes, functions and other definitions and their respective values.
- dir() is a built-in function that also returns the list of all attributes and functions in a module.

### Locating Modules:
- When you import a module, the Python interpreter searches for the module in a given list of directories in following sequences−
    - 1) The current directory.
    - 2) If the module isn't found, Python then searches each directory in the shell variable PYTHONPATH.
    - 3) If all else fails, Python checks in the list of installation-dependent default path (e.g., /usr/lib/python).
<br/><br/>    
- The list of directories searched by Python is given by the ```sys.path``` variable.
- Modules must be located in the search path, therefore we can:
    - write our own modules within directories already defined in the search path (e.g. ...local/lib/python3.7/dist-packages).
    - modify the environment variable PYTHONPATH to include the directories containing the user-defined modules.
<br/><br/> 
- **To modify the module path:**
```python
import sys
new_path = <module path>
if new_path not in sys.path:
    sys.path.append(new_path)
```

### How importing actually works

Three things happen the first time you `import x`:

1. **Search** — Python walks `sys.path` in order, looking for `x`.
2. **Execute** — the module's top-level code runs, **once**.
3. **Cache** — the module object is stored in `sys.modules`. Every later `import x` in the
   same process is just a dictionary lookup.

That second point matters: **anything at module level runs on import**, which is why
executable code belongs behind a `__main__` guard.

### The four import forms

| Form | Binds | Use when |
|---|---|---|
| `import math` | `math` | The default — provenance is obvious |
| `import numpy as np` | `np` | Long names, or a community convention |
| `from math import sqrt` | `sqrt` | You use one or two names heavily |
| `from math import *` | ⚠️ everything | Interactive sessions only |

### `__name__` and the `__main__` guard

Every module has a `__name__`. It is `"__main__"` when the file is run directly, and the
module's own name when it is imported. That single fact is what makes a file usable as both
a script and a library:

```python
def main():
    ...

if __name__ == "__main__":
    main()
```

````python
import sys

# ---- Modules are cached: top-level code runs ONCE per process ----
print("json already imported?", "json" in sys.modules)
import json
print("after import        :", "json" in sys.modules)
import json                                     # no re-execution
print("importing again is a cheap dict lookup\n")


# ---- The four import forms ----
import math                        # bind the module
from math import sqrt              # bind one name
from math import sqrt as sq        # bind under another name
import math as m                   # bind the module under another name

print("math.sqrt(16) :", math.sqrt(16))
print("sqrt(16)      :", sqrt(16))
print("sq(16)        :", sq(16))
print("m.sqrt(16)    :", m.sqrt(16))


# ---- ⚠️ from module import * ----
# Pulls every public name into your namespace. It hides provenance and can
# silently overwrite your own names.
gamma = "my important value"
from math import *                 # math also defines `gamma`
print("\nafter 'from math import *', gamma =", gamma, " <- overwritten!")


# ---- __all__ controls what `import *` exports ----
print("\n__all__ is a module's declared public surface.")
print("os.__all__ has", len(getattr(__import__('os'), '__all__', [])), "names")


# ---- __name__: 'the module knows how it was loaded' ----
print("\nthis notebook's __name__ :", __name__)
print("math.__name__            :", math.__name__)

print("""
In a .py file:
    __name__ == "__main__"      when you run it directly
    __name__ == "modulename"    when it is imported

That is why executable code goes behind:

    if __name__ == "__main__":
        main()

Without the guard, importing your script RUNS it.
""")
````

#### deque()
- A deque object support appends and pops from either ends of a list. 
- It is more memory efficient than a normal list object. 
- In a normal list object, the removal of any item causes all items to the right to be shifted towards left by one index. Hence, it is very slow.

````python
q= deque([10,20,30,40])
q.appendleft(0)
print(q)
````

## `pathlib` — the modern replacement for `os.path`

> **Version note:** `pathlib` arrived in **Python 3.4** and is now the recommended way to
> work with the filesystem. The `os.path` functions above still work and you will meet them
> constantly in existing code — but **write `pathlib` in new code.**

The difference is that `os.path` treats paths as **strings** you pass to functions, while
`pathlib` treats a path as an **object** with methods and attributes. Everything gets
shorter, and the platform separator stops being your problem.

### Translation table

| `os.path` | `pathlib` |
|---|---|
| `os.path.join(a, b, c)` | `Path(a) / b / c` |
| `os.path.abspath(p)` | `Path(p).resolve()` |
| `os.path.dirname(p)` | `Path(p).parent` |
| `os.path.basename(p)` | `Path(p).name` |
| `os.path.splitext(p)[0]` | `Path(p).stem` |
| `os.path.splitext(p)[1]` | `Path(p).suffix` |
| `os.path.exists(p)` | `Path(p).exists()` |
| `os.path.isfile(p)` / `isdir(p)` | `Path(p).is_file()` / `.is_dir()` |
| `os.path.expanduser('~')` | `Path.home()` |
| `os.listdir(d)` | `Path(d).iterdir()` |
| `os.makedirs(d, exist_ok=True)` | `Path(d).mkdir(parents=True, exist_ok=True)` |
| `os.remove(p)` | `Path(p).unlink()` |
| `glob.glob(pat)` | `Path(d).glob(pat)` |
| `open(p).read()` | `Path(p).read_text()` |

### The headline: `/` joins paths

```python
config = Path.home() / ".config" / "myapp" / "settings.toml"
```

That reads like a path, works identically on Windows and Linux, and needs no import of
`os.sep` or `os.path.join`.

**Real-world use case:** any script that touches files. `pathlib` removes an entire category
of Windows/Linux separator bugs, and `Path.read_text()` / `write_text()` cover most small-file
I/O in one line (full file handling is **08**).

````python
from pathlib import Path

# ---- Building paths: the / operator, instead of os.path.join ----
base = Path.cwd()
target = base / "data" / "reports" / "q4.csv"
print("built path   :", target)
print("as posix     :", target.as_posix())

# ---- The parts of a path are attributes, not function calls ----
print("\nname         :", target.name)
print("stem         :", target.stem)
print("suffix       :", target.suffix)
print("parent       :", target.parent.name)
print("parents      :", [p.name for p in list(target.parents)[:3]])
print("parts        :", target.parts[-3:])

# ---- Modifying a path ----
print("\nwith_suffix  :", target.with_suffix(".parquet").name)
print("with_name    :", target.with_name("q1.csv").name)
print("with_stem    :", target.with_stem("annual").name)

# ---- Tests ----
here = Path.cwd()
print("\nexists       :", here.exists())
print("is_dir       :", here.is_dir())
print("is_file      :", here.is_file())
print("is_absolute  :", here.is_absolute())

# ---- Home and expansion ----
print("\nhome         :", Path.home())
print("expanduser   :", Path("~/local/bin").expanduser())

# ---- Reading and writing: no open() needed for small files ----
tmp = Path("pathlib_demo.txt")
tmp.write_text("line one\nline two\n", encoding="utf-8")
print("\nread_text    :", repr(tmp.read_text(encoding="utf-8")))
print("size         :", tmp.stat().st_size, "bytes")
tmp.unlink()
print("deleted      :", not tmp.exists())

# ---- Searching: glob and rglob ----
notebooks = sorted(Path("..").glob("*/*.ipynb"))[:3]
print("\nglob '*/*.ipynb' (first 3):")
for nb in notebooks:
    print("  ", nb.parent.name, "/", nb.name)

# rglob recurses to any depth
count = sum(1 for _ in Path("..").rglob("*.ipynb"))
print(f"\nrglob found {count} notebooks in the whole tree")

# ---- mkdir with parents, and cleanup ----
nested = Path("demo_a/demo_b/demo_c")
nested.mkdir(parents=True, exist_ok=True)
print("\ncreated nested:", nested.exists())
for p in [nested, nested.parent, nested.parent.parent]:
    p.rmdir()
print("removed       :", not Path("demo_a").exists())
````

### `importlib`, and the removal of `imp`

> ### 🔴 Version note
> **The `imp` module was removed in Python 3.12.** It had been deprecated since 3.4. Any
> tutorial or codebase using `imp.reload()` or `imp.load_source()` will fail outright on a
> modern interpreter. The replacement is **`importlib`**.

`importlib` is the import system, exposed as a normal library. Three things it is good for:

| Task | Call |
|---|---|
| Import a module whose **name is a string** | `importlib.import_module("json")` |
| Check whether something **is installed** | `importlib.util.find_spec("numpy")` |
| **Reload** a module you have edited | `importlib.reload(mod)` |

The first is how plugin systems work: the module name comes from a config file, not from
your source.

````python
import importlib
import importlib.util          # submodules are not auto-imported
import sys
from pathlib import Path

# ---- importlib: importing by NAME, decided at run time ----
name = "statistics"
mod = importlib.import_module(name)
print(f"imported {name!r} dynamically:", mod.mean([1, 2, 3, 4]))

# Useful for plugin systems: the module name comes from config, not source
for name in ["math", "random", "string"]:
    m = importlib.import_module(name)
    public = [n for n in dir(m) if not n.startswith("_")]
    print(f"  {name:<8} exposes {len(public):>3} public names")

# ---- Checking whether something is installed, without importing it ----
for name in ["json", "numpy", "definitely_not_a_real_module"]:
    spec = importlib.util.find_spec(name)
    print(f"\n  {name:<30} installed: {spec is not None}")

# ---- importlib.reload(): pick up a changed module without restarting ----
demo = Path("reload_demo.py")
demo.write_text("VALUE = 'first'\n", encoding="utf-8")
sys.path.insert(0, str(Path.cwd()))

import reload_demo
print("\nfirst load :", reload_demo.VALUE)

demo.write_text("VALUE = 'second'\n", encoding="utf-8")
import reload_demo                      # cached - does NOT pick up the change
print("re-import  :", reload_demo.VALUE, " <- still cached")

importlib.reload(reload_demo)
print("after reload:", reload_demo.VALUE)

# cleanup
del sys.modules["reload_demo"]
sys.path.remove(str(Path.cwd()))
demo.unlink()

print("""
⚠️ The `imp` module was REMOVED in Python 3.12.
   Old code using imp.reload() / imp.load_source() must move to importlib.
""")
````

### Circular imports

Two modules that import each other produce:

```
ImportError: cannot import name 'X' from partially initialized module 'Y'
             (most likely due to a circular import)
```

**Why:** module `orders` starts executing, hits `from customers import ...`, which starts
executing `customers`, which hits `from orders import ...` — but `orders` is only *partially*
built, so the name it wants does not exist yet.

Three fixes, in increasing order of how much you should prefer them:

| Fix | How | Verdict |
|---|---|---|
| Import the **module**, not the name | `import customers` then `customers.get(...)` | Works — resolution is deferred to call time |
| Move the import **inside the function** | `def f(): from customers import get` | Works, but hides a dependency |
| **Extract the shared concept** into a third module | both import `common` | ✅ The real fix — the cycle was a design signal |

````python
import shutil, sys, tempfile, textwrap
from pathlib import Path

work = Path(tempfile.mkdtemp(prefix="circular_demo_"))
sys.path.insert(0, str(work.resolve()))

# ---- The broken version: two modules importing each other at top level ----
(work / "orders.py").write_text(textwrap.dedent("""
    from customers import get_customer

    def describe(order_id):
        return f"order {order_id} for {get_customer(order_id)}"
"""), encoding="utf-8")

(work / "customers.py").write_text(textwrap.dedent("""
    from orders import describe

    def get_customer(order_id):
        return "Aditya"
"""), encoding="utf-8")

try:
    import orders
except ImportError as exc:
    print("circular import fails:")
    print("  ", exc)

for name in ("orders", "customers"):
    sys.modules.pop(name, None)


# ---- Fix 1: import the MODULE, not the name (resolution is deferred) ----
(work / "orders.py").write_text(textwrap.dedent("""
    import customers

    def describe(order_id):
        return f"order {order_id} for {customers.get_customer(order_id)}"
"""), encoding="utf-8")

(work / "customers.py").write_text(textwrap.dedent("""
    import orders

    def get_customer(order_id):
        return "Aditya"
"""), encoding="utf-8")

import orders
print("\nfix 1 (import the module):", orders.describe(101))

for name in ("orders", "customers"):
    sys.modules.pop(name, None)


# ---- Fix 2: move the import INSIDE the function ----
(work / "orders.py").write_text(textwrap.dedent("""
    def describe(order_id):
        from customers import get_customer      # imported at CALL time
        return f"order {order_id} for {get_customer(order_id)}"
"""), encoding="utf-8")

import orders
print("fix 2 (local import)     :", orders.describe(102))

# ---- Fix 3 (the real one): extract the shared piece into a third module ----
print("""
fix 3 (the proper fix)    : if two modules need each other, they usually
                            share a concept that belongs in a third module.
                            Extract it and have both import that.
""")

# cleanup
for name in ("orders", "customers"):
    sys.modules.pop(name, None)
sys.path.remove(str(work.resolve()))
shutil.rmtree(work)          # rmtree: importing leaves a __pycache__ behind
print("cleaned up:", not work.exists())
````

````python
import shutil, tempfile
from pathlib import Path

# NOTE: cells 94-96 originally used hardcoded C:\Users\Aditya\... paths.
# Rewritten to work anywhere.

work = Path(tempfile.mkdtemp(prefix="shutil_demo_"))
source = work / "report.txt"
source.write_text("quarterly numbers\n", encoding="utf-8")

extra = work / "Extra"
extra.mkdir()

# ---- shutil.copy(src, dst): copy a file ----
shutil.copy(source, extra / "report_copy.txt")
print("after copy :", sorted(p.name for p in work.rglob("*")))
print("original still there:", source.exists())
````

````python
# ---- shutil.move(src, dst): move a file or directory ----
extra2 = work / "Extra2"
extra2.mkdir()

shutil.move(str(source), str(extra2 / "report.txt"))
print("after move :", sorted(p.name for p in work.rglob("*")))
print("original gone:", not source.exists())

# ---- shutil.copytree(): copy a whole directory tree ----
shutil.copytree(extra, work / "ExtraBackup")
print("\nafter copytree:", sorted(p.name for p in work.rglob("*")))

# ---- shutil.disk_usage(): a handy extra ----
usage = shutil.disk_usage(work)
print(f"\ndisk: {usage.used / 1e9:.1f} GB used of {usage.total / 1e9:.1f} GB")
````

````python
# ---- shutil.rmtree(): recursively delete a directory tree ----
# This is the answer to os.rmdir()'s refusal to remove non-empty directories.
print("before rmtree:", work.exists(), "-", len(list(work.rglob("*"))), "entries")

shutil.rmtree(work)
print("after rmtree :", work.exists())

print("""
⚠️ shutil.rmtree() is genuinely destructive and does not ask. Always build
   the path from a known-safe root (a tempfile directory, or a path you
   validated), never from raw user input.
""")
````

## Pickle Module:
- It is a way to save python data objects & next time when we will need it ,we can simply load it up & get back data objects. This process known as object serialization.
- **pickle.dump()** is the method for saving the data to the pickle file with .p & .pkl extension.
- **pickle.load()** is the method for loading pickle file.

````python
import pickle
print(pickle.__doc__)
````

````python
import pickle, tempfile
from pathlib import Path

pkl_dir = Path(tempfile.mkdtemp(prefix="pickle_demo_"))
pkl_path = pkl_dir / "test.pkl"

# ---- Save a pickle file ----
my_dict = {"name": "ravi", "age": 24, 1: "app"}
with open(pkl_path, "wb") as f:
    pickle.dump(my_dict, f)

print("written:", pkl_path.name, f"({pkl_path.stat().st_size} bytes)")
````

````python
# ---- Load a pickle file ----
with open(pkl_path, "rb") as f:
    ndict = pickle.load(f)

print(ndict)
print("round trip identical:", ndict == my_dict)

# ---- Clean up ----
import shutil
shutil.rmtree(pkl_dir)
print("cleaned up:", not pkl_dir.exists())

print("""
⚠️ SECURITY: never unpickle data you did not create.
   pickle.load() can execute arbitrary code during deserialisation, so an
   untrusted .pkl file is equivalent to running an untrusted script.
   For data exchange use JSON (08) or another safe format.
""")
````

## Common Mistakes & Pitfalls

1. 🔴 **Hardcoding an absolute path** like `r'C:\\Users\\Aditya\\...'`. It works on exactly one machine. Derive paths from `__file__`, `Path.cwd()` or a config value.
2. **Naming your file after a stdlib module.** A local `random.py` or `json.py` shadows the real one, and the error message is baffling.
3. **`from module import *`.** It hides where names came from and can silently overwrite your own. Only acceptable in an interactive session.
4. **Forgetting `if __name__ == '__main__':`.** Any top-level code runs on *import*, so importing your script executes it.
5. **Assuming a module re-executes on every import.** It is cached in `sys.modules` and its top-level code runs **once** per process.
6. **Mutating `sys.path` at run time** to make an import work. It is a smell — the real fix is a package plus an editable install (**7.2**).
7. **Using `imp`.** It was **removed in Python 3.12**. Use `importlib`.
8. **Circular imports.** Two modules importing each other gives `ImportError: cannot import name ... (most likely due to a circular import)`.
9. **Reaching for `os.path` in new code.** `pathlib` is clearer and cross-platform by design.

## Best Practices

- One module = one coherent responsibility. Split when the file stops fitting in your head.
- Put imports at the top, grouped **stdlib / third-party / local**, per PEP 8.
- Prefer `from package import module` over `from module import *`.
- Guard executable code with `if __name__ == '__main__':`.
- Use **`pathlib`** for filesystem work in new code.
- Give every module a docstring — `help(module)` and `module.__doc__` read it.
- Use `__all__` to declare a module's public surface.
- Never hardcode absolute paths; build them from `Path(__file__).parent` or configuration.
