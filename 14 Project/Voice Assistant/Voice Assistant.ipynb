```python
# Importing libraries
from speech_recognition import Recognizer, Microphone, UnknownValueError, RequestError
from re import search
import os
from sys import exit
import webbrowser
import smtplib
import subprocess
from datetime import datetime as dt
from time import sleep, strftime
from pyowm import OWM
import vlc
from bs4 import BeautifulSoup as soup
import urllib
import requests
import json
import youtube_dl
import wikipedia
import random

import warnings
warnings.simplefilter("ignore")
```

- create the method which will interpret user voice response.

```python
def myCommand():
    r = Recognizer()
    with Microphone() as source:
        print('Say something...')
        r.pause_threshold = 1
        r.adjust_for_ambient_noise(source, duration=1)
        audio = r.record(source, duration=5)
#         audio = r.listen(source)
    try:
        print('Analysing...')
        command = r.recognize_google(audio).lower()
        print('You said: ' + command + '\n')
    except UnknownValueError:
        print("Google Speech Recognition could not understand audio")
        # loop back to continue to listen for commands if unrecognizable speech is received
        command = myCommand()
    return command
```

- create a method that will convert text to speech.

```python
def sofiaResponse(audio):
    print(audio)
    for line in audio.splitlines():
        os.system("say " + audio)
```

- create a loop to continue executing multiple commands. 
- Inside the method assistant() passing user command(myCommand()) as parameters.

### 1. Open the subreddit Reddit in the browser.
- The user will give any command to open any subreddit from Reddit and the command should be 
    - *Hey Sofia!* Can you please **open Reddit subreddit_name**. 
- Only the italic bold phrase should be used as it is. You can use any kind of prefix, just take care of the italic bold phrase.
<br/><br/>
- **How it works:** If you have said the phrase *open reddit* in your command then it will search for subreddit name in the user command using re.search(). The subreddit will be searched using www.reddit.com and will get opened in the browser using pythons Webbrowser module.
- The webbrowser module provides a high-level interface to allow displaying Web-based documents to users.

```python
def assistant(command):
    # shutdown
    if 'exit' in command:
        sofiaResponse('Bye bye Sir. Have a nice day')
        exit()
        
    #reddit
    elif 'open reddit' in command:
        reg_ex = search('open reddit (.*)', command)
        url = 'https://www.reddit.com/'
        if reg_ex:
            subreddit = reg_ex.group(1)
            url = url + 'r/' + subreddit
        webbrowser.open(url)
        sofiaResponse('The Reddit content has been opened for you Sir.')
        

while True:
    assistant(myCommand())
    sleep(5)
```

### 2. Open any website in the browser.

### 3. Send Email.
- You can also ask your desktop assistant to send the email.
- **How it works:** If you have said the word email in your command then the bot will ask for receipient, If my response is rajat, the bot will use pthons smtplib library. The smtplib module defines an SMTP client session object that can be used to send mail to any Internet machine with an SMTP or ESMTP listener daemon. Sending mail is done with Python’s smtplib using an SMTP server. First it will initaite gmail SMTP using smtplib.SMTP(), then identify the server using ehlo() function, then encypting the session starttls(), then login to your mailbox using login(), then sending the message using sendmail().

```python
import os          # credentials come from the environment
def assistant(command):
    # shutdown
    if 'exit' in command:
        sofiaResponse('Bye bye Sir. Have a nice day')
        exit()
    
    # email
    elif 'email' in command:
        contact={'alice': 'alice@example.com', 'bob': 'bob@example.com'}
        sofiaResponse('Who is the recipient?')
        recipient = myCommand()
        if recipient in contact.keys():
            sofiaResponse('What should I say to him/her?')
            content = myCommand()
            server = smtplib.SMTP('smtp.gmail.com', 587)
            server.ehlo()
            server.starttls()
            server.login(os.environ["MAIL_ADDRESS"], os.environ["MAIL_PASSWORD"])
            server.sendmail(os.environ["MAIL_ADDRESS"], f'{contact.get(recipient)}', content)
            server.close()
            sofiaResponse('Email has been sent successfuly.')
        else:
            sofiaResponse('Person not in contact list')
    
while True:
    assistant(myCommand())
```

### 4. Launch any system application.
- Say “launch calc” or “can you please launch notepad” or “Sofia launch writer” etc. and Sofia will launch that system application for you.
- **How it works:** If you have said the word launch in your command then it will search for application name(if it is present in your system) in the user command using re.search(). It will then append the suffix “.app” to the application name. Now your application name is for example say calender.app(In macOS the executable files end with extension .app unlike in Windows which ends with .exe). So the executable application name will be launched using python subprocess’s Popen() function. The subprocess module enables you to start new applications from your Python program.

```python
def assistant(command):
    # shutdown
    if 'exit' in command:
        sofiaResponse('Bye bye Sir. Have a nice day')
        exit()
    
    # launch application
    elif 'launch' in command:
        reg_ex = search('launch (.*)', command)
        if reg_ex:
            appname = reg_ex.group(1)
            appname1 = appname+".exe"
            try:
                subprocess.Popen('C:\\Windows\\System32\\'+ appname1)
                sofiaResponse('I have launched the desired application')
            except:
                sofiaResponse("Sorry didn't Understood!")

while True:
    assistant(myCommand())
```

### 5. Tells you the current weather and temperature of almost any city.
- Sofia can also tell you the weather, maximum and minimum temperature of any city around the world. The user just needs to say something like “what is the current weather in London” or “tell me the current weather in Delhi”.
- **How it works:** If you have said the phrase current weather in your command then it will search for city name using re.search(). I have used pythons pyowm library to get the weather of any city. get_status() will tell you about the weather condition like haze, cloudy, rainy etc and get_temperature() will tell you about the max and min temperature of the city.

```python
import os          # credentials come from the environment
def assistant(command):
    # shutdown
    if 'exit' in command:
        sofiaResponse('Bye bye Sir. Have a nice day')
        exit()
        
    # weather    
    elif 'current weather' in command:
        reg_ex = search('current weather in (.*)', command)
        try:
            if reg_ex:
                city = reg_ex.group(1)
                owm = OWM(API_key=os.environ["OWM_API_KEY"])
                obs = owm.weather_at_place(city)
                w = obs.get_weather()
                k = w.get_status()
                x = w.get_temperature(unit='celsius')
                sofiaResponse('''Current weather in %s is %s.
                The maximum temperature is %0.2f degree celcius.
                The minimum temperature is %0.2f degree celcius.''' % (city, k, x['temp_max'], x['temp_min']))
        except:
            sofiaResponse('Sorry! Didn\'t understood your request.')

while True:
    assistant(myCommand())
```

And when you give command as shutdown, sys.exit() will be called to terminate the program.

```python
def assistant(command):
    # to terminate the program
    if 'exit' in command:
        sofiaResponse('Bye bye Sir. Have a nice day')
        exit()
        
    #Greet Sofia
    elif 'hello' in command:
        day_time = int(strftime('%H'))
        if day_time < 12:
            sofiaResponse('Hello Sir. Good morning')
        elif 12 <= day_time < 18:
            sofiaResponse('Hello Sir. Good afternoon')
        else:
            sofiaResponse('Hello Sir. Good evening')
    
while True:
    assistant(myCommand())
```
